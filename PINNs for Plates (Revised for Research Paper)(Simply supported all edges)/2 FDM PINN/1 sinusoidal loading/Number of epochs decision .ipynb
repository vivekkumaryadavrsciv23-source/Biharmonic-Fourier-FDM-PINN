# FDM-PINN Kirchhoff-Love plate bending under SINUSOIDAL loading 

## all-simply-supported (SS-SS-SS-SS)

# SiLU Activation function

### Epochs= 5000

```python
# ---------------------------------------------------------------------------
# Physics-Informed Neural Network for Kirchhoff Plate Bending
# Architecture : FDM-PINN
# BC Type      : SSSS (Simply Supported: w=0 and Mn=0 on all edges)
# Loading      : N/A - Epoch selection study
# Authors      : Vivek Kumar Yadav, Kshitij Kumar Yadav, Rajesh Kumar
# Institution  : Dept. of Civil Engineering, IIT (BHU) Varanasi
# ---------------------------------------------------------------------------


import os
import json
import time
import random

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from matplotlib import cm

# ============================================================
# CONFIG
# ============================================================
MODEL_NAME = "FDM_PINN"
LOADING_CASE = "sinusoidal"

OUT_DIR = os.path.join("outputs", MODEL_NAME)
FIG_DIR = os.path.join(OUT_DIR, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

SEED = 42
MAX_EPOCHS = 5000     # matches original notebook's training budget (range(15000)); same PATIENCE=3000
PATIENCE = 3000
PRINT_EVERY = 100
LR_INIT = 1e-3
LR_DECAY_STEPS = 2000
LR_DECAY_RATE = 0.95

FD_STEP = 0.01     # finite-difference step size h (non-dimensional units).

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
DTYPE = torch.float64

# ============================================================
# SEEDING
# ============================================================
np.random.seed(SEED)
random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)

# ============================================================
# ============================================================
real_a, real_b, real_h = 1.0, 1.0, 0.01
real_E, real_nu, real_q0 = 200e9, 0.3, 1000.0

D_dim = real_E * (real_h ** 3) / (12 * (1 - real_nu ** 2))
L = real_a
W = real_q0 * (L ** 4) / D_dim
D_nd = 1.0
a_nd, b_nd, q0_nd, nu_nd = 1.0, 1.0, 1.0, real_nu

# ============================================================
# DOMAIN
# ============================================================
nx, ny = 51, 51
x_nd = np.linspace(0, a_nd, nx)
y_nd = np.linspace(0, b_nd, ny)
X_nd, Y_nd = np.meshgrid(x_nd, y_nd)
grid_points_nd = np.vstack([X_nd.ravel(), Y_nd.ravel()]).T

X_grid = torch.tensor(grid_points_nd, dtype=DTYPE, device=DEVICE)

center_x, center_y = a_nd / 2, b_nd / 2
center_idx = int(np.argmin((grid_points_nd[:, 0] - center_x) ** 2 +
                            (grid_points_nd[:, 1] - center_y) ** 2))

boundary_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
interior_indices = np.setdiff1d(np.arange(nx * ny), boundary_indices)

edge_x_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
edge_y_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
]))

bc_idx = torch.tensor(boundary_indices, dtype=torch.long, device=DEVICE)
int_idx = torch.tensor(interior_indices, dtype=torch.long, device=DEVICE)
edge_x_idx = torch.tensor(edge_x_indices, dtype=torch.long, device=DEVICE)
edge_y_idx = torch.tensor(edge_y_indices, dtype=torch.long, device=DEVICE)

X_int = X_grid[int_idx]        # (N_int, 2) interior collocation points
X_bc = X_grid[bc_idx]          # (N_bc, 2) boundary points (for w=0)
X_edge_x = X_grid[edge_x_idx]  # (N_ex, 2) x=0,a edge points (for Mx=0)
X_edge_y = X_grid[edge_y_idx]  # (N_ey, 2) y=0,b edge points (for My=0)

# ============================================================
# ============================================================
def analytical_w_nd(x, y):
    denom = np.pi ** 4 * D_nd * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mx_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (1 / a_nd ** 2 + nu_nd / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_My_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (nu_nd / a_nd ** 2 + 1 / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mxy_nd(x, y):
    denom = ((1 / a_nd ** 2) + (1 / b_nd ** 2)) ** 2
    return (-q0_nd * (1 - nu_nd) / (np.pi ** 2 * denom * a_nd * b_nd)) * np.cos(np.pi * x / a_nd) * np.cos(np.pi * y / b_nd)


def sinusoidal_load_nd(x, y):
    return q0_nd * torch.sin(np.pi * x / a_nd) * torch.sin(np.pi * y / b_nd)


# ============================================================
# ============================================================
class PlatePINN(nn.Module):
    """
    w(x,y) = D(x,y) * scale * NN(x,y), where D(x,y) = x(a-x)y(b-y) / [(a/2)^2(b/2)^2]
    is a smooth distance function that is EXACTLY zero on all four edges
    (x=0, x=a, y=0, y=b) and positive inside the domain.

    This hard-enforces the essential SS boundary condition w=0 by construction,
    for ANY choice of network weights -- the optimizer can no longer trade off
    boundary satisfaction against the PDE/moment residuals, which was causing
    a visible, non-shrinking violation of w=0 at the edges (and a systematic
    bias in the interior solution) under the previous soft bc_loss penalty.
    Normalizing by (a/2)^2*(b/2)^2 keeps D(x,y)=1 at the plate center, so it
    doesn't change the overall output scale.
    """
    def __init__(self, hidden=64, n_layers=3):
        super().__init__()
        layers = []
        in_dim = 2
        for _ in range(n_layers):
            linear = nn.Linear(in_dim, hidden, dtype=DTYPE)
            nn.init.xavier_normal_(linear.weight)   # matches original kernel_initializer='glorot_normal'
            nn.init.zeros_(linear.bias)              # matches keras Dense default bias_initializer='zeros'
            layers.append(linear)
            layers.append(nn.SiLU())
            in_dim = hidden
        self.hidden = nn.Sequential(*layers)
        self.out = nn.Linear(in_dim, 1, dtype=DTYPE)
        nn.init.zeros_(self.out.weight)   # matches original output_layer kernel_initializer='zeros'
        nn.init.zeros_(self.out.bias)     # network output is exactly 0 at init, controlled by `scale` below
        self.scale = nn.Parameter(torch.tensor(0.01, dtype=DTYPE))

    def forward(self, x, y):
        h = torch.cat([x / a_nd, y / b_nd], dim=1)
        h = self.hidden(h)
        n_out = self.out(h)
        dist = (x * (a_nd - x) * y * (b_nd - y)) / ((a_nd / 2) ** 2 * (b_nd / 2) ** 2)
        return self.scale * dist * n_out


model = PlatePINN().to(DEVICE)

optimizer = torch.optim.Adam(model.parameters(), lr=LR_INIT)
scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer, lr_lambda=lambda step: LR_DECAY_RATE ** (step / LR_DECAY_STEPS)
)


# ============================================================
# FINITE-DIFFERENCE STENCIL HELPERS
# ============================================================
def batched_stencil_eval(x, y, offsets):
    """
    Evaluate the model at (x+dx, y+dy) for every (dx,dy) in `offsets`,
    all in a single stacked forward pass.

    x, y : (N,1) tensors
    offsets : list of (dx, dy) tuples
    returns : dict mapping (dx,dy) -> (N,1) tensor of model outputs
    """
    n = x.shape[0]
    xs = torch.cat([x + dx for dx, dy in offsets], dim=0)
    ys = torch.cat([y + dy for dx, dy in offsets], dim=0)
    w_all = model(xs, ys)
    return {off: w_all[i * n:(i + 1) * n] for i, off in enumerate(offsets)}


def fd_interior_derivatives(x, y, h):
    """Return w_c, w_xxxx, w_yyyy, w_xxyy at interior points via FD stencils."""
    offsets = [
        (0, 0),
        (-2 * h, 0), (-h, 0), (h, 0), (2 * h, 0),      # x-line (for xxxx)
        (0, -2 * h), (0, -h), (0, h), (0, 2 * h),      # y-line (for yyyy)
        (h, h), (h, -h), (-h, h), (-h, -h),            # diagonals (for xxyy)
    ]
    v = batched_stencil_eval(x, y, offsets)

    w_c = v[(0, 0)]
    w_xxxx = (v[(-2 * h, 0)] - 4 * v[(-h, 0)] + 6 * w_c - 4 * v[(h, 0)] + v[(2 * h, 0)]) / h ** 4
    w_yyyy = (v[(0, -2 * h)] - 4 * v[(0, -h)] + 6 * w_c - 4 * v[(0, h)] + v[(0, 2 * h)]) / h ** 4
    w_xxyy = (4 * w_c
              - 2 * (v[(h, 0)] + v[(-h, 0)] + v[(0, h)] + v[(0, -h)])
              + v[(h, h)] + v[(h, -h)] + v[(-h, h)] + v[(-h, -h)]) / h ** 4
    return w_c, w_xxxx, w_yyyy, w_xxyy


def fd_boundary_moments(x, y, h):
    """Return w_c, w_xx, w_yy at boundary points via 3-point FD stencils.

    Note: the central stencil evaluates the network at x-h/x+h (or y-h/y+h),
    which for points exactly on the domain edge means one of the two
    evaluation points lies slightly outside [0,a]x[0,b] (e.g. at x=0 this
    needs w(-h,y)). This is not automatically wrong -- the network is a
    smooth function defined everywhere, not only on the physical domain --
    but it is a methodological choice a reviewer could question. It is
    documented here explicitly as the chosen FDM-PINN stencil treatment for
    this study: central differences, evaluated with a small extension of the
    network beyond the physical edge. A one-sided (forward/backward)
    boundary stencil is a valid alternative if strict domain-only evaluation
    is required; it was not used here so that all boundary derivatives share
    the same (central, 2nd-order accurate) truncation-error behavior as the
    interior stencils."""
    offsets = [(0, 0), (-h, 0), (h, 0), (0, -h), (0, h)]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    return w_c, w_xx, w_yy


def fd_full_field_derivatives(x, y, h):
    """Same as fd_interior_derivatives but also returns w_xy (for Mxy) —
    used only in post-processing over the full grid."""
    offsets = [
        (0, 0),
        (-h, 0), (h, 0), (0, -h), (0, h),
        (h, h), (h, -h), (-h, h), (-h, -h),
    ]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    w_xy = (v[(h, h)] - v[(h, -h)] - v[(-h, h)] + v[(-h, -h)]) / (4 * h ** 2)
    return w_c, w_xx, w_yy, w_xy


# ============================================================
# LOSS (finite-difference residual — replaces autograd chain from Vanilla)
# ============================================================
def compute_loss():
    x_int, y_int = X_int[:, 0:1], X_int[:, 1:2]
    x_bc, y_bc = X_bc[:, 0:1], X_bc[:, 1:2]
    x_ex, y_ex = X_edge_x[:, 0:1], X_edge_x[:, 1:2]
    x_ey, y_ey = X_edge_y[:, 0:1], X_edge_y[:, 1:2]

    _, w_xxxx, w_yyyy, w_xxyy = fd_interior_derivatives(x_int, y_int, FD_STEP)
    q = sinusoidal_load_nd(x_int, y_int)
    pde = w_xxxx + 2 * w_xxyy + w_yyyy - q
    pde_loss = torch.mean(pde ** 2)

    w_bc = model(x_bc, y_bc)
    bc_loss = torch.mean(w_bc ** 2)

    _, w_xx_ex, w_yy_ex = fd_boundary_moments(x_ex, y_ex, FD_STEP)
    Mx_edge = -(w_xx_ex + nu_nd * w_yy_ex)

    _, w_xx_ey, w_yy_ey = fd_boundary_moments(x_ey, y_ey, FD_STEP)
    My_edge = -(w_yy_ey + nu_nd * w_xx_ey)

    moment_loss = torch.mean(Mx_edge ** 2) + torch.mean(My_edge ** 2)

    total = pde_loss + moment_loss

    with torch.no_grad():
        w_full = model(X_grid[:, 0:1], X_grid[:, 1:2])

    return total, pde_loss, bc_loss, moment_loss, w_full


# ============================================================
# ============================================================
epoch_history, loss_history, pde_history, bc_history, moment_history = [], [], [], [], []
w_center_history, epoch_time_history = [], []

best_loss = float("inf")
wait = 0
weights_path = os.path.join(OUT_DIR, "best_weights.pt")

if DEVICE.type == "cuda":
    torch.cuda.reset_peak_memory_stats()

train_start = time.perf_counter()
early_stop_epoch = MAX_EPOCHS  # epoch index where early stopping triggered; MAX_EPOCHS if it never did. This is an early-stopping bookkeeping value, NOT a proven mathematical convergence criterion.

for epoch in range(MAX_EPOCHS):
    epoch_t0 = time.perf_counter()

    optimizer.zero_grad()
    total, pde, bc, moment, w_full = compute_loss()
    total.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)  # matches original Adam(clipnorm=1.0)
    optimizer.step()
    scheduler.step()

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    epoch_dt = time.perf_counter() - epoch_t0

    loss_val = total.item()
    w_center_val = w_full[center_idx, 0].item()

    epoch_history.append(epoch)
    loss_history.append(loss_val)
    pde_history.append(pde.item())
    bc_history.append(bc.item())
    moment_history.append(moment.item())
    w_center_history.append(w_center_val)
    epoch_time_history.append(epoch_dt)

    if epoch % PRINT_EVERY == 0:
        print(f"Epoch {epoch} Loss {loss_val:.6e} Center {w_center_val:.6e} "
              f"({epoch_dt*1000:.2f} ms/epoch)")

    if loss_val < best_loss:
        best_loss = loss_val
        wait = 0
        torch.save(model.state_dict(), weights_path)
    else:
        wait += 1
        if wait >= PATIENCE:
            print(f"Early stopping triggered at epoch {epoch}")
            early_stop_epoch = epoch
            break

total_train_time = time.perf_counter() - train_start
peak_mem_mb = (torch.cuda.max_memory_allocated() / 1e6) if DEVICE.type == "cuda" else float("nan")

model.load_state_dict(torch.load(weights_path, map_location=DEVICE))
model.eval()

# ============================================================
# SAVE TRAINING CONVERGENCE CSV
# ============================================================
w_center_dim_history = np.array(w_center_history) * W
df_hist = pd.DataFrame({
    "epoch": epoch_history,
    "total_loss": loss_history,
    "pde_loss": pde_history,
    "bc_loss": bc_history,
    "moment_loss": moment_history,
    "w_center_nd": w_center_history,
    "w_center_dim": w_center_dim_history,
    "epoch_time_s": epoch_time_history,
})
df_hist.to_csv(os.path.join(OUT_DIR, "training_convergence.csv"), index=False)
print("Saved training_convergence.csv")

# ============================================================
# ============================================================
cost_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "device": str(DEVICE),
    "fd_step_h": FD_STEP,
    "epochs_run": len(epoch_history),
    "epochs_to_early_stop": early_stop_epoch,  # renamed from epochs_to_convergence: this marks when the
    "mean_time_per_epoch_s": float(np.mean(epoch_time_history)),
    "median_time_per_epoch_s": float(np.median(epoch_time_history)),
    "total_training_time_s": total_train_time,
    "peak_gpu_memory_mb": peak_mem_mb,
    "n_trainable_params": sum(p.numel() for p in model.parameters() if p.requires_grad),
}
with open(os.path.join(OUT_DIR, "cost_metrics.json"), "w") as f:
    json.dump(cost_metrics, f, indent=2)
print("Saved cost_metrics.json:", cost_metrics)

# ============================================================
# POST-PROCESSING: full-field predictions via FD stencils
# ============================================================
def compute_full_fields():
    x = X_grid[:, 0:1]
    y = X_grid[:, 1:2]
    with torch.no_grad():
        w_c, w_xx, w_yy, w_xy = fd_full_field_derivatives(x, y, FD_STEP)
        Mx = -(w_xx + nu_nd * w_yy)
        My = -(w_yy + nu_nd * w_xx)
        Mxy = -(1 - nu_nd) * w_xy

    w_grid = w_c.cpu().numpy().reshape(ny, nx)
    Mx_grid = Mx.cpu().numpy().reshape(ny, nx)
    My_grid = My.cpu().numpy().reshape(ny, nx)
    Mxy_grid = Mxy.cpu().numpy().reshape(ny, nx)
    return w_grid, Mx_grid, My_grid, Mxy_grid


w_pred_nd, Mx_pred_nd, My_pred_nd, Mxy_pred_nd = compute_full_fields()

w_anal_nd = analytical_w_nd(X_nd, Y_nd)
Mx_anal_nd = analytical_Mx_nd(X_nd, Y_nd)
My_anal_nd = analytical_My_nd(X_nd, Y_nd)
Mxy_anal_nd = analytical_Mxy_nd(X_nd, Y_nd)

w_dim_pred, w_dim_anal = w_pred_nd * W, w_anal_nd * W
moment_scale = D_dim * W / L ** 2
Mx_dim_pred, Mx_dim_anal = Mx_pred_nd * moment_scale, Mx_anal_nd * moment_scale
My_dim_pred, My_dim_anal = My_pred_nd * moment_scale, My_anal_nd * moment_scale
Mxy_dim_pred, Mxy_dim_anal = Mxy_pred_nd * moment_scale, Mxy_anal_nd * moment_scale

w_signed_error = w_dim_pred - w_dim_anal
w_error = np.abs(w_signed_error)
Mx_error = np.abs(Mx_dim_pred - Mx_dim_anal)
My_error = np.abs(My_dim_pred - My_dim_anal)
Mxy_error = np.abs(Mxy_dim_pred - Mxy_dim_anal)

# ============================================================
# ============================================================
iy, ix = ny // 2, nx // 2
w_pred_center, w_anal_center = w_dim_pred[iy, ix], w_dim_anal[iy, ix]
center_abs_err = abs(w_pred_center - w_anal_center)
center_pct_err = center_abs_err / abs(w_anal_center) * 100

L2_error_w = np.sqrt(np.mean((w_dim_pred - w_dim_anal) ** 2))
rel_L2_error_w = np.linalg.norm(w_dim_pred - w_dim_anal) / np.linalg.norm(w_dim_anal)

error_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "center_deflection_pred_m": float(w_pred_center),
    "center_deflection_anal_m": float(w_anal_center),
    "center_abs_error_m": float(center_abs_err),
    "center_pct_error": float(center_pct_err),
    "w_max_abs_error_m": float(np.max(w_error)),
    "w_rms_error_m": float(np.sqrt(np.mean(w_error ** 2))),
    "w_L2_error_m": float(L2_error_w),
    "w_relative_L2_error": float(rel_L2_error_w),
    "Mx_max_abs_error_Nm": float(np.max(Mx_error)),
    "Mx_rms_error_Nm": float(np.sqrt(np.mean(Mx_error ** 2))),
    "My_max_abs_error_Nm": float(np.max(My_error)),
    "My_rms_error_Nm": float(np.sqrt(np.mean(My_error ** 2))),
    "Mxy_max_abs_error_Nm": float(np.max(Mxy_error)),
    "Mxy_rms_error_Nm": float(np.sqrt(np.mean(Mxy_error ** 2))),
}
with open(os.path.join(OUT_DIR, "error_metrics.json"), "w") as f:
    json.dump(error_metrics, f, indent=2)
print("Saved error_metrics.json:", error_metrics)

df_field = pd.DataFrame({
    "x_nd": X_nd.ravel(), "y_nd": Y_nd.ravel(),
    "w_dim_anal": w_dim_anal.ravel(), "w_dim_pred": w_dim_pred.ravel(),
    "Mx_dim_anal": Mx_dim_anal.ravel(), "Mx_dim_pred": Mx_dim_pred.ravel(),
    "My_dim_anal": My_dim_anal.ravel(), "My_dim_pred": My_dim_pred.ravel(),
    "Mxy_dim_anal": Mxy_dim_anal.ravel(), "Mxy_dim_pred": Mxy_dim_pred.ravel(),
})
df_field["w_abs_error"] = np.abs(df_field["w_dim_anal"] - df_field["w_dim_pred"])
df_field["w_rel_error"] = df_field["w_abs_error"] / (np.abs(df_field["w_dim_anal"]) + 1e-12)
df_field.to_csv(os.path.join(OUT_DIR, "field_data.csv"), index=False)
print("Saved field_data.csv")

# ============================================================
# ============================================================
plt.rcParams.update({"figure.dpi": 150})


def save_contour(field, title, cbar_label, fname, cmap=cm.jet):
    plt.figure(figsize=(7, 6))
    cf = plt.contourf(X_nd, Y_nd, field, levels=50, cmap=cmap)
    plt.colorbar(cf, label=cbar_label)
    plt.title(title, fontstyle="italic", fontsize=13)
    plt.xlabel("x/a", fontstyle="italic")
    plt.ylabel("y/b", fontstyle="italic")
    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, fname))
    plt.close()


plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], df_hist["w_center_dim"], linewidth=2, label=f"{MODEL_NAME} Prediction")
plt.axhline(y=w_anal_center, linestyle="--", linewidth=2, color="k", label="Analytical")
plt.xlabel("Epoch"); plt.ylabel("Center Deflection w (m)")
plt.title("Convergence of Center Deflection")
plt.legend(); plt.grid(True, linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Center_Deflection.png")); plt.close()

plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], np.abs(df_hist["w_center_dim"] - w_anal_center), linewidth=2)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Absolute Error at Center (m)")
plt.title("Convergence Error (Log Scale)")
plt.grid(True, which="both", linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Error_Log_Scale.png")); plt.close()

plt.figure(figsize=(8, 6))
plt.plot(df_hist["epoch"], df_hist["total_loss"], label="Total Loss")
plt.plot(df_hist["epoch"], df_hist["pde_loss"], label="PDE Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["bc_loss"], label="BC Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["moment_loss"], label="Moment Loss", alpha=0.7)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Training Loss Curve")
plt.legend(); plt.grid(True, which="both", ls="--", alpha=0.7); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Training_Loss_Curve.png")); plt.close()

save_contour(w_dim_anal, "Analytical Plate Deflection", "Deflection (m)",
             "Analytical_Plate_Deflection.png")
save_contour(w_dim_pred, f"{MODEL_NAME} Predicted Plate Deflection", "Deflection (m)",
             "FDM-PINN_Predicted_Plate_Deflection.png")
save_contour(w_error, "Absolute Error in Plate Deflection", "|w_pred - w_anal| (m)",
             "Absolute_Error_in_Plate_Deflection.png", cmap=cm.inferno)
save_contour(w_signed_error, "Signed Error w_pred - w_anal", "w_pred - w_anal (m)",
             "Absolute_Error_w_pred_-_w_anal.png", cmap="coolwarm")

save_contour(Mx_dim_anal, "Analytical Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x_Analytical.png")
save_contour(Mx_dim_pred, "Predicted Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x.png")
save_contour(Mx_error, "Absolute Error in $M_x$", "|Mx_pred - Mx_anal| (N·m/m)",
             "Absolute_Error_in_M_x.png", cmap="hot")

save_contour(My_dim_anal, "Analytical Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y_Analytical.png")
save_contour(My_dim_pred, "Predicted Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y.png")
save_contour(My_error, "Absolute Error in $M_y$", "|My_pred - My_anal| (N·m/m)",
             "Absolute_Error_in_M_y.png", cmap="hot")

save_contour(Mxy_dim_anal, "Analytical Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy_Analytical.png")
save_contour(Mxy_dim_pred, "Predicted Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy.png")
save_contour(Mxy_error, "Absolute Error in $M_{xy}$", "|Mxy_pred - Mxy_anal| (N·m/m)",
             "Absolute_Error_in_M_xy.png", cmap="hot")

plt.figure(figsize=(6, 6))
plt.scatter(w_dim_anal.ravel(), w_dim_pred.ravel(), s=8, alpha=0.5)
lims = [min(w_dim_anal.min(), w_dim_pred.min()), max(w_dim_anal.max(), w_dim_pred.max())]
plt.plot(lims, lims, "k--", linewidth=1, label="Perfect agreement")
plt.xlabel("Analytical displacement w (m)", fontstyle="italic")
plt.ylabel("Predicted displacement w (m)", fontstyle="italic")
plt.title("Analytical vs Predicted Displacement", fontstyle="italic")
plt.legend(); plt.grid(True); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Analytical_vs_Predicted_Displacement.png")); plt.close()

err_fft = np.fft.fftshift(np.fft.fft2(w_error))
err_fft_mag = np.log1p(np.abs(err_fft))
plt.figure(figsize=(7, 6))
plt.imshow(err_fft_mag, cmap="magma", extent=[-nx // 2, nx // 2, -ny // 2, ny // 2])
plt.colorbar(label="log(1 + |FFT error|)")
plt.title(f"Error Spectrum (FFT) — {LOADING_CASE}")
plt.xlabel("kx"); plt.ylabel("ky")
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Error_Spectrum_FFT.png")); plt.close()

fig, axes = plt.subplots(4, 3, figsize=(16, 18))
rows = [
    ("Displacement w (m)", w_dim_anal, w_dim_pred, w_error),
    ("Moment Mx (N·m)", Mx_dim_anal, Mx_dim_pred, Mx_error),
    ("Moment My (N·m)", My_dim_anal, My_dim_pred, My_error),
    ("Moment Mxy (N·m)", Mxy_dim_anal, Mxy_dim_pred, Mxy_error),
]
for r, (label, anal, pred, err) in enumerate(rows):
    for c, (title, field, cmap) in enumerate([
        (f"Analytical {label}", anal, cm.jet),
        (f"Predicted {label}", pred, cm.jet),
        (f"Absolute Error {label}", err, cm.inferno),
    ]):
        ax = axes[r, c]
        cf = ax.contourf(X_nd, Y_nd, field, levels=20, cmap=cmap)
        fig.colorbar(cf, ax=ax)
        ax.set_title(title, fontsize=10)
        ax.set_xlabel("x/a"); ax.set_ylabel("y/b")
plt.suptitle(f"{MODEL_NAME} — {LOADING_CASE} loading, SS-SS-SS-SS (summary grid)", fontsize=14)
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "_summary_grid_all_fields.png")); plt.close()

print(f"\nAll outputs written to: {os.path.abspath(OUT_DIR)}")
print(f"Total training time: {total_train_time:.2f}s | "
      f"Epochs to early stop: {early_stop_epoch} | "
      f"FD step h: {FD_STEP}")
```

```python
# ---------------------------------------------------------------------------
# Physics-Informed Neural Network for Kirchhoff Plate Bending
# Architecture : FDM-PINN
# BC Type      : SSSS (Simply Supported: w=0 and Mn=0 on all edges)
# Loading      : N/A - Epoch selection study
# Authors      : Vivek Kumar Yadav, Kshitij Kumar Yadav, Rajesh Kumar
# Institution  : Dept. of Civil Engineering, IIT (BHU) Varanasi
# ---------------------------------------------------------------------------


import os
import json
import time
import random

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from matplotlib import cm

# ============================================================
# CONFIG
# ============================================================
MODEL_NAME = "FDM_PINN"
LOADING_CASE = "sinusoidal"

OUT_DIR = os.path.join("outputs", MODEL_NAME)
FIG_DIR = os.path.join(OUT_DIR, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

SEED = 42
MAX_EPOCHS = 10000     # matches original notebook's training budget (range(15000)); same PATIENCE=3000
PATIENCE = 3000
PRINT_EVERY = 100
LR_INIT = 1e-3
LR_DECAY_STEPS = 2000
LR_DECAY_RATE = 0.95

FD_STEP = 0.01     # finite-difference step size h (non-dimensional units).

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
DTYPE = torch.float64

# ============================================================
# SEEDING
# ============================================================
np.random.seed(SEED)
random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)

# ============================================================
# ============================================================
real_a, real_b, real_h = 1.0, 1.0, 0.01
real_E, real_nu, real_q0 = 200e9, 0.3, 1000.0

D_dim = real_E * (real_h ** 3) / (12 * (1 - real_nu ** 2))
L = real_a
W = real_q0 * (L ** 4) / D_dim
D_nd = 1.0
a_nd, b_nd, q0_nd, nu_nd = 1.0, 1.0, 1.0, real_nu

# ============================================================
# DOMAIN
# ============================================================
nx, ny = 51, 51
x_nd = np.linspace(0, a_nd, nx)
y_nd = np.linspace(0, b_nd, ny)
X_nd, Y_nd = np.meshgrid(x_nd, y_nd)
grid_points_nd = np.vstack([X_nd.ravel(), Y_nd.ravel()]).T

X_grid = torch.tensor(grid_points_nd, dtype=DTYPE, device=DEVICE)

center_x, center_y = a_nd / 2, b_nd / 2
center_idx = int(np.argmin((grid_points_nd[:, 0] - center_x) ** 2 +
                            (grid_points_nd[:, 1] - center_y) ** 2))

boundary_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
interior_indices = np.setdiff1d(np.arange(nx * ny), boundary_indices)

edge_x_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
edge_y_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
]))

bc_idx = torch.tensor(boundary_indices, dtype=torch.long, device=DEVICE)
int_idx = torch.tensor(interior_indices, dtype=torch.long, device=DEVICE)
edge_x_idx = torch.tensor(edge_x_indices, dtype=torch.long, device=DEVICE)
edge_y_idx = torch.tensor(edge_y_indices, dtype=torch.long, device=DEVICE)

X_int = X_grid[int_idx]        # (N_int, 2) interior collocation points
X_bc = X_grid[bc_idx]          # (N_bc, 2) boundary points (for w=0)
X_edge_x = X_grid[edge_x_idx]  # (N_ex, 2) x=0,a edge points (for Mx=0)
X_edge_y = X_grid[edge_y_idx]  # (N_ey, 2) y=0,b edge points (for My=0)

# ============================================================
# ============================================================
def analytical_w_nd(x, y):
    denom = np.pi ** 4 * D_nd * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mx_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (1 / a_nd ** 2 + nu_nd / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_My_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (nu_nd / a_nd ** 2 + 1 / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mxy_nd(x, y):
    denom = ((1 / a_nd ** 2) + (1 / b_nd ** 2)) ** 2
    return (-q0_nd * (1 - nu_nd) / (np.pi ** 2 * denom * a_nd * b_nd)) * np.cos(np.pi * x / a_nd) * np.cos(np.pi * y / b_nd)


def sinusoidal_load_nd(x, y):
    return q0_nd * torch.sin(np.pi * x / a_nd) * torch.sin(np.pi * y / b_nd)


# ============================================================
# ============================================================
class PlatePINN(nn.Module):
    """
    w(x,y) = D(x,y) * scale * NN(x,y), where D(x,y) = x(a-x)y(b-y) / [(a/2)^2(b/2)^2]
    is a smooth distance function that is EXACTLY zero on all four edges
    (x=0, x=a, y=0, y=b) and positive inside the domain.

    This hard-enforces the essential SS boundary condition w=0 by construction,
    for ANY choice of network weights -- the optimizer can no longer trade off
    boundary satisfaction against the PDE/moment residuals, which was causing
    a visible, non-shrinking violation of w=0 at the edges (and a systematic
    bias in the interior solution) under the previous soft bc_loss penalty.
    Normalizing by (a/2)^2*(b/2)^2 keeps D(x,y)=1 at the plate center, so it
    doesn't change the overall output scale.
    """
    def __init__(self, hidden=64, n_layers=3):
        super().__init__()
        layers = []
        in_dim = 2
        for _ in range(n_layers):
            linear = nn.Linear(in_dim, hidden, dtype=DTYPE)
            nn.init.xavier_normal_(linear.weight)   # matches original kernel_initializer='glorot_normal'
            nn.init.zeros_(linear.bias)              # matches keras Dense default bias_initializer='zeros'
            layers.append(linear)
            layers.append(nn.SiLU())
            in_dim = hidden
        self.hidden = nn.Sequential(*layers)
        self.out = nn.Linear(in_dim, 1, dtype=DTYPE)
        nn.init.zeros_(self.out.weight)   # matches original output_layer kernel_initializer='zeros'
        nn.init.zeros_(self.out.bias)     # network output is exactly 0 at init, controlled by `scale` below
        self.scale = nn.Parameter(torch.tensor(0.01, dtype=DTYPE))

    def forward(self, x, y):
        h = torch.cat([x / a_nd, y / b_nd], dim=1)
        h = self.hidden(h)
        n_out = self.out(h)
        dist = (x * (a_nd - x) * y * (b_nd - y)) / ((a_nd / 2) ** 2 * (b_nd / 2) ** 2)
        return self.scale * dist * n_out


model = PlatePINN().to(DEVICE)

optimizer = torch.optim.Adam(model.parameters(), lr=LR_INIT)
scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer, lr_lambda=lambda step: LR_DECAY_RATE ** (step / LR_DECAY_STEPS)
)


# ============================================================
# FINITE-DIFFERENCE STENCIL HELPERS
# ============================================================
def batched_stencil_eval(x, y, offsets):
    """
    Evaluate the model at (x+dx, y+dy) for every (dx,dy) in `offsets`,
    all in a single stacked forward pass.

    x, y : (N,1) tensors
    offsets : list of (dx, dy) tuples
    returns : dict mapping (dx,dy) -> (N,1) tensor of model outputs
    """
    n = x.shape[0]
    xs = torch.cat([x + dx for dx, dy in offsets], dim=0)
    ys = torch.cat([y + dy for dx, dy in offsets], dim=0)
    w_all = model(xs, ys)
    return {off: w_all[i * n:(i + 1) * n] for i, off in enumerate(offsets)}


def fd_interior_derivatives(x, y, h):
    """Return w_c, w_xxxx, w_yyyy, w_xxyy at interior points via FD stencils."""
    offsets = [
        (0, 0),
        (-2 * h, 0), (-h, 0), (h, 0), (2 * h, 0),      # x-line (for xxxx)
        (0, -2 * h), (0, -h), (0, h), (0, 2 * h),      # y-line (for yyyy)
        (h, h), (h, -h), (-h, h), (-h, -h),            # diagonals (for xxyy)
    ]
    v = batched_stencil_eval(x, y, offsets)

    w_c = v[(0, 0)]
    w_xxxx = (v[(-2 * h, 0)] - 4 * v[(-h, 0)] + 6 * w_c - 4 * v[(h, 0)] + v[(2 * h, 0)]) / h ** 4
    w_yyyy = (v[(0, -2 * h)] - 4 * v[(0, -h)] + 6 * w_c - 4 * v[(0, h)] + v[(0, 2 * h)]) / h ** 4
    w_xxyy = (4 * w_c
              - 2 * (v[(h, 0)] + v[(-h, 0)] + v[(0, h)] + v[(0, -h)])
              + v[(h, h)] + v[(h, -h)] + v[(-h, h)] + v[(-h, -h)]) / h ** 4
    return w_c, w_xxxx, w_yyyy, w_xxyy


def fd_boundary_moments(x, y, h):
    """Return w_c, w_xx, w_yy at boundary points via 3-point FD stencils.

    Note: the central stencil evaluates the network at x-h/x+h (or y-h/y+h),
    which for points exactly on the domain edge means one of the two
    evaluation points lies slightly outside [0,a]x[0,b] (e.g. at x=0 this
    needs w(-h,y)). This is not automatically wrong -- the network is a
    smooth function defined everywhere, not only on the physical domain --
    but it is a methodological choice a reviewer could question. It is
    documented here explicitly as the chosen FDM-PINN stencil treatment for
    this study: central differences, evaluated with a small extension of the
    network beyond the physical edge. A one-sided (forward/backward)
    boundary stencil is a valid alternative if strict domain-only evaluation
    is required; it was not used here so that all boundary derivatives share
    the same (central, 2nd-order accurate) truncation-error behavior as the
    interior stencils."""
    offsets = [(0, 0), (-h, 0), (h, 0), (0, -h), (0, h)]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    return w_c, w_xx, w_yy


def fd_full_field_derivatives(x, y, h):
    """Same as fd_interior_derivatives but also returns w_xy (for Mxy) —
    used only in post-processing over the full grid."""
    offsets = [
        (0, 0),
        (-h, 0), (h, 0), (0, -h), (0, h),
        (h, h), (h, -h), (-h, h), (-h, -h),
    ]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    w_xy = (v[(h, h)] - v[(h, -h)] - v[(-h, h)] + v[(-h, -h)]) / (4 * h ** 2)
    return w_c, w_xx, w_yy, w_xy


# ============================================================
# LOSS (finite-difference residual — replaces autograd chain from Vanilla)
# ============================================================
def compute_loss():
    x_int, y_int = X_int[:, 0:1], X_int[:, 1:2]
    x_bc, y_bc = X_bc[:, 0:1], X_bc[:, 1:2]
    x_ex, y_ex = X_edge_x[:, 0:1], X_edge_x[:, 1:2]
    x_ey, y_ey = X_edge_y[:, 0:1], X_edge_y[:, 1:2]

    _, w_xxxx, w_yyyy, w_xxyy = fd_interior_derivatives(x_int, y_int, FD_STEP)
    q = sinusoidal_load_nd(x_int, y_int)
    pde = w_xxxx + 2 * w_xxyy + w_yyyy - q
    pde_loss = torch.mean(pde ** 2)

    w_bc = model(x_bc, y_bc)
    bc_loss = torch.mean(w_bc ** 2)

    _, w_xx_ex, w_yy_ex = fd_boundary_moments(x_ex, y_ex, FD_STEP)
    Mx_edge = -(w_xx_ex + nu_nd * w_yy_ex)

    _, w_xx_ey, w_yy_ey = fd_boundary_moments(x_ey, y_ey, FD_STEP)
    My_edge = -(w_yy_ey + nu_nd * w_xx_ey)

    moment_loss = torch.mean(Mx_edge ** 2) + torch.mean(My_edge ** 2)

    total = pde_loss + moment_loss

    with torch.no_grad():
        w_full = model(X_grid[:, 0:1], X_grid[:, 1:2])

    return total, pde_loss, bc_loss, moment_loss, w_full


# ============================================================
# ============================================================
epoch_history, loss_history, pde_history, bc_history, moment_history = [], [], [], [], []
w_center_history, epoch_time_history = [], []

best_loss = float("inf")
wait = 0
weights_path = os.path.join(OUT_DIR, "best_weights.pt")

if DEVICE.type == "cuda":
    torch.cuda.reset_peak_memory_stats()

train_start = time.perf_counter()
early_stop_epoch = MAX_EPOCHS  # epoch index where early stopping triggered; MAX_EPOCHS if it never did. This is an early-stopping bookkeeping value, NOT a proven mathematical convergence criterion.

for epoch in range(MAX_EPOCHS):
    epoch_t0 = time.perf_counter()

    optimizer.zero_grad()
    total, pde, bc, moment, w_full = compute_loss()
    total.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)  # matches original Adam(clipnorm=1.0)
    optimizer.step()
    scheduler.step()

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    epoch_dt = time.perf_counter() - epoch_t0

    loss_val = total.item()
    w_center_val = w_full[center_idx, 0].item()

    epoch_history.append(epoch)
    loss_history.append(loss_val)
    pde_history.append(pde.item())
    bc_history.append(bc.item())
    moment_history.append(moment.item())
    w_center_history.append(w_center_val)
    epoch_time_history.append(epoch_dt)

    if epoch % PRINT_EVERY == 0:
        print(f"Epoch {epoch} Loss {loss_val:.6e} Center {w_center_val:.6e} "
              f"({epoch_dt*1000:.2f} ms/epoch)")

    if loss_val < best_loss:
        best_loss = loss_val
        wait = 0
        torch.save(model.state_dict(), weights_path)
    else:
        wait += 1
        if wait >= PATIENCE:
            print(f"Early stopping triggered at epoch {epoch}")
            early_stop_epoch = epoch
            break

total_train_time = time.perf_counter() - train_start
peak_mem_mb = (torch.cuda.max_memory_allocated() / 1e6) if DEVICE.type == "cuda" else float("nan")

model.load_state_dict(torch.load(weights_path, map_location=DEVICE))
model.eval()

# ============================================================
# SAVE TRAINING CONVERGENCE CSV
# ============================================================
w_center_dim_history = np.array(w_center_history) * W
df_hist = pd.DataFrame({
    "epoch": epoch_history,
    "total_loss": loss_history,
    "pde_loss": pde_history,
    "bc_loss": bc_history,
    "moment_loss": moment_history,
    "w_center_nd": w_center_history,
    "w_center_dim": w_center_dim_history,
    "epoch_time_s": epoch_time_history,
})
df_hist.to_csv(os.path.join(OUT_DIR, "training_convergence.csv"), index=False)
print("Saved training_convergence.csv")

# ============================================================
# ============================================================
cost_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "device": str(DEVICE),
    "fd_step_h": FD_STEP,
    "epochs_run": len(epoch_history),
    "epochs_to_early_stop": early_stop_epoch,  # renamed from epochs_to_convergence: this marks when the
    "mean_time_per_epoch_s": float(np.mean(epoch_time_history)),
    "median_time_per_epoch_s": float(np.median(epoch_time_history)),
    "total_training_time_s": total_train_time,
    "peak_gpu_memory_mb": peak_mem_mb,
    "n_trainable_params": sum(p.numel() for p in model.parameters() if p.requires_grad),
}
with open(os.path.join(OUT_DIR, "cost_metrics.json"), "w") as f:
    json.dump(cost_metrics, f, indent=2)
print("Saved cost_metrics.json:", cost_metrics)

# ============================================================
# POST-PROCESSING: full-field predictions via FD stencils
# ============================================================
def compute_full_fields():
    x = X_grid[:, 0:1]
    y = X_grid[:, 1:2]
    with torch.no_grad():
        w_c, w_xx, w_yy, w_xy = fd_full_field_derivatives(x, y, FD_STEP)
        Mx = -(w_xx + nu_nd * w_yy)
        My = -(w_yy + nu_nd * w_xx)
        Mxy = -(1 - nu_nd) * w_xy

    w_grid = w_c.cpu().numpy().reshape(ny, nx)
    Mx_grid = Mx.cpu().numpy().reshape(ny, nx)
    My_grid = My.cpu().numpy().reshape(ny, nx)
    Mxy_grid = Mxy.cpu().numpy().reshape(ny, nx)
    return w_grid, Mx_grid, My_grid, Mxy_grid


w_pred_nd, Mx_pred_nd, My_pred_nd, Mxy_pred_nd = compute_full_fields()

w_anal_nd = analytical_w_nd(X_nd, Y_nd)
Mx_anal_nd = analytical_Mx_nd(X_nd, Y_nd)
My_anal_nd = analytical_My_nd(X_nd, Y_nd)
Mxy_anal_nd = analytical_Mxy_nd(X_nd, Y_nd)

w_dim_pred, w_dim_anal = w_pred_nd * W, w_anal_nd * W
moment_scale = D_dim * W / L ** 2
Mx_dim_pred, Mx_dim_anal = Mx_pred_nd * moment_scale, Mx_anal_nd * moment_scale
My_dim_pred, My_dim_anal = My_pred_nd * moment_scale, My_anal_nd * moment_scale
Mxy_dim_pred, Mxy_dim_anal = Mxy_pred_nd * moment_scale, Mxy_anal_nd * moment_scale

w_signed_error = w_dim_pred - w_dim_anal
w_error = np.abs(w_signed_error)
Mx_error = np.abs(Mx_dim_pred - Mx_dim_anal)
My_error = np.abs(My_dim_pred - My_dim_anal)
Mxy_error = np.abs(Mxy_dim_pred - Mxy_dim_anal)

# ============================================================
# ============================================================
iy, ix = ny // 2, nx // 2
w_pred_center, w_anal_center = w_dim_pred[iy, ix], w_dim_anal[iy, ix]
center_abs_err = abs(w_pred_center - w_anal_center)
center_pct_err = center_abs_err / abs(w_anal_center) * 100

L2_error_w = np.sqrt(np.mean((w_dim_pred - w_dim_anal) ** 2))
rel_L2_error_w = np.linalg.norm(w_dim_pred - w_dim_anal) / np.linalg.norm(w_dim_anal)

error_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "center_deflection_pred_m": float(w_pred_center),
    "center_deflection_anal_m": float(w_anal_center),
    "center_abs_error_m": float(center_abs_err),
    "center_pct_error": float(center_pct_err),
    "w_max_abs_error_m": float(np.max(w_error)),
    "w_rms_error_m": float(np.sqrt(np.mean(w_error ** 2))),
    "w_L2_error_m": float(L2_error_w),
    "w_relative_L2_error": float(rel_L2_error_w),
    "Mx_max_abs_error_Nm": float(np.max(Mx_error)),
    "Mx_rms_error_Nm": float(np.sqrt(np.mean(Mx_error ** 2))),
    "My_max_abs_error_Nm": float(np.max(My_error)),
    "My_rms_error_Nm": float(np.sqrt(np.mean(My_error ** 2))),
    "Mxy_max_abs_error_Nm": float(np.max(Mxy_error)),
    "Mxy_rms_error_Nm": float(np.sqrt(np.mean(Mxy_error ** 2))),
}
with open(os.path.join(OUT_DIR, "error_metrics.json"), "w") as f:
    json.dump(error_metrics, f, indent=2)
print("Saved error_metrics.json:", error_metrics)

df_field = pd.DataFrame({
    "x_nd": X_nd.ravel(), "y_nd": Y_nd.ravel(),
    "w_dim_anal": w_dim_anal.ravel(), "w_dim_pred": w_dim_pred.ravel(),
    "Mx_dim_anal": Mx_dim_anal.ravel(), "Mx_dim_pred": Mx_dim_pred.ravel(),
    "My_dim_anal": My_dim_anal.ravel(), "My_dim_pred": My_dim_pred.ravel(),
    "Mxy_dim_anal": Mxy_dim_anal.ravel(), "Mxy_dim_pred": Mxy_dim_pred.ravel(),
})
df_field["w_abs_error"] = np.abs(df_field["w_dim_anal"] - df_field["w_dim_pred"])
df_field["w_rel_error"] = df_field["w_abs_error"] / (np.abs(df_field["w_dim_anal"]) + 1e-12)
df_field.to_csv(os.path.join(OUT_DIR, "field_data.csv"), index=False)
print("Saved field_data.csv")

# ============================================================
# ============================================================
plt.rcParams.update({"figure.dpi": 150})


def save_contour(field, title, cbar_label, fname, cmap=cm.jet):
    plt.figure(figsize=(7, 6))
    cf = plt.contourf(X_nd, Y_nd, field, levels=50, cmap=cmap)
    plt.colorbar(cf, label=cbar_label)
    plt.title(title, fontstyle="italic", fontsize=13)
    plt.xlabel("x/a", fontstyle="italic")
    plt.ylabel("y/b", fontstyle="italic")
    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, fname))
    plt.close()


plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], df_hist["w_center_dim"], linewidth=2, label=f"{MODEL_NAME} Prediction")
plt.axhline(y=w_anal_center, linestyle="--", linewidth=2, color="k", label="Analytical")
plt.xlabel("Epoch"); plt.ylabel("Center Deflection w (m)")
plt.title("Convergence of Center Deflection")
plt.legend(); plt.grid(True, linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Center_Deflection.png")); plt.close()

plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], np.abs(df_hist["w_center_dim"] - w_anal_center), linewidth=2)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Absolute Error at Center (m)")
plt.title("Convergence Error (Log Scale)")
plt.grid(True, which="both", linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Error_Log_Scale.png")); plt.close()

plt.figure(figsize=(8, 6))
plt.plot(df_hist["epoch"], df_hist["total_loss"], label="Total Loss")
plt.plot(df_hist["epoch"], df_hist["pde_loss"], label="PDE Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["bc_loss"], label="BC Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["moment_loss"], label="Moment Loss", alpha=0.7)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Training Loss Curve")
plt.legend(); plt.grid(True, which="both", ls="--", alpha=0.7); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Training_Loss_Curve.png")); plt.close()

save_contour(w_dim_anal, "Analytical Plate Deflection", "Deflection (m)",
             "Analytical_Plate_Deflection.png")
save_contour(w_dim_pred, f"{MODEL_NAME} Predicted Plate Deflection", "Deflection (m)",
             "FDM-PINN_Predicted_Plate_Deflection.png")
save_contour(w_error, "Absolute Error in Plate Deflection", "|w_pred - w_anal| (m)",
             "Absolute_Error_in_Plate_Deflection.png", cmap=cm.inferno)
save_contour(w_signed_error, "Signed Error w_pred - w_anal", "w_pred - w_anal (m)",
             "Absolute_Error_w_pred_-_w_anal.png", cmap="coolwarm")

save_contour(Mx_dim_anal, "Analytical Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x_Analytical.png")
save_contour(Mx_dim_pred, "Predicted Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x.png")
save_contour(Mx_error, "Absolute Error in $M_x$", "|Mx_pred - Mx_anal| (N·m/m)",
             "Absolute_Error_in_M_x.png", cmap="hot")

save_contour(My_dim_anal, "Analytical Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y_Analytical.png")
save_contour(My_dim_pred, "Predicted Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y.png")
save_contour(My_error, "Absolute Error in $M_y$", "|My_pred - My_anal| (N·m/m)",
             "Absolute_Error_in_M_y.png", cmap="hot")

save_contour(Mxy_dim_anal, "Analytical Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy_Analytical.png")
save_contour(Mxy_dim_pred, "Predicted Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy.png")
save_contour(Mxy_error, "Absolute Error in $M_{xy}$", "|Mxy_pred - Mxy_anal| (N·m/m)",
             "Absolute_Error_in_M_xy.png", cmap="hot")

plt.figure(figsize=(6, 6))
plt.scatter(w_dim_anal.ravel(), w_dim_pred.ravel(), s=8, alpha=0.5)
lims = [min(w_dim_anal.min(), w_dim_pred.min()), max(w_dim_anal.max(), w_dim_pred.max())]
plt.plot(lims, lims, "k--", linewidth=1, label="Perfect agreement")
plt.xlabel("Analytical displacement w (m)", fontstyle="italic")
plt.ylabel("Predicted displacement w (m)", fontstyle="italic")
plt.title("Analytical vs Predicted Displacement", fontstyle="italic")
plt.legend(); plt.grid(True); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Analytical_vs_Predicted_Displacement.png")); plt.close()

err_fft = np.fft.fftshift(np.fft.fft2(w_error))
err_fft_mag = np.log1p(np.abs(err_fft))
plt.figure(figsize=(7, 6))
plt.imshow(err_fft_mag, cmap="magma", extent=[-nx // 2, nx // 2, -ny // 2, ny // 2])
plt.colorbar(label="log(1 + |FFT error|)")
plt.title(f"Error Spectrum (FFT) — {LOADING_CASE}")
plt.xlabel("kx"); plt.ylabel("ky")
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Error_Spectrum_FFT.png")); plt.close()

fig, axes = plt.subplots(4, 3, figsize=(16, 18))
rows = [
    ("Displacement w (m)", w_dim_anal, w_dim_pred, w_error),
    ("Moment Mx (N·m)", Mx_dim_anal, Mx_dim_pred, Mx_error),
    ("Moment My (N·m)", My_dim_anal, My_dim_pred, My_error),
    ("Moment Mxy (N·m)", Mxy_dim_anal, Mxy_dim_pred, Mxy_error),
]
for r, (label, anal, pred, err) in enumerate(rows):
    for c, (title, field, cmap) in enumerate([
        (f"Analytical {label}", anal, cm.jet),
        (f"Predicted {label}", pred, cm.jet),
        (f"Absolute Error {label}", err, cm.inferno),
    ]):
        ax = axes[r, c]
        cf = ax.contourf(X_nd, Y_nd, field, levels=20, cmap=cmap)
        fig.colorbar(cf, ax=ax)
        ax.set_title(title, fontsize=10)
        ax.set_xlabel("x/a"); ax.set_ylabel("y/b")
plt.suptitle(f"{MODEL_NAME} — {LOADING_CASE} loading, SS-SS-SS-SS (summary grid)", fontsize=14)
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "_summary_grid_all_fields.png")); plt.close()

print(f"\nAll outputs written to: {os.path.abspath(OUT_DIR)}")
print(f"Total training time: {total_train_time:.2f}s | "
      f"Epochs to early stop: {early_stop_epoch} | "
      f"FD step h: {FD_STEP}")
```

```python
# ---------------------------------------------------------------------------
# Physics-Informed Neural Network for Kirchhoff Plate Bending
# Architecture : FDM-PINN
# BC Type      : SSSS (Simply Supported: w=0 and Mn=0 on all edges)
# Loading      : N/A - Epoch selection study
# Authors      : Vivek Kumar Yadav, Kshitij Kumar Yadav, Rajesh Kumar
# Institution  : Dept. of Civil Engineering, IIT (BHU) Varanasi
# ---------------------------------------------------------------------------


import os
import json
import time
import random

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from matplotlib import cm

# ============================================================
# CONFIG
# ============================================================
MODEL_NAME = "FDM_PINN"
LOADING_CASE = "sinusoidal"

OUT_DIR = os.path.join("outputs", MODEL_NAME)
FIG_DIR = os.path.join(OUT_DIR, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

SEED = 42
MAX_EPOCHS = 15000     # matches original notebook's training budget (range(15000)); same PATIENCE=3000
PATIENCE = 3000
PRINT_EVERY = 100
LR_INIT = 1e-3
LR_DECAY_STEPS = 2000
LR_DECAY_RATE = 0.95

FD_STEP = 0.01     # finite-difference step size h (non-dimensional units).

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
DTYPE = torch.float64

# ============================================================
# SEEDING
# ============================================================
np.random.seed(SEED)
random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)

# ============================================================
# ============================================================
real_a, real_b, real_h = 1.0, 1.0, 0.01
real_E, real_nu, real_q0 = 200e9, 0.3, 1000.0

D_dim = real_E * (real_h ** 3) / (12 * (1 - real_nu ** 2))
L = real_a
W = real_q0 * (L ** 4) / D_dim
D_nd = 1.0
a_nd, b_nd, q0_nd, nu_nd = 1.0, 1.0, 1.0, real_nu

# ============================================================
# DOMAIN
# ============================================================
nx, ny = 51, 51
x_nd = np.linspace(0, a_nd, nx)
y_nd = np.linspace(0, b_nd, ny)
X_nd, Y_nd = np.meshgrid(x_nd, y_nd)
grid_points_nd = np.vstack([X_nd.ravel(), Y_nd.ravel()]).T

X_grid = torch.tensor(grid_points_nd, dtype=DTYPE, device=DEVICE)

center_x, center_y = a_nd / 2, b_nd / 2
center_idx = int(np.argmin((grid_points_nd[:, 0] - center_x) ** 2 +
                            (grid_points_nd[:, 1] - center_y) ** 2))

boundary_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
interior_indices = np.setdiff1d(np.arange(nx * ny), boundary_indices)

edge_x_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
edge_y_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
]))

bc_idx = torch.tensor(boundary_indices, dtype=torch.long, device=DEVICE)
int_idx = torch.tensor(interior_indices, dtype=torch.long, device=DEVICE)
edge_x_idx = torch.tensor(edge_x_indices, dtype=torch.long, device=DEVICE)
edge_y_idx = torch.tensor(edge_y_indices, dtype=torch.long, device=DEVICE)

X_int = X_grid[int_idx]        # (N_int, 2) interior collocation points
X_bc = X_grid[bc_idx]          # (N_bc, 2) boundary points (for w=0)
X_edge_x = X_grid[edge_x_idx]  # (N_ex, 2) x=0,a edge points (for Mx=0)
X_edge_y = X_grid[edge_y_idx]  # (N_ey, 2) y=0,b edge points (for My=0)

# ============================================================
# ============================================================
def analytical_w_nd(x, y):
    denom = np.pi ** 4 * D_nd * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mx_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (1 / a_nd ** 2 + nu_nd / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_My_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (nu_nd / a_nd ** 2 + 1 / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mxy_nd(x, y):
    denom = ((1 / a_nd ** 2) + (1 / b_nd ** 2)) ** 2
    return (-q0_nd * (1 - nu_nd) / (np.pi ** 2 * denom * a_nd * b_nd)) * np.cos(np.pi * x / a_nd) * np.cos(np.pi * y / b_nd)


def sinusoidal_load_nd(x, y):
    return q0_nd * torch.sin(np.pi * x / a_nd) * torch.sin(np.pi * y / b_nd)


# ============================================================
# ============================================================
class PlatePINN(nn.Module):
    """
    w(x,y) = D(x,y) * scale * NN(x,y), where D(x,y) = x(a-x)y(b-y) / [(a/2)^2(b/2)^2]
    is a smooth distance function that is EXACTLY zero on all four edges
    (x=0, x=a, y=0, y=b) and positive inside the domain.

    This hard-enforces the essential SS boundary condition w=0 by construction,
    for ANY choice of network weights -- the optimizer can no longer trade off
    boundary satisfaction against the PDE/moment residuals, which was causing
    a visible, non-shrinking violation of w=0 at the edges (and a systematic
    bias in the interior solution) under the previous soft bc_loss penalty.
    Normalizing by (a/2)^2*(b/2)^2 keeps D(x,y)=1 at the plate center, so it
    doesn't change the overall output scale.
    """
    def __init__(self, hidden=64, n_layers=3):
        super().__init__()
        layers = []
        in_dim = 2
        for _ in range(n_layers):
            linear = nn.Linear(in_dim, hidden, dtype=DTYPE)
            nn.init.xavier_normal_(linear.weight)   # matches original kernel_initializer='glorot_normal'
            nn.init.zeros_(linear.bias)              # matches keras Dense default bias_initializer='zeros'
            layers.append(linear)
            layers.append(nn.SiLU())
            in_dim = hidden
        self.hidden = nn.Sequential(*layers)
        self.out = nn.Linear(in_dim, 1, dtype=DTYPE)
        nn.init.zeros_(self.out.weight)   # matches original output_layer kernel_initializer='zeros'
        nn.init.zeros_(self.out.bias)     # network output is exactly 0 at init, controlled by `scale` below
        self.scale = nn.Parameter(torch.tensor(0.01, dtype=DTYPE))

    def forward(self, x, y):
        h = torch.cat([x / a_nd, y / b_nd], dim=1)
        h = self.hidden(h)
        n_out = self.out(h)
        dist = (x * (a_nd - x) * y * (b_nd - y)) / ((a_nd / 2) ** 2 * (b_nd / 2) ** 2)
        return self.scale * dist * n_out


model = PlatePINN().to(DEVICE)

optimizer = torch.optim.Adam(model.parameters(), lr=LR_INIT)
scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer, lr_lambda=lambda step: LR_DECAY_RATE ** (step / LR_DECAY_STEPS)
)


# ============================================================
# FINITE-DIFFERENCE STENCIL HELPERS
# ============================================================
def batched_stencil_eval(x, y, offsets):
    """
    Evaluate the model at (x+dx, y+dy) for every (dx,dy) in `offsets`,
    all in a single stacked forward pass.

    x, y : (N,1) tensors
    offsets : list of (dx, dy) tuples
    returns : dict mapping (dx,dy) -> (N,1) tensor of model outputs
    """
    n = x.shape[0]
    xs = torch.cat([x + dx for dx, dy in offsets], dim=0)
    ys = torch.cat([y + dy for dx, dy in offsets], dim=0)
    w_all = model(xs, ys)
    return {off: w_all[i * n:(i + 1) * n] for i, off in enumerate(offsets)}


def fd_interior_derivatives(x, y, h):
    """Return w_c, w_xxxx, w_yyyy, w_xxyy at interior points via FD stencils."""
    offsets = [
        (0, 0),
        (-2 * h, 0), (-h, 0), (h, 0), (2 * h, 0),      # x-line (for xxxx)
        (0, -2 * h), (0, -h), (0, h), (0, 2 * h),      # y-line (for yyyy)
        (h, h), (h, -h), (-h, h), (-h, -h),            # diagonals (for xxyy)
    ]
    v = batched_stencil_eval(x, y, offsets)

    w_c = v[(0, 0)]
    w_xxxx = (v[(-2 * h, 0)] - 4 * v[(-h, 0)] + 6 * w_c - 4 * v[(h, 0)] + v[(2 * h, 0)]) / h ** 4
    w_yyyy = (v[(0, -2 * h)] - 4 * v[(0, -h)] + 6 * w_c - 4 * v[(0, h)] + v[(0, 2 * h)]) / h ** 4
    w_xxyy = (4 * w_c
              - 2 * (v[(h, 0)] + v[(-h, 0)] + v[(0, h)] + v[(0, -h)])
              + v[(h, h)] + v[(h, -h)] + v[(-h, h)] + v[(-h, -h)]) / h ** 4
    return w_c, w_xxxx, w_yyyy, w_xxyy


def fd_boundary_moments(x, y, h):
    """Return w_c, w_xx, w_yy at boundary points via 3-point FD stencils.

    Note: the central stencil evaluates the network at x-h/x+h (or y-h/y+h),
    which for points exactly on the domain edge means one of the two
    evaluation points lies slightly outside [0,a]x[0,b] (e.g. at x=0 this
    needs w(-h,y)). This is not automatically wrong -- the network is a
    smooth function defined everywhere, not only on the physical domain --
    but it is a methodological choice a reviewer could question. It is
    documented here explicitly as the chosen FDM-PINN stencil treatment for
    this study: central differences, evaluated with a small extension of the
    network beyond the physical edge. A one-sided (forward/backward)
    boundary stencil is a valid alternative if strict domain-only evaluation
    is required; it was not used here so that all boundary derivatives share
    the same (central, 2nd-order accurate) truncation-error behavior as the
    interior stencils."""
    offsets = [(0, 0), (-h, 0), (h, 0), (0, -h), (0, h)]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    return w_c, w_xx, w_yy


def fd_full_field_derivatives(x, y, h):
    """Same as fd_interior_derivatives but also returns w_xy (for Mxy) —
    used only in post-processing over the full grid."""
    offsets = [
        (0, 0),
        (-h, 0), (h, 0), (0, -h), (0, h),
        (h, h), (h, -h), (-h, h), (-h, -h),
    ]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    w_xy = (v[(h, h)] - v[(h, -h)] - v[(-h, h)] + v[(-h, -h)]) / (4 * h ** 2)
    return w_c, w_xx, w_yy, w_xy


# ============================================================
# LOSS (finite-difference residual — replaces autograd chain from Vanilla)
# ============================================================
def compute_loss():
    x_int, y_int = X_int[:, 0:1], X_int[:, 1:2]
    x_bc, y_bc = X_bc[:, 0:1], X_bc[:, 1:2]
    x_ex, y_ex = X_edge_x[:, 0:1], X_edge_x[:, 1:2]
    x_ey, y_ey = X_edge_y[:, 0:1], X_edge_y[:, 1:2]

    _, w_xxxx, w_yyyy, w_xxyy = fd_interior_derivatives(x_int, y_int, FD_STEP)
    q = sinusoidal_load_nd(x_int, y_int)
    pde = w_xxxx + 2 * w_xxyy + w_yyyy - q
    pde_loss = torch.mean(pde ** 2)

    w_bc = model(x_bc, y_bc)
    bc_loss = torch.mean(w_bc ** 2)

    _, w_xx_ex, w_yy_ex = fd_boundary_moments(x_ex, y_ex, FD_STEP)
    Mx_edge = -(w_xx_ex + nu_nd * w_yy_ex)

    _, w_xx_ey, w_yy_ey = fd_boundary_moments(x_ey, y_ey, FD_STEP)
    My_edge = -(w_yy_ey + nu_nd * w_xx_ey)

    moment_loss = torch.mean(Mx_edge ** 2) + torch.mean(My_edge ** 2)

    total = pde_loss + moment_loss

    with torch.no_grad():
        w_full = model(X_grid[:, 0:1], X_grid[:, 1:2])

    return total, pde_loss, bc_loss, moment_loss, w_full


# ============================================================
# ============================================================
epoch_history, loss_history, pde_history, bc_history, moment_history = [], [], [], [], []
w_center_history, epoch_time_history = [], []

best_loss = float("inf")
wait = 0
weights_path = os.path.join(OUT_DIR, "best_weights.pt")

if DEVICE.type == "cuda":
    torch.cuda.reset_peak_memory_stats()

train_start = time.perf_counter()
early_stop_epoch = MAX_EPOCHS  # epoch index where early stopping triggered; MAX_EPOCHS if it never did. This is an early-stopping bookkeeping value, NOT a proven mathematical convergence criterion.

for epoch in range(MAX_EPOCHS):
    epoch_t0 = time.perf_counter()

    optimizer.zero_grad()
    total, pde, bc, moment, w_full = compute_loss()
    total.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)  # matches original Adam(clipnorm=1.0)
    optimizer.step()
    scheduler.step()

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    epoch_dt = time.perf_counter() - epoch_t0

    loss_val = total.item()
    w_center_val = w_full[center_idx, 0].item()

    epoch_history.append(epoch)
    loss_history.append(loss_val)
    pde_history.append(pde.item())
    bc_history.append(bc.item())
    moment_history.append(moment.item())
    w_center_history.append(w_center_val)
    epoch_time_history.append(epoch_dt)

    if epoch % PRINT_EVERY == 0:
        print(f"Epoch {epoch} Loss {loss_val:.6e} Center {w_center_val:.6e} "
              f"({epoch_dt*1000:.2f} ms/epoch)")

    if loss_val < best_loss:
        best_loss = loss_val
        wait = 0
        torch.save(model.state_dict(), weights_path)
    else:
        wait += 1
        if wait >= PATIENCE:
            print(f"Early stopping triggered at epoch {epoch}")
            early_stop_epoch = epoch
            break

total_train_time = time.perf_counter() - train_start
peak_mem_mb = (torch.cuda.max_memory_allocated() / 1e6) if DEVICE.type == "cuda" else float("nan")

model.load_state_dict(torch.load(weights_path, map_location=DEVICE))
model.eval()

# ============================================================
# SAVE TRAINING CONVERGENCE CSV
# ============================================================
w_center_dim_history = np.array(w_center_history) * W
df_hist = pd.DataFrame({
    "epoch": epoch_history,
    "total_loss": loss_history,
    "pde_loss": pde_history,
    "bc_loss": bc_history,
    "moment_loss": moment_history,
    "w_center_nd": w_center_history,
    "w_center_dim": w_center_dim_history,
    "epoch_time_s": epoch_time_history,
})
df_hist.to_csv(os.path.join(OUT_DIR, "training_convergence.csv"), index=False)
print("Saved training_convergence.csv")

# ============================================================
# ============================================================
cost_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "device": str(DEVICE),
    "fd_step_h": FD_STEP,
    "epochs_run": len(epoch_history),
    "epochs_to_early_stop": early_stop_epoch,  # renamed from epochs_to_convergence: this marks when the
    "mean_time_per_epoch_s": float(np.mean(epoch_time_history)),
    "median_time_per_epoch_s": float(np.median(epoch_time_history)),
    "total_training_time_s": total_train_time,
    "peak_gpu_memory_mb": peak_mem_mb,
    "n_trainable_params": sum(p.numel() for p in model.parameters() if p.requires_grad),
}
with open(os.path.join(OUT_DIR, "cost_metrics.json"), "w") as f:
    json.dump(cost_metrics, f, indent=2)
print("Saved cost_metrics.json:", cost_metrics)

# ============================================================
# POST-PROCESSING: full-field predictions via FD stencils
# ============================================================
def compute_full_fields():
    x = X_grid[:, 0:1]
    y = X_grid[:, 1:2]
    with torch.no_grad():
        w_c, w_xx, w_yy, w_xy = fd_full_field_derivatives(x, y, FD_STEP)
        Mx = -(w_xx + nu_nd * w_yy)
        My = -(w_yy + nu_nd * w_xx)
        Mxy = -(1 - nu_nd) * w_xy

    w_grid = w_c.cpu().numpy().reshape(ny, nx)
    Mx_grid = Mx.cpu().numpy().reshape(ny, nx)
    My_grid = My.cpu().numpy().reshape(ny, nx)
    Mxy_grid = Mxy.cpu().numpy().reshape(ny, nx)
    return w_grid, Mx_grid, My_grid, Mxy_grid


w_pred_nd, Mx_pred_nd, My_pred_nd, Mxy_pred_nd = compute_full_fields()

w_anal_nd = analytical_w_nd(X_nd, Y_nd)
Mx_anal_nd = analytical_Mx_nd(X_nd, Y_nd)
My_anal_nd = analytical_My_nd(X_nd, Y_nd)
Mxy_anal_nd = analytical_Mxy_nd(X_nd, Y_nd)

w_dim_pred, w_dim_anal = w_pred_nd * W, w_anal_nd * W
moment_scale = D_dim * W / L ** 2
Mx_dim_pred, Mx_dim_anal = Mx_pred_nd * moment_scale, Mx_anal_nd * moment_scale
My_dim_pred, My_dim_anal = My_pred_nd * moment_scale, My_anal_nd * moment_scale
Mxy_dim_pred, Mxy_dim_anal = Mxy_pred_nd * moment_scale, Mxy_anal_nd * moment_scale

w_signed_error = w_dim_pred - w_dim_anal
w_error = np.abs(w_signed_error)
Mx_error = np.abs(Mx_dim_pred - Mx_dim_anal)
My_error = np.abs(My_dim_pred - My_dim_anal)
Mxy_error = np.abs(Mxy_dim_pred - Mxy_dim_anal)

# ============================================================
# ============================================================
iy, ix = ny // 2, nx // 2
w_pred_center, w_anal_center = w_dim_pred[iy, ix], w_dim_anal[iy, ix]
center_abs_err = abs(w_pred_center - w_anal_center)
center_pct_err = center_abs_err / abs(w_anal_center) * 100

L2_error_w = np.sqrt(np.mean((w_dim_pred - w_dim_anal) ** 2))
rel_L2_error_w = np.linalg.norm(w_dim_pred - w_dim_anal) / np.linalg.norm(w_dim_anal)

error_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "center_deflection_pred_m": float(w_pred_center),
    "center_deflection_anal_m": float(w_anal_center),
    "center_abs_error_m": float(center_abs_err),
    "center_pct_error": float(center_pct_err),
    "w_max_abs_error_m": float(np.max(w_error)),
    "w_rms_error_m": float(np.sqrt(np.mean(w_error ** 2))),
    "w_L2_error_m": float(L2_error_w),
    "w_relative_L2_error": float(rel_L2_error_w),
    "Mx_max_abs_error_Nm": float(np.max(Mx_error)),
    "Mx_rms_error_Nm": float(np.sqrt(np.mean(Mx_error ** 2))),
    "My_max_abs_error_Nm": float(np.max(My_error)),
    "My_rms_error_Nm": float(np.sqrt(np.mean(My_error ** 2))),
    "Mxy_max_abs_error_Nm": float(np.max(Mxy_error)),
    "Mxy_rms_error_Nm": float(np.sqrt(np.mean(Mxy_error ** 2))),
}
with open(os.path.join(OUT_DIR, "error_metrics.json"), "w") as f:
    json.dump(error_metrics, f, indent=2)
print("Saved error_metrics.json:", error_metrics)

df_field = pd.DataFrame({
    "x_nd": X_nd.ravel(), "y_nd": Y_nd.ravel(),
    "w_dim_anal": w_dim_anal.ravel(), "w_dim_pred": w_dim_pred.ravel(),
    "Mx_dim_anal": Mx_dim_anal.ravel(), "Mx_dim_pred": Mx_dim_pred.ravel(),
    "My_dim_anal": My_dim_anal.ravel(), "My_dim_pred": My_dim_pred.ravel(),
    "Mxy_dim_anal": Mxy_dim_anal.ravel(), "Mxy_dim_pred": Mxy_dim_pred.ravel(),
})
df_field["w_abs_error"] = np.abs(df_field["w_dim_anal"] - df_field["w_dim_pred"])
df_field["w_rel_error"] = df_field["w_abs_error"] / (np.abs(df_field["w_dim_anal"]) + 1e-12)
df_field.to_csv(os.path.join(OUT_DIR, "field_data.csv"), index=False)
print("Saved field_data.csv")

# ============================================================
# ============================================================
plt.rcParams.update({"figure.dpi": 150})


def save_contour(field, title, cbar_label, fname, cmap=cm.jet):
    plt.figure(figsize=(7, 6))
    cf = plt.contourf(X_nd, Y_nd, field, levels=50, cmap=cmap)
    plt.colorbar(cf, label=cbar_label)
    plt.title(title, fontstyle="italic", fontsize=13)
    plt.xlabel("x/a", fontstyle="italic")
    plt.ylabel("y/b", fontstyle="italic")
    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, fname))
    plt.close()


plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], df_hist["w_center_dim"], linewidth=2, label=f"{MODEL_NAME} Prediction")
plt.axhline(y=w_anal_center, linestyle="--", linewidth=2, color="k", label="Analytical")
plt.xlabel("Epoch"); plt.ylabel("Center Deflection w (m)")
plt.title("Convergence of Center Deflection")
plt.legend(); plt.grid(True, linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Center_Deflection.png")); plt.close()

plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], np.abs(df_hist["w_center_dim"] - w_anal_center), linewidth=2)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Absolute Error at Center (m)")
plt.title("Convergence Error (Log Scale)")
plt.grid(True, which="both", linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Error_Log_Scale.png")); plt.close()

plt.figure(figsize=(8, 6))
plt.plot(df_hist["epoch"], df_hist["total_loss"], label="Total Loss")
plt.plot(df_hist["epoch"], df_hist["pde_loss"], label="PDE Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["bc_loss"], label="BC Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["moment_loss"], label="Moment Loss", alpha=0.7)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Training Loss Curve")
plt.legend(); plt.grid(True, which="both", ls="--", alpha=0.7); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Training_Loss_Curve.png")); plt.close()

save_contour(w_dim_anal, "Analytical Plate Deflection", "Deflection (m)",
             "Analytical_Plate_Deflection.png")
save_contour(w_dim_pred, f"{MODEL_NAME} Predicted Plate Deflection", "Deflection (m)",
             "FDM-PINN_Predicted_Plate_Deflection.png")
save_contour(w_error, "Absolute Error in Plate Deflection", "|w_pred - w_anal| (m)",
             "Absolute_Error_in_Plate_Deflection.png", cmap=cm.inferno)
save_contour(w_signed_error, "Signed Error w_pred - w_anal", "w_pred - w_anal (m)",
             "Absolute_Error_w_pred_-_w_anal.png", cmap="coolwarm")

save_contour(Mx_dim_anal, "Analytical Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x_Analytical.png")
save_contour(Mx_dim_pred, "Predicted Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x.png")
save_contour(Mx_error, "Absolute Error in $M_x$", "|Mx_pred - Mx_anal| (N·m/m)",
             "Absolute_Error_in_M_x.png", cmap="hot")

save_contour(My_dim_anal, "Analytical Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y_Analytical.png")
save_contour(My_dim_pred, "Predicted Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y.png")
save_contour(My_error, "Absolute Error in $M_y$", "|My_pred - My_anal| (N·m/m)",
             "Absolute_Error_in_M_y.png", cmap="hot")

save_contour(Mxy_dim_anal, "Analytical Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy_Analytical.png")
save_contour(Mxy_dim_pred, "Predicted Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy.png")
save_contour(Mxy_error, "Absolute Error in $M_{xy}$", "|Mxy_pred - Mxy_anal| (N·m/m)",
             "Absolute_Error_in_M_xy.png", cmap="hot")

plt.figure(figsize=(6, 6))
plt.scatter(w_dim_anal.ravel(), w_dim_pred.ravel(), s=8, alpha=0.5)
lims = [min(w_dim_anal.min(), w_dim_pred.min()), max(w_dim_anal.max(), w_dim_pred.max())]
plt.plot(lims, lims, "k--", linewidth=1, label="Perfect agreement")
plt.xlabel("Analytical displacement w (m)", fontstyle="italic")
plt.ylabel("Predicted displacement w (m)", fontstyle="italic")
plt.title("Analytical vs Predicted Displacement", fontstyle="italic")
plt.legend(); plt.grid(True); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Analytical_vs_Predicted_Displacement.png")); plt.close()

err_fft = np.fft.fftshift(np.fft.fft2(w_error))
err_fft_mag = np.log1p(np.abs(err_fft))
plt.figure(figsize=(7, 6))
plt.imshow(err_fft_mag, cmap="magma", extent=[-nx // 2, nx // 2, -ny // 2, ny // 2])
plt.colorbar(label="log(1 + |FFT error|)")
plt.title(f"Error Spectrum (FFT) — {LOADING_CASE}")
plt.xlabel("kx"); plt.ylabel("ky")
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Error_Spectrum_FFT.png")); plt.close()

fig, axes = plt.subplots(4, 3, figsize=(16, 18))
rows = [
    ("Displacement w (m)", w_dim_anal, w_dim_pred, w_error),
    ("Moment Mx (N·m)", Mx_dim_anal, Mx_dim_pred, Mx_error),
    ("Moment My (N·m)", My_dim_anal, My_dim_pred, My_error),
    ("Moment Mxy (N·m)", Mxy_dim_anal, Mxy_dim_pred, Mxy_error),
]
for r, (label, anal, pred, err) in enumerate(rows):
    for c, (title, field, cmap) in enumerate([
        (f"Analytical {label}", anal, cm.jet),
        (f"Predicted {label}", pred, cm.jet),
        (f"Absolute Error {label}", err, cm.inferno),
    ]):
        ax = axes[r, c]
        cf = ax.contourf(X_nd, Y_nd, field, levels=20, cmap=cmap)
        fig.colorbar(cf, ax=ax)
        ax.set_title(title, fontsize=10)
        ax.set_xlabel("x/a"); ax.set_ylabel("y/b")
plt.suptitle(f"{MODEL_NAME} — {LOADING_CASE} loading, SS-SS-SS-SS (summary grid)", fontsize=14)
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "_summary_grid_all_fields.png")); plt.close()

print(f"\nAll outputs written to: {os.path.abspath(OUT_DIR)}")
print(f"Total training time: {total_train_time:.2f}s | "
      f"Epochs to early stop: {early_stop_epoch} | "
      f"FD step h: {FD_STEP}")
```

```python
# ---------------------------------------------------------------------------
# Physics-Informed Neural Network for Kirchhoff Plate Bending
# Architecture : FDM-PINN
# BC Type      : SSSS (Simply Supported: w=0 and Mn=0 on all edges)
# Loading      : N/A - Epoch selection study
# Authors      : Vivek Kumar Yadav, Kshitij Kumar Yadav, Rajesh Kumar
# Institution  : Dept. of Civil Engineering, IIT (BHU) Varanasi
# ---------------------------------------------------------------------------


import os
import json
import time
import random

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from matplotlib import cm

# ============================================================
# CONFIG
# ============================================================
MODEL_NAME = "FDM_PINN"
LOADING_CASE = "sinusoidal"

OUT_DIR = os.path.join("outputs", MODEL_NAME)
FIG_DIR = os.path.join(OUT_DIR, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

SEED = 42
MAX_EPOCHS = 20000     # matches original notebook's training budget (range(15000)); same PATIENCE=3000
PATIENCE = 3000
PRINT_EVERY = 100
LR_INIT = 1e-3
LR_DECAY_STEPS = 2000
LR_DECAY_RATE = 0.95

FD_STEP = 0.01     # finite-difference step size h (non-dimensional units).

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
DTYPE = torch.float64

# ============================================================
# SEEDING
# ============================================================
np.random.seed(SEED)
random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)

# ============================================================
# ============================================================
real_a, real_b, real_h = 1.0, 1.0, 0.01
real_E, real_nu, real_q0 = 200e9, 0.3, 1000.0

D_dim = real_E * (real_h ** 3) / (12 * (1 - real_nu ** 2))
L = real_a
W = real_q0 * (L ** 4) / D_dim
D_nd = 1.0
a_nd, b_nd, q0_nd, nu_nd = 1.0, 1.0, 1.0, real_nu

# ============================================================
# DOMAIN
# ============================================================
nx, ny = 51, 51
x_nd = np.linspace(0, a_nd, nx)
y_nd = np.linspace(0, b_nd, ny)
X_nd, Y_nd = np.meshgrid(x_nd, y_nd)
grid_points_nd = np.vstack([X_nd.ravel(), Y_nd.ravel()]).T

X_grid = torch.tensor(grid_points_nd, dtype=DTYPE, device=DEVICE)

center_x, center_y = a_nd / 2, b_nd / 2
center_idx = int(np.argmin((grid_points_nd[:, 0] - center_x) ** 2 +
                            (grid_points_nd[:, 1] - center_y) ** 2))

boundary_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
interior_indices = np.setdiff1d(np.arange(nx * ny), boundary_indices)

edge_x_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
edge_y_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
]))

bc_idx = torch.tensor(boundary_indices, dtype=torch.long, device=DEVICE)
int_idx = torch.tensor(interior_indices, dtype=torch.long, device=DEVICE)
edge_x_idx = torch.tensor(edge_x_indices, dtype=torch.long, device=DEVICE)
edge_y_idx = torch.tensor(edge_y_indices, dtype=torch.long, device=DEVICE)

X_int = X_grid[int_idx]        # (N_int, 2) interior collocation points
X_bc = X_grid[bc_idx]          # (N_bc, 2) boundary points (for w=0)
X_edge_x = X_grid[edge_x_idx]  # (N_ex, 2) x=0,a edge points (for Mx=0)
X_edge_y = X_grid[edge_y_idx]  # (N_ey, 2) y=0,b edge points (for My=0)

# ============================================================
# ============================================================
def analytical_w_nd(x, y):
    denom = np.pi ** 4 * D_nd * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mx_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (1 / a_nd ** 2 + nu_nd / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_My_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (nu_nd / a_nd ** 2 + 1 / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mxy_nd(x, y):
    denom = ((1 / a_nd ** 2) + (1 / b_nd ** 2)) ** 2
    return (-q0_nd * (1 - nu_nd) / (np.pi ** 2 * denom * a_nd * b_nd)) * np.cos(np.pi * x / a_nd) * np.cos(np.pi * y / b_nd)


def sinusoidal_load_nd(x, y):
    return q0_nd * torch.sin(np.pi * x / a_nd) * torch.sin(np.pi * y / b_nd)


# ============================================================
# ============================================================
class PlatePINN(nn.Module):
    """
    w(x,y) = D(x,y) * scale * NN(x,y), where D(x,y) = x(a-x)y(b-y) / [(a/2)^2(b/2)^2]
    is a smooth distance function that is EXACTLY zero on all four edges
    (x=0, x=a, y=0, y=b) and positive inside the domain.

    This hard-enforces the essential SS boundary condition w=0 by construction,
    for ANY choice of network weights -- the optimizer can no longer trade off
    boundary satisfaction against the PDE/moment residuals, which was causing
    a visible, non-shrinking violation of w=0 at the edges (and a systematic
    bias in the interior solution) under the previous soft bc_loss penalty.
    Normalizing by (a/2)^2*(b/2)^2 keeps D(x,y)=1 at the plate center, so it
    doesn't change the overall output scale.
    """
    def __init__(self, hidden=64, n_layers=3):
        super().__init__()
        layers = []
        in_dim = 2
        for _ in range(n_layers):
            linear = nn.Linear(in_dim, hidden, dtype=DTYPE)
            nn.init.xavier_normal_(linear.weight)   # matches original kernel_initializer='glorot_normal'
            nn.init.zeros_(linear.bias)              # matches keras Dense default bias_initializer='zeros'
            layers.append(linear)
            layers.append(nn.SiLU())
            in_dim = hidden
        self.hidden = nn.Sequential(*layers)
        self.out = nn.Linear(in_dim, 1, dtype=DTYPE)
        nn.init.zeros_(self.out.weight)   # matches original output_layer kernel_initializer='zeros'
        nn.init.zeros_(self.out.bias)     # network output is exactly 0 at init, controlled by `scale` below
        self.scale = nn.Parameter(torch.tensor(0.01, dtype=DTYPE))

    def forward(self, x, y):
        h = torch.cat([x / a_nd, y / b_nd], dim=1)
        h = self.hidden(h)
        n_out = self.out(h)
        dist = (x * (a_nd - x) * y * (b_nd - y)) / ((a_nd / 2) ** 2 * (b_nd / 2) ** 2)
        return self.scale * dist * n_out


model = PlatePINN().to(DEVICE)

optimizer = torch.optim.Adam(model.parameters(), lr=LR_INIT)
scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer, lr_lambda=lambda step: LR_DECAY_RATE ** (step / LR_DECAY_STEPS)
)


# ============================================================
# FINITE-DIFFERENCE STENCIL HELPERS
# ============================================================
def batched_stencil_eval(x, y, offsets):
    """
    Evaluate the model at (x+dx, y+dy) for every (dx,dy) in `offsets`,
    all in a single stacked forward pass.

    x, y : (N,1) tensors
    offsets : list of (dx, dy) tuples
    returns : dict mapping (dx,dy) -> (N,1) tensor of model outputs
    """
    n = x.shape[0]
    xs = torch.cat([x + dx for dx, dy in offsets], dim=0)
    ys = torch.cat([y + dy for dx, dy in offsets], dim=0)
    w_all = model(xs, ys)
    return {off: w_all[i * n:(i + 1) * n] for i, off in enumerate(offsets)}


def fd_interior_derivatives(x, y, h):
    """Return w_c, w_xxxx, w_yyyy, w_xxyy at interior points via FD stencils."""
    offsets = [
        (0, 0),
        (-2 * h, 0), (-h, 0), (h, 0), (2 * h, 0),      # x-line (for xxxx)
        (0, -2 * h), (0, -h), (0, h), (0, 2 * h),      # y-line (for yyyy)
        (h, h), (h, -h), (-h, h), (-h, -h),            # diagonals (for xxyy)
    ]
    v = batched_stencil_eval(x, y, offsets)

    w_c = v[(0, 0)]
    w_xxxx = (v[(-2 * h, 0)] - 4 * v[(-h, 0)] + 6 * w_c - 4 * v[(h, 0)] + v[(2 * h, 0)]) / h ** 4
    w_yyyy = (v[(0, -2 * h)] - 4 * v[(0, -h)] + 6 * w_c - 4 * v[(0, h)] + v[(0, 2 * h)]) / h ** 4
    w_xxyy = (4 * w_c
              - 2 * (v[(h, 0)] + v[(-h, 0)] + v[(0, h)] + v[(0, -h)])
              + v[(h, h)] + v[(h, -h)] + v[(-h, h)] + v[(-h, -h)]) / h ** 4
    return w_c, w_xxxx, w_yyyy, w_xxyy


def fd_boundary_moments(x, y, h):
    """Return w_c, w_xx, w_yy at boundary points via 3-point FD stencils.

    Note: the central stencil evaluates the network at x-h/x+h (or y-h/y+h),
    which for points exactly on the domain edge means one of the two
    evaluation points lies slightly outside [0,a]x[0,b] (e.g. at x=0 this
    needs w(-h,y)). This is not automatically wrong -- the network is a
    smooth function defined everywhere, not only on the physical domain --
    but it is a methodological choice a reviewer could question. It is
    documented here explicitly as the chosen FDM-PINN stencil treatment for
    this study: central differences, evaluated with a small extension of the
    network beyond the physical edge. A one-sided (forward/backward)
    boundary stencil is a valid alternative if strict domain-only evaluation
    is required; it was not used here so that all boundary derivatives share
    the same (central, 2nd-order accurate) truncation-error behavior as the
    interior stencils."""
    offsets = [(0, 0), (-h, 0), (h, 0), (0, -h), (0, h)]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    return w_c, w_xx, w_yy


def fd_full_field_derivatives(x, y, h):
    """Same as fd_interior_derivatives but also returns w_xy (for Mxy) —
    used only in post-processing over the full grid."""
    offsets = [
        (0, 0),
        (-h, 0), (h, 0), (0, -h), (0, h),
        (h, h), (h, -h), (-h, h), (-h, -h),
    ]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    w_xy = (v[(h, h)] - v[(h, -h)] - v[(-h, h)] + v[(-h, -h)]) / (4 * h ** 2)
    return w_c, w_xx, w_yy, w_xy


# ============================================================
# LOSS (finite-difference residual — replaces autograd chain from Vanilla)
# ============================================================
def compute_loss():
    x_int, y_int = X_int[:, 0:1], X_int[:, 1:2]
    x_bc, y_bc = X_bc[:, 0:1], X_bc[:, 1:2]
    x_ex, y_ex = X_edge_x[:, 0:1], X_edge_x[:, 1:2]
    x_ey, y_ey = X_edge_y[:, 0:1], X_edge_y[:, 1:2]

    _, w_xxxx, w_yyyy, w_xxyy = fd_interior_derivatives(x_int, y_int, FD_STEP)
    q = sinusoidal_load_nd(x_int, y_int)
    pde = w_xxxx + 2 * w_xxyy + w_yyyy - q
    pde_loss = torch.mean(pde ** 2)

    w_bc = model(x_bc, y_bc)
    bc_loss = torch.mean(w_bc ** 2)

    _, w_xx_ex, w_yy_ex = fd_boundary_moments(x_ex, y_ex, FD_STEP)
    Mx_edge = -(w_xx_ex + nu_nd * w_yy_ex)

    _, w_xx_ey, w_yy_ey = fd_boundary_moments(x_ey, y_ey, FD_STEP)
    My_edge = -(w_yy_ey + nu_nd * w_xx_ey)

    moment_loss = torch.mean(Mx_edge ** 2) + torch.mean(My_edge ** 2)

    total = pde_loss + moment_loss

    with torch.no_grad():
        w_full = model(X_grid[:, 0:1], X_grid[:, 1:2])

    return total, pde_loss, bc_loss, moment_loss, w_full


# ============================================================
# ============================================================
epoch_history, loss_history, pde_history, bc_history, moment_history = [], [], [], [], []
w_center_history, epoch_time_history = [], []

best_loss = float("inf")
wait = 0
weights_path = os.path.join(OUT_DIR, "best_weights.pt")

if DEVICE.type == "cuda":
    torch.cuda.reset_peak_memory_stats()

train_start = time.perf_counter()
early_stop_epoch = MAX_EPOCHS  # epoch index where early stopping triggered; MAX_EPOCHS if it never did. This is an early-stopping bookkeeping value, NOT a proven mathematical convergence criterion.

for epoch in range(MAX_EPOCHS):
    epoch_t0 = time.perf_counter()

    optimizer.zero_grad()
    total, pde, bc, moment, w_full = compute_loss()
    total.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)  # matches original Adam(clipnorm=1.0)
    optimizer.step()
    scheduler.step()

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    epoch_dt = time.perf_counter() - epoch_t0

    loss_val = total.item()
    w_center_val = w_full[center_idx, 0].item()

    epoch_history.append(epoch)
    loss_history.append(loss_val)
    pde_history.append(pde.item())
    bc_history.append(bc.item())
    moment_history.append(moment.item())
    w_center_history.append(w_center_val)
    epoch_time_history.append(epoch_dt)

    if epoch % PRINT_EVERY == 0:
        print(f"Epoch {epoch} Loss {loss_val:.6e} Center {w_center_val:.6e} "
              f"({epoch_dt*1000:.2f} ms/epoch)")

    if loss_val < best_loss:
        best_loss = loss_val
        wait = 0
        torch.save(model.state_dict(), weights_path)
    else:
        wait += 1
        if wait >= PATIENCE:
            print(f"Early stopping triggered at epoch {epoch}")
            early_stop_epoch = epoch
            break

total_train_time = time.perf_counter() - train_start
peak_mem_mb = (torch.cuda.max_memory_allocated() / 1e6) if DEVICE.type == "cuda" else float("nan")

model.load_state_dict(torch.load(weights_path, map_location=DEVICE))
model.eval()

# ============================================================
# SAVE TRAINING CONVERGENCE CSV
# ============================================================
w_center_dim_history = np.array(w_center_history) * W
df_hist = pd.DataFrame({
    "epoch": epoch_history,
    "total_loss": loss_history,
    "pde_loss": pde_history,
    "bc_loss": bc_history,
    "moment_loss": moment_history,
    "w_center_nd": w_center_history,
    "w_center_dim": w_center_dim_history,
    "epoch_time_s": epoch_time_history,
})
df_hist.to_csv(os.path.join(OUT_DIR, "training_convergence.csv"), index=False)
print("Saved training_convergence.csv")

# ============================================================
# ============================================================
cost_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "device": str(DEVICE),
    "fd_step_h": FD_STEP,
    "epochs_run": len(epoch_history),
    "epochs_to_early_stop": early_stop_epoch,  # renamed from epochs_to_convergence: this marks when the
    "mean_time_per_epoch_s": float(np.mean(epoch_time_history)),
    "median_time_per_epoch_s": float(np.median(epoch_time_history)),
    "total_training_time_s": total_train_time,
    "peak_gpu_memory_mb": peak_mem_mb,
    "n_trainable_params": sum(p.numel() for p in model.parameters() if p.requires_grad),
}
with open(os.path.join(OUT_DIR, "cost_metrics.json"), "w") as f:
    json.dump(cost_metrics, f, indent=2)
print("Saved cost_metrics.json:", cost_metrics)

# ============================================================
# POST-PROCESSING: full-field predictions via FD stencils
# ============================================================
def compute_full_fields():
    x = X_grid[:, 0:1]
    y = X_grid[:, 1:2]
    with torch.no_grad():
        w_c, w_xx, w_yy, w_xy = fd_full_field_derivatives(x, y, FD_STEP)
        Mx = -(w_xx + nu_nd * w_yy)
        My = -(w_yy + nu_nd * w_xx)
        Mxy = -(1 - nu_nd) * w_xy

    w_grid = w_c.cpu().numpy().reshape(ny, nx)
    Mx_grid = Mx.cpu().numpy().reshape(ny, nx)
    My_grid = My.cpu().numpy().reshape(ny, nx)
    Mxy_grid = Mxy.cpu().numpy().reshape(ny, nx)
    return w_grid, Mx_grid, My_grid, Mxy_grid


w_pred_nd, Mx_pred_nd, My_pred_nd, Mxy_pred_nd = compute_full_fields()

w_anal_nd = analytical_w_nd(X_nd, Y_nd)
Mx_anal_nd = analytical_Mx_nd(X_nd, Y_nd)
My_anal_nd = analytical_My_nd(X_nd, Y_nd)
Mxy_anal_nd = analytical_Mxy_nd(X_nd, Y_nd)

w_dim_pred, w_dim_anal = w_pred_nd * W, w_anal_nd * W
moment_scale = D_dim * W / L ** 2
Mx_dim_pred, Mx_dim_anal = Mx_pred_nd * moment_scale, Mx_anal_nd * moment_scale
My_dim_pred, My_dim_anal = My_pred_nd * moment_scale, My_anal_nd * moment_scale
Mxy_dim_pred, Mxy_dim_anal = Mxy_pred_nd * moment_scale, Mxy_anal_nd * moment_scale

w_signed_error = w_dim_pred - w_dim_anal
w_error = np.abs(w_signed_error)
Mx_error = np.abs(Mx_dim_pred - Mx_dim_anal)
My_error = np.abs(My_dim_pred - My_dim_anal)
Mxy_error = np.abs(Mxy_dim_pred - Mxy_dim_anal)

# ============================================================
# ============================================================
iy, ix = ny // 2, nx // 2
w_pred_center, w_anal_center = w_dim_pred[iy, ix], w_dim_anal[iy, ix]
center_abs_err = abs(w_pred_center - w_anal_center)
center_pct_err = center_abs_err / abs(w_anal_center) * 100

L2_error_w = np.sqrt(np.mean((w_dim_pred - w_dim_anal) ** 2))
rel_L2_error_w = np.linalg.norm(w_dim_pred - w_dim_anal) / np.linalg.norm(w_dim_anal)

error_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "center_deflection_pred_m": float(w_pred_center),
    "center_deflection_anal_m": float(w_anal_center),
    "center_abs_error_m": float(center_abs_err),
    "center_pct_error": float(center_pct_err),
    "w_max_abs_error_m": float(np.max(w_error)),
    "w_rms_error_m": float(np.sqrt(np.mean(w_error ** 2))),
    "w_L2_error_m": float(L2_error_w),
    "w_relative_L2_error": float(rel_L2_error_w),
    "Mx_max_abs_error_Nm": float(np.max(Mx_error)),
    "Mx_rms_error_Nm": float(np.sqrt(np.mean(Mx_error ** 2))),
    "My_max_abs_error_Nm": float(np.max(My_error)),
    "My_rms_error_Nm": float(np.sqrt(np.mean(My_error ** 2))),
    "Mxy_max_abs_error_Nm": float(np.max(Mxy_error)),
    "Mxy_rms_error_Nm": float(np.sqrt(np.mean(Mxy_error ** 2))),
}
with open(os.path.join(OUT_DIR, "error_metrics.json"), "w") as f:
    json.dump(error_metrics, f, indent=2)
print("Saved error_metrics.json:", error_metrics)

df_field = pd.DataFrame({
    "x_nd": X_nd.ravel(), "y_nd": Y_nd.ravel(),
    "w_dim_anal": w_dim_anal.ravel(), "w_dim_pred": w_dim_pred.ravel(),
    "Mx_dim_anal": Mx_dim_anal.ravel(), "Mx_dim_pred": Mx_dim_pred.ravel(),
    "My_dim_anal": My_dim_anal.ravel(), "My_dim_pred": My_dim_pred.ravel(),
    "Mxy_dim_anal": Mxy_dim_anal.ravel(), "Mxy_dim_pred": Mxy_dim_pred.ravel(),
})
df_field["w_abs_error"] = np.abs(df_field["w_dim_anal"] - df_field["w_dim_pred"])
df_field["w_rel_error"] = df_field["w_abs_error"] / (np.abs(df_field["w_dim_anal"]) + 1e-12)
df_field.to_csv(os.path.join(OUT_DIR, "field_data.csv"), index=False)
print("Saved field_data.csv")

# ============================================================
# ============================================================
plt.rcParams.update({"figure.dpi": 150})


def save_contour(field, title, cbar_label, fname, cmap=cm.jet):
    plt.figure(figsize=(7, 6))
    cf = plt.contourf(X_nd, Y_nd, field, levels=50, cmap=cmap)
    plt.colorbar(cf, label=cbar_label)
    plt.title(title, fontstyle="italic", fontsize=13)
    plt.xlabel("x/a", fontstyle="italic")
    plt.ylabel("y/b", fontstyle="italic")
    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, fname))
    plt.close()


plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], df_hist["w_center_dim"], linewidth=2, label=f"{MODEL_NAME} Prediction")
plt.axhline(y=w_anal_center, linestyle="--", linewidth=2, color="k", label="Analytical")
plt.xlabel("Epoch"); plt.ylabel("Center Deflection w (m)")
plt.title("Convergence of Center Deflection")
plt.legend(); plt.grid(True, linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Center_Deflection.png")); plt.close()

plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], np.abs(df_hist["w_center_dim"] - w_anal_center), linewidth=2)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Absolute Error at Center (m)")
plt.title("Convergence Error (Log Scale)")
plt.grid(True, which="both", linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Error_Log_Scale.png")); plt.close()

plt.figure(figsize=(8, 6))
plt.plot(df_hist["epoch"], df_hist["total_loss"], label="Total Loss")
plt.plot(df_hist["epoch"], df_hist["pde_loss"], label="PDE Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["bc_loss"], label="BC Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["moment_loss"], label="Moment Loss", alpha=0.7)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Training Loss Curve")
plt.legend(); plt.grid(True, which="both", ls="--", alpha=0.7); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Training_Loss_Curve.png")); plt.close()

save_contour(w_dim_anal, "Analytical Plate Deflection", "Deflection (m)",
             "Analytical_Plate_Deflection.png")
save_contour(w_dim_pred, f"{MODEL_NAME} Predicted Plate Deflection", "Deflection (m)",
             "FDM-PINN_Predicted_Plate_Deflection.png")
save_contour(w_error, "Absolute Error in Plate Deflection", "|w_pred - w_anal| (m)",
             "Absolute_Error_in_Plate_Deflection.png", cmap=cm.inferno)
save_contour(w_signed_error, "Signed Error w_pred - w_anal", "w_pred - w_anal (m)",
             "Absolute_Error_w_pred_-_w_anal.png", cmap="coolwarm")

save_contour(Mx_dim_anal, "Analytical Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x_Analytical.png")
save_contour(Mx_dim_pred, "Predicted Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x.png")
save_contour(Mx_error, "Absolute Error in $M_x$", "|Mx_pred - Mx_anal| (N·m/m)",
             "Absolute_Error_in_M_x.png", cmap="hot")

save_contour(My_dim_anal, "Analytical Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y_Analytical.png")
save_contour(My_dim_pred, "Predicted Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y.png")
save_contour(My_error, "Absolute Error in $M_y$", "|My_pred - My_anal| (N·m/m)",
             "Absolute_Error_in_M_y.png", cmap="hot")

save_contour(Mxy_dim_anal, "Analytical Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy_Analytical.png")
save_contour(Mxy_dim_pred, "Predicted Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy.png")
save_contour(Mxy_error, "Absolute Error in $M_{xy}$", "|Mxy_pred - Mxy_anal| (N·m/m)",
             "Absolute_Error_in_M_xy.png", cmap="hot")

plt.figure(figsize=(6, 6))
plt.scatter(w_dim_anal.ravel(), w_dim_pred.ravel(), s=8, alpha=0.5)
lims = [min(w_dim_anal.min(), w_dim_pred.min()), max(w_dim_anal.max(), w_dim_pred.max())]
plt.plot(lims, lims, "k--", linewidth=1, label="Perfect agreement")
plt.xlabel("Analytical displacement w (m)", fontstyle="italic")
plt.ylabel("Predicted displacement w (m)", fontstyle="italic")
plt.title("Analytical vs Predicted Displacement", fontstyle="italic")
plt.legend(); plt.grid(True); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Analytical_vs_Predicted_Displacement.png")); plt.close()

err_fft = np.fft.fftshift(np.fft.fft2(w_error))
err_fft_mag = np.log1p(np.abs(err_fft))
plt.figure(figsize=(7, 6))
plt.imshow(err_fft_mag, cmap="magma", extent=[-nx // 2, nx // 2, -ny // 2, ny // 2])
plt.colorbar(label="log(1 + |FFT error|)")
plt.title(f"Error Spectrum (FFT) — {LOADING_CASE}")
plt.xlabel("kx"); plt.ylabel("ky")
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Error_Spectrum_FFT.png")); plt.close()

fig, axes = plt.subplots(4, 3, figsize=(16, 18))
rows = [
    ("Displacement w (m)", w_dim_anal, w_dim_pred, w_error),
    ("Moment Mx (N·m)", Mx_dim_anal, Mx_dim_pred, Mx_error),
    ("Moment My (N·m)", My_dim_anal, My_dim_pred, My_error),
    ("Moment Mxy (N·m)", Mxy_dim_anal, Mxy_dim_pred, Mxy_error),
]
for r, (label, anal, pred, err) in enumerate(rows):
    for c, (title, field, cmap) in enumerate([
        (f"Analytical {label}", anal, cm.jet),
        (f"Predicted {label}", pred, cm.jet),
        (f"Absolute Error {label}", err, cm.inferno),
    ]):
        ax = axes[r, c]
        cf = ax.contourf(X_nd, Y_nd, field, levels=20, cmap=cmap)
        fig.colorbar(cf, ax=ax)
        ax.set_title(title, fontsize=10)
        ax.set_xlabel("x/a"); ax.set_ylabel("y/b")
plt.suptitle(f"{MODEL_NAME} — {LOADING_CASE} loading, SS-SS-SS-SS (summary grid)", fontsize=14)
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "_summary_grid_all_fields.png")); plt.close()

print(f"\nAll outputs written to: {os.path.abspath(OUT_DIR)}")
print(f"Total training time: {total_train_time:.2f}s | "
      f"Epochs to early stop: {early_stop_epoch} | "
      f"FD step h: {FD_STEP}")
```

```python
# ---------------------------------------------------------------------------
# Physics-Informed Neural Network for Kirchhoff Plate Bending
# Architecture : FDM-PINN
# BC Type      : SSSS (Simply Supported: w=0 and Mn=0 on all edges)
# Loading      : N/A - Epoch selection study
# Authors      : Vivek Kumar Yadav, Kshitij Kumar Yadav, Rajesh Kumar
# Institution  : Dept. of Civil Engineering, IIT (BHU) Varanasi
# ---------------------------------------------------------------------------


import os
import json
import time
import random

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from matplotlib import cm

# ============================================================
# CONFIG
# ============================================================
MODEL_NAME = "FDM_PINN"
LOADING_CASE = "sinusoidal"

OUT_DIR = os.path.join("outputs", MODEL_NAME)
FIG_DIR = os.path.join(OUT_DIR, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

SEED = 42
MAX_EPOCHS = 30000     # matches original notebook's training budget (range(15000)); same PATIENCE=3000
PATIENCE = 3000
PRINT_EVERY = 100
LR_INIT = 1e-3
LR_DECAY_STEPS = 2000
LR_DECAY_RATE = 0.95

FD_STEP = 0.01     # finite-difference step size h (non-dimensional units).

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
DTYPE = torch.float64

# ============================================================
# SEEDING
# ============================================================
np.random.seed(SEED)
random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)

# ============================================================
# ============================================================
real_a, real_b, real_h = 1.0, 1.0, 0.01
real_E, real_nu, real_q0 = 200e9, 0.3, 1000.0

D_dim = real_E * (real_h ** 3) / (12 * (1 - real_nu ** 2))
L = real_a
W = real_q0 * (L ** 4) / D_dim
D_nd = 1.0
a_nd, b_nd, q0_nd, nu_nd = 1.0, 1.0, 1.0, real_nu

# ============================================================
# DOMAIN
# ============================================================
nx, ny = 51, 51
x_nd = np.linspace(0, a_nd, nx)
y_nd = np.linspace(0, b_nd, ny)
X_nd, Y_nd = np.meshgrid(x_nd, y_nd)
grid_points_nd = np.vstack([X_nd.ravel(), Y_nd.ravel()]).T

X_grid = torch.tensor(grid_points_nd, dtype=DTYPE, device=DEVICE)

center_x, center_y = a_nd / 2, b_nd / 2
center_idx = int(np.argmin((grid_points_nd[:, 0] - center_x) ** 2 +
                            (grid_points_nd[:, 1] - center_y) ** 2))

boundary_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
interior_indices = np.setdiff1d(np.arange(nx * ny), boundary_indices)

edge_x_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
edge_y_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
]))

bc_idx = torch.tensor(boundary_indices, dtype=torch.long, device=DEVICE)
int_idx = torch.tensor(interior_indices, dtype=torch.long, device=DEVICE)
edge_x_idx = torch.tensor(edge_x_indices, dtype=torch.long, device=DEVICE)
edge_y_idx = torch.tensor(edge_y_indices, dtype=torch.long, device=DEVICE)

X_int = X_grid[int_idx]        # (N_int, 2) interior collocation points
X_bc = X_grid[bc_idx]          # (N_bc, 2) boundary points (for w=0)
X_edge_x = X_grid[edge_x_idx]  # (N_ex, 2) x=0,a edge points (for Mx=0)
X_edge_y = X_grid[edge_y_idx]  # (N_ey, 2) y=0,b edge points (for My=0)

# ============================================================
# ============================================================
def analytical_w_nd(x, y):
    denom = np.pi ** 4 * D_nd * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mx_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (1 / a_nd ** 2 + nu_nd / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_My_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (nu_nd / a_nd ** 2 + 1 / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mxy_nd(x, y):
    denom = ((1 / a_nd ** 2) + (1 / b_nd ** 2)) ** 2
    return (-q0_nd * (1 - nu_nd) / (np.pi ** 2 * denom * a_nd * b_nd)) * np.cos(np.pi * x / a_nd) * np.cos(np.pi * y / b_nd)


def sinusoidal_load_nd(x, y):
    return q0_nd * torch.sin(np.pi * x / a_nd) * torch.sin(np.pi * y / b_nd)


# ============================================================
# ============================================================
class PlatePINN(nn.Module):
    """
    w(x,y) = D(x,y) * scale * NN(x,y), where D(x,y) = x(a-x)y(b-y) / [(a/2)^2(b/2)^2]
    is a smooth distance function that is EXACTLY zero on all four edges
    (x=0, x=a, y=0, y=b) and positive inside the domain.

    This hard-enforces the essential SS boundary condition w=0 by construction,
    for ANY choice of network weights -- the optimizer can no longer trade off
    boundary satisfaction against the PDE/moment residuals, which was causing
    a visible, non-shrinking violation of w=0 at the edges (and a systematic
    bias in the interior solution) under the previous soft bc_loss penalty.
    Normalizing by (a/2)^2*(b/2)^2 keeps D(x,y)=1 at the plate center, so it
    doesn't change the overall output scale.
    """
    def __init__(self, hidden=64, n_layers=3):
        super().__init__()
        layers = []
        in_dim = 2
        for _ in range(n_layers):
            linear = nn.Linear(in_dim, hidden, dtype=DTYPE)
            nn.init.xavier_normal_(linear.weight)   # matches original kernel_initializer='glorot_normal'
            nn.init.zeros_(linear.bias)              # matches keras Dense default bias_initializer='zeros'
            layers.append(linear)
            layers.append(nn.SiLU())
            in_dim = hidden
        self.hidden = nn.Sequential(*layers)
        self.out = nn.Linear(in_dim, 1, dtype=DTYPE)
        nn.init.zeros_(self.out.weight)   # matches original output_layer kernel_initializer='zeros'
        nn.init.zeros_(self.out.bias)     # network output is exactly 0 at init, controlled by `scale` below
        self.scale = nn.Parameter(torch.tensor(0.01, dtype=DTYPE))

    def forward(self, x, y):
        h = torch.cat([x / a_nd, y / b_nd], dim=1)
        h = self.hidden(h)
        n_out = self.out(h)
        dist = (x * (a_nd - x) * y * (b_nd - y)) / ((a_nd / 2) ** 2 * (b_nd / 2) ** 2)
        return self.scale * dist * n_out


model = PlatePINN().to(DEVICE)

optimizer = torch.optim.Adam(model.parameters(), lr=LR_INIT)
scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer, lr_lambda=lambda step: LR_DECAY_RATE ** (step / LR_DECAY_STEPS)
)


# ============================================================
# FINITE-DIFFERENCE STENCIL HELPERS
# ============================================================
def batched_stencil_eval(x, y, offsets):
    """
    Evaluate the model at (x+dx, y+dy) for every (dx,dy) in `offsets`,
    all in a single stacked forward pass.

    x, y : (N,1) tensors
    offsets : list of (dx, dy) tuples
    returns : dict mapping (dx,dy) -> (N,1) tensor of model outputs
    """
    n = x.shape[0]
    xs = torch.cat([x + dx for dx, dy in offsets], dim=0)
    ys = torch.cat([y + dy for dx, dy in offsets], dim=0)
    w_all = model(xs, ys)
    return {off: w_all[i * n:(i + 1) * n] for i, off in enumerate(offsets)}


def fd_interior_derivatives(x, y, h):
    """Return w_c, w_xxxx, w_yyyy, w_xxyy at interior points via FD stencils."""
    offsets = [
        (0, 0),
        (-2 * h, 0), (-h, 0), (h, 0), (2 * h, 0),      # x-line (for xxxx)
        (0, -2 * h), (0, -h), (0, h), (0, 2 * h),      # y-line (for yyyy)
        (h, h), (h, -h), (-h, h), (-h, -h),            # diagonals (for xxyy)
    ]
    v = batched_stencil_eval(x, y, offsets)

    w_c = v[(0, 0)]
    w_xxxx = (v[(-2 * h, 0)] - 4 * v[(-h, 0)] + 6 * w_c - 4 * v[(h, 0)] + v[(2 * h, 0)]) / h ** 4
    w_yyyy = (v[(0, -2 * h)] - 4 * v[(0, -h)] + 6 * w_c - 4 * v[(0, h)] + v[(0, 2 * h)]) / h ** 4
    w_xxyy = (4 * w_c
              - 2 * (v[(h, 0)] + v[(-h, 0)] + v[(0, h)] + v[(0, -h)])
              + v[(h, h)] + v[(h, -h)] + v[(-h, h)] + v[(-h, -h)]) / h ** 4
    return w_c, w_xxxx, w_yyyy, w_xxyy


def fd_boundary_moments(x, y, h):
    """Return w_c, w_xx, w_yy at boundary points via 3-point FD stencils.

    Note: the central stencil evaluates the network at x-h/x+h (or y-h/y+h),
    which for points exactly on the domain edge means one of the two
    evaluation points lies slightly outside [0,a]x[0,b] (e.g. at x=0 this
    needs w(-h,y)). This is not automatically wrong -- the network is a
    smooth function defined everywhere, not only on the physical domain --
    but it is a methodological choice a reviewer could question. It is
    documented here explicitly as the chosen FDM-PINN stencil treatment for
    this study: central differences, evaluated with a small extension of the
    network beyond the physical edge. A one-sided (forward/backward)
    boundary stencil is a valid alternative if strict domain-only evaluation
    is required; it was not used here so that all boundary derivatives share
    the same (central, 2nd-order accurate) truncation-error behavior as the
    interior stencils."""
    offsets = [(0, 0), (-h, 0), (h, 0), (0, -h), (0, h)]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    return w_c, w_xx, w_yy


def fd_full_field_derivatives(x, y, h):
    """Same as fd_interior_derivatives but also returns w_xy (for Mxy) —
    used only in post-processing over the full grid."""
    offsets = [
        (0, 0),
        (-h, 0), (h, 0), (0, -h), (0, h),
        (h, h), (h, -h), (-h, h), (-h, -h),
    ]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    w_xy = (v[(h, h)] - v[(h, -h)] - v[(-h, h)] + v[(-h, -h)]) / (4 * h ** 2)
    return w_c, w_xx, w_yy, w_xy


# ============================================================
# LOSS (finite-difference residual — replaces autograd chain from Vanilla)
# ============================================================
def compute_loss():
    x_int, y_int = X_int[:, 0:1], X_int[:, 1:2]
    x_bc, y_bc = X_bc[:, 0:1], X_bc[:, 1:2]
    x_ex, y_ex = X_edge_x[:, 0:1], X_edge_x[:, 1:2]
    x_ey, y_ey = X_edge_y[:, 0:1], X_edge_y[:, 1:2]

    _, w_xxxx, w_yyyy, w_xxyy = fd_interior_derivatives(x_int, y_int, FD_STEP)
    q = sinusoidal_load_nd(x_int, y_int)
    pde = w_xxxx + 2 * w_xxyy + w_yyyy - q
    pde_loss = torch.mean(pde ** 2)

    w_bc = model(x_bc, y_bc)
    bc_loss = torch.mean(w_bc ** 2)

    _, w_xx_ex, w_yy_ex = fd_boundary_moments(x_ex, y_ex, FD_STEP)
    Mx_edge = -(w_xx_ex + nu_nd * w_yy_ex)

    _, w_xx_ey, w_yy_ey = fd_boundary_moments(x_ey, y_ey, FD_STEP)
    My_edge = -(w_yy_ey + nu_nd * w_xx_ey)

    moment_loss = torch.mean(Mx_edge ** 2) + torch.mean(My_edge ** 2)

    total = pde_loss + moment_loss

    with torch.no_grad():
        w_full = model(X_grid[:, 0:1], X_grid[:, 1:2])

    return total, pde_loss, bc_loss, moment_loss, w_full


# ============================================================
# ============================================================
epoch_history, loss_history, pde_history, bc_history, moment_history = [], [], [], [], []
w_center_history, epoch_time_history = [], []

best_loss = float("inf")
wait = 0
weights_path = os.path.join(OUT_DIR, "best_weights.pt")

if DEVICE.type == "cuda":
    torch.cuda.reset_peak_memory_stats()

train_start = time.perf_counter()
early_stop_epoch = MAX_EPOCHS  # epoch index where early stopping triggered; MAX_EPOCHS if it never did. This is an early-stopping bookkeeping value, NOT a proven mathematical convergence criterion.

for epoch in range(MAX_EPOCHS):
    epoch_t0 = time.perf_counter()

    optimizer.zero_grad()
    total, pde, bc, moment, w_full = compute_loss()
    total.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)  # matches original Adam(clipnorm=1.0)
    optimizer.step()
    scheduler.step()

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    epoch_dt = time.perf_counter() - epoch_t0

    loss_val = total.item()
    w_center_val = w_full[center_idx, 0].item()

    epoch_history.append(epoch)
    loss_history.append(loss_val)
    pde_history.append(pde.item())
    bc_history.append(bc.item())
    moment_history.append(moment.item())
    w_center_history.append(w_center_val)
    epoch_time_history.append(epoch_dt)

    if epoch % PRINT_EVERY == 0:
        print(f"Epoch {epoch} Loss {loss_val:.6e} Center {w_center_val:.6e} "
              f"({epoch_dt*1000:.2f} ms/epoch)")

    if loss_val < best_loss:
        best_loss = loss_val
        wait = 0
        torch.save(model.state_dict(), weights_path)
    else:
        wait += 1
        if wait >= PATIENCE:
            print(f"Early stopping triggered at epoch {epoch}")
            early_stop_epoch = epoch
            break

total_train_time = time.perf_counter() - train_start
peak_mem_mb = (torch.cuda.max_memory_allocated() / 1e6) if DEVICE.type == "cuda" else float("nan")

model.load_state_dict(torch.load(weights_path, map_location=DEVICE))
model.eval()

# ============================================================
# SAVE TRAINING CONVERGENCE CSV
# ============================================================
w_center_dim_history = np.array(w_center_history) * W
df_hist = pd.DataFrame({
    "epoch": epoch_history,
    "total_loss": loss_history,
    "pde_loss": pde_history,
    "bc_loss": bc_history,
    "moment_loss": moment_history,
    "w_center_nd": w_center_history,
    "w_center_dim": w_center_dim_history,
    "epoch_time_s": epoch_time_history,
})
df_hist.to_csv(os.path.join(OUT_DIR, "training_convergence.csv"), index=False)
print("Saved training_convergence.csv")

# ============================================================
# ============================================================
cost_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "device": str(DEVICE),
    "fd_step_h": FD_STEP,
    "epochs_run": len(epoch_history),
    "epochs_to_early_stop": early_stop_epoch,  # renamed from epochs_to_convergence: this marks when the
    "mean_time_per_epoch_s": float(np.mean(epoch_time_history)),
    "median_time_per_epoch_s": float(np.median(epoch_time_history)),
    "total_training_time_s": total_train_time,
    "peak_gpu_memory_mb": peak_mem_mb,
    "n_trainable_params": sum(p.numel() for p in model.parameters() if p.requires_grad),
}
with open(os.path.join(OUT_DIR, "cost_metrics.json"), "w") as f:
    json.dump(cost_metrics, f, indent=2)
print("Saved cost_metrics.json:", cost_metrics)

# ============================================================
# POST-PROCESSING: full-field predictions via FD stencils
# ============================================================
def compute_full_fields():
    x = X_grid[:, 0:1]
    y = X_grid[:, 1:2]
    with torch.no_grad():
        w_c, w_xx, w_yy, w_xy = fd_full_field_derivatives(x, y, FD_STEP)
        Mx = -(w_xx + nu_nd * w_yy)
        My = -(w_yy + nu_nd * w_xx)
        Mxy = -(1 - nu_nd) * w_xy

    w_grid = w_c.cpu().numpy().reshape(ny, nx)
    Mx_grid = Mx.cpu().numpy().reshape(ny, nx)
    My_grid = My.cpu().numpy().reshape(ny, nx)
    Mxy_grid = Mxy.cpu().numpy().reshape(ny, nx)
    return w_grid, Mx_grid, My_grid, Mxy_grid


w_pred_nd, Mx_pred_nd, My_pred_nd, Mxy_pred_nd = compute_full_fields()

w_anal_nd = analytical_w_nd(X_nd, Y_nd)
Mx_anal_nd = analytical_Mx_nd(X_nd, Y_nd)
My_anal_nd = analytical_My_nd(X_nd, Y_nd)
Mxy_anal_nd = analytical_Mxy_nd(X_nd, Y_nd)

w_dim_pred, w_dim_anal = w_pred_nd * W, w_anal_nd * W
moment_scale = D_dim * W / L ** 2
Mx_dim_pred, Mx_dim_anal = Mx_pred_nd * moment_scale, Mx_anal_nd * moment_scale
My_dim_pred, My_dim_anal = My_pred_nd * moment_scale, My_anal_nd * moment_scale
Mxy_dim_pred, Mxy_dim_anal = Mxy_pred_nd * moment_scale, Mxy_anal_nd * moment_scale

w_signed_error = w_dim_pred - w_dim_anal
w_error = np.abs(w_signed_error)
Mx_error = np.abs(Mx_dim_pred - Mx_dim_anal)
My_error = np.abs(My_dim_pred - My_dim_anal)
Mxy_error = np.abs(Mxy_dim_pred - Mxy_dim_anal)

# ============================================================
# ============================================================
iy, ix = ny // 2, nx // 2
w_pred_center, w_anal_center = w_dim_pred[iy, ix], w_dim_anal[iy, ix]
center_abs_err = abs(w_pred_center - w_anal_center)
center_pct_err = center_abs_err / abs(w_anal_center) * 100

L2_error_w = np.sqrt(np.mean((w_dim_pred - w_dim_anal) ** 2))
rel_L2_error_w = np.linalg.norm(w_dim_pred - w_dim_anal) / np.linalg.norm(w_dim_anal)

error_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "center_deflection_pred_m": float(w_pred_center),
    "center_deflection_anal_m": float(w_anal_center),
    "center_abs_error_m": float(center_abs_err),
    "center_pct_error": float(center_pct_err),
    "w_max_abs_error_m": float(np.max(w_error)),
    "w_rms_error_m": float(np.sqrt(np.mean(w_error ** 2))),
    "w_L2_error_m": float(L2_error_w),
    "w_relative_L2_error": float(rel_L2_error_w),
    "Mx_max_abs_error_Nm": float(np.max(Mx_error)),
    "Mx_rms_error_Nm": float(np.sqrt(np.mean(Mx_error ** 2))),
    "My_max_abs_error_Nm": float(np.max(My_error)),
    "My_rms_error_Nm": float(np.sqrt(np.mean(My_error ** 2))),
    "Mxy_max_abs_error_Nm": float(np.max(Mxy_error)),
    "Mxy_rms_error_Nm": float(np.sqrt(np.mean(Mxy_error ** 2))),
}
with open(os.path.join(OUT_DIR, "error_metrics.json"), "w") as f:
    json.dump(error_metrics, f, indent=2)
print("Saved error_metrics.json:", error_metrics)

df_field = pd.DataFrame({
    "x_nd": X_nd.ravel(), "y_nd": Y_nd.ravel(),
    "w_dim_anal": w_dim_anal.ravel(), "w_dim_pred": w_dim_pred.ravel(),
    "Mx_dim_anal": Mx_dim_anal.ravel(), "Mx_dim_pred": Mx_dim_pred.ravel(),
    "My_dim_anal": My_dim_anal.ravel(), "My_dim_pred": My_dim_pred.ravel(),
    "Mxy_dim_anal": Mxy_dim_anal.ravel(), "Mxy_dim_pred": Mxy_dim_pred.ravel(),
})
df_field["w_abs_error"] = np.abs(df_field["w_dim_anal"] - df_field["w_dim_pred"])
df_field["w_rel_error"] = df_field["w_abs_error"] / (np.abs(df_field["w_dim_anal"]) + 1e-12)
df_field.to_csv(os.path.join(OUT_DIR, "field_data.csv"), index=False)
print("Saved field_data.csv")

# ============================================================
# ============================================================
plt.rcParams.update({"figure.dpi": 150})


def save_contour(field, title, cbar_label, fname, cmap=cm.jet):
    plt.figure(figsize=(7, 6))
    cf = plt.contourf(X_nd, Y_nd, field, levels=50, cmap=cmap)
    plt.colorbar(cf, label=cbar_label)
    plt.title(title, fontstyle="italic", fontsize=13)
    plt.xlabel("x/a", fontstyle="italic")
    plt.ylabel("y/b", fontstyle="italic")
    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, fname))
    plt.close()


plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], df_hist["w_center_dim"], linewidth=2, label=f"{MODEL_NAME} Prediction")
plt.axhline(y=w_anal_center, linestyle="--", linewidth=2, color="k", label="Analytical")
plt.xlabel("Epoch"); plt.ylabel("Center Deflection w (m)")
plt.title("Convergence of Center Deflection")
plt.legend(); plt.grid(True, linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Center_Deflection.png")); plt.close()

plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], np.abs(df_hist["w_center_dim"] - w_anal_center), linewidth=2)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Absolute Error at Center (m)")
plt.title("Convergence Error (Log Scale)")
plt.grid(True, which="both", linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Error_Log_Scale.png")); plt.close()

plt.figure(figsize=(8, 6))
plt.plot(df_hist["epoch"], df_hist["total_loss"], label="Total Loss")
plt.plot(df_hist["epoch"], df_hist["pde_loss"], label="PDE Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["bc_loss"], label="BC Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["moment_loss"], label="Moment Loss", alpha=0.7)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Training Loss Curve")
plt.legend(); plt.grid(True, which="both", ls="--", alpha=0.7); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Training_Loss_Curve.png")); plt.close()

save_contour(w_dim_anal, "Analytical Plate Deflection", "Deflection (m)",
             "Analytical_Plate_Deflection.png")
save_contour(w_dim_pred, f"{MODEL_NAME} Predicted Plate Deflection", "Deflection (m)",
             "FDM-PINN_Predicted_Plate_Deflection.png")
save_contour(w_error, "Absolute Error in Plate Deflection", "|w_pred - w_anal| (m)",
             "Absolute_Error_in_Plate_Deflection.png", cmap=cm.inferno)
save_contour(w_signed_error, "Signed Error w_pred - w_anal", "w_pred - w_anal (m)",
             "Absolute_Error_w_pred_-_w_anal.png", cmap="coolwarm")

save_contour(Mx_dim_anal, "Analytical Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x_Analytical.png")
save_contour(Mx_dim_pred, "Predicted Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x.png")
save_contour(Mx_error, "Absolute Error in $M_x$", "|Mx_pred - Mx_anal| (N·m/m)",
             "Absolute_Error_in_M_x.png", cmap="hot")

save_contour(My_dim_anal, "Analytical Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y_Analytical.png")
save_contour(My_dim_pred, "Predicted Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y.png")
save_contour(My_error, "Absolute Error in $M_y$", "|My_pred - My_anal| (N·m/m)",
             "Absolute_Error_in_M_y.png", cmap="hot")

save_contour(Mxy_dim_anal, "Analytical Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy_Analytical.png")
save_contour(Mxy_dim_pred, "Predicted Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy.png")
save_contour(Mxy_error, "Absolute Error in $M_{xy}$", "|Mxy_pred - Mxy_anal| (N·m/m)",
             "Absolute_Error_in_M_xy.png", cmap="hot")

plt.figure(figsize=(6, 6))
plt.scatter(w_dim_anal.ravel(), w_dim_pred.ravel(), s=8, alpha=0.5)
lims = [min(w_dim_anal.min(), w_dim_pred.min()), max(w_dim_anal.max(), w_dim_pred.max())]
plt.plot(lims, lims, "k--", linewidth=1, label="Perfect agreement")
plt.xlabel("Analytical displacement w (m)", fontstyle="italic")
plt.ylabel("Predicted displacement w (m)", fontstyle="italic")
plt.title("Analytical vs Predicted Displacement", fontstyle="italic")
plt.legend(); plt.grid(True); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Analytical_vs_Predicted_Displacement.png")); plt.close()

err_fft = np.fft.fftshift(np.fft.fft2(w_error))
err_fft_mag = np.log1p(np.abs(err_fft))
plt.figure(figsize=(7, 6))
plt.imshow(err_fft_mag, cmap="magma", extent=[-nx // 2, nx // 2, -ny // 2, ny // 2])
plt.colorbar(label="log(1 + |FFT error|)")
plt.title(f"Error Spectrum (FFT) — {LOADING_CASE}")
plt.xlabel("kx"); plt.ylabel("ky")
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Error_Spectrum_FFT.png")); plt.close()

fig, axes = plt.subplots(4, 3, figsize=(16, 18))
rows = [
    ("Displacement w (m)", w_dim_anal, w_dim_pred, w_error),
    ("Moment Mx (N·m)", Mx_dim_anal, Mx_dim_pred, Mx_error),
    ("Moment My (N·m)", My_dim_anal, My_dim_pred, My_error),
    ("Moment Mxy (N·m)", Mxy_dim_anal, Mxy_dim_pred, Mxy_error),
]
for r, (label, anal, pred, err) in enumerate(rows):
    for c, (title, field, cmap) in enumerate([
        (f"Analytical {label}", anal, cm.jet),
        (f"Predicted {label}", pred, cm.jet),
        (f"Absolute Error {label}", err, cm.inferno),
    ]):
        ax = axes[r, c]
        cf = ax.contourf(X_nd, Y_nd, field, levels=20, cmap=cmap)
        fig.colorbar(cf, ax=ax)
        ax.set_title(title, fontsize=10)
        ax.set_xlabel("x/a"); ax.set_ylabel("y/b")
plt.suptitle(f"{MODEL_NAME} — {LOADING_CASE} loading, SS-SS-SS-SS (summary grid)", fontsize=14)
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "_summary_grid_all_fields.png")); plt.close()

print(f"\nAll outputs written to: {os.path.abspath(OUT_DIR)}")
print(f"Total training time: {total_train_time:.2f}s | "
      f"Epochs to early stop: {early_stop_epoch} | "
      f"FD step h: {FD_STEP}")
```

```python
# ---------------------------------------------------------------------------
# Physics-Informed Neural Network for Kirchhoff Plate Bending
# Architecture : FDM-PINN
# BC Type      : SSSS (Simply Supported: w=0 and Mn=0 on all edges)
# Loading      : N/A - Epoch selection study
# Authors      : Vivek Kumar Yadav, Kshitij Kumar Yadav, Rajesh Kumar
# Institution  : Dept. of Civil Engineering, IIT (BHU) Varanasi
# ---------------------------------------------------------------------------


import os
import json
import time
import random

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from matplotlib import cm

# ============================================================
# CONFIG
# ============================================================
MODEL_NAME = "FDM_PINN"
LOADING_CASE = "sinusoidal"

OUT_DIR = os.path.join("outputs", MODEL_NAME)
FIG_DIR = os.path.join(OUT_DIR, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

SEED = 42
MAX_EPOCHS = 40000     # matches original notebook's training budget (range(15000)); same PATIENCE=3000
PATIENCE = 3000
PRINT_EVERY = 100
LR_INIT = 1e-3
LR_DECAY_STEPS = 2000
LR_DECAY_RATE = 0.95

FD_STEP = 0.01     # finite-difference step size h (non-dimensional units).

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
DTYPE = torch.float64

# ============================================================
# SEEDING
# ============================================================
np.random.seed(SEED)
random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)

# ============================================================
# ============================================================
real_a, real_b, real_h = 1.0, 1.0, 0.01
real_E, real_nu, real_q0 = 200e9, 0.3, 1000.0

D_dim = real_E * (real_h ** 3) / (12 * (1 - real_nu ** 2))
L = real_a
W = real_q0 * (L ** 4) / D_dim
D_nd = 1.0
a_nd, b_nd, q0_nd, nu_nd = 1.0, 1.0, 1.0, real_nu

# ============================================================
# DOMAIN
# ============================================================
nx, ny = 51, 51
x_nd = np.linspace(0, a_nd, nx)
y_nd = np.linspace(0, b_nd, ny)
X_nd, Y_nd = np.meshgrid(x_nd, y_nd)
grid_points_nd = np.vstack([X_nd.ravel(), Y_nd.ravel()]).T

X_grid = torch.tensor(grid_points_nd, dtype=DTYPE, device=DEVICE)

center_x, center_y = a_nd / 2, b_nd / 2
center_idx = int(np.argmin((grid_points_nd[:, 0] - center_x) ** 2 +
                            (grid_points_nd[:, 1] - center_y) ** 2))

boundary_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
interior_indices = np.setdiff1d(np.arange(nx * ny), boundary_indices)

edge_x_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
edge_y_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
]))

bc_idx = torch.tensor(boundary_indices, dtype=torch.long, device=DEVICE)
int_idx = torch.tensor(interior_indices, dtype=torch.long, device=DEVICE)
edge_x_idx = torch.tensor(edge_x_indices, dtype=torch.long, device=DEVICE)
edge_y_idx = torch.tensor(edge_y_indices, dtype=torch.long, device=DEVICE)

X_int = X_grid[int_idx]        # (N_int, 2) interior collocation points
X_bc = X_grid[bc_idx]          # (N_bc, 2) boundary points (for w=0)
X_edge_x = X_grid[edge_x_idx]  # (N_ex, 2) x=0,a edge points (for Mx=0)
X_edge_y = X_grid[edge_y_idx]  # (N_ey, 2) y=0,b edge points (for My=0)

# ============================================================
# ============================================================
def analytical_w_nd(x, y):
    denom = np.pi ** 4 * D_nd * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mx_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (1 / a_nd ** 2 + nu_nd / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_My_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (nu_nd / a_nd ** 2 + 1 / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mxy_nd(x, y):
    denom = ((1 / a_nd ** 2) + (1 / b_nd ** 2)) ** 2
    return (-q0_nd * (1 - nu_nd) / (np.pi ** 2 * denom * a_nd * b_nd)) * np.cos(np.pi * x / a_nd) * np.cos(np.pi * y / b_nd)


def sinusoidal_load_nd(x, y):
    return q0_nd * torch.sin(np.pi * x / a_nd) * torch.sin(np.pi * y / b_nd)


# ============================================================
# ============================================================
class PlatePINN(nn.Module):
    """
    w(x,y) = D(x,y) * scale * NN(x,y), where D(x,y) = x(a-x)y(b-y) / [(a/2)^2(b/2)^2]
    is a smooth distance function that is EXACTLY zero on all four edges
    (x=0, x=a, y=0, y=b) and positive inside the domain.

    This hard-enforces the essential SS boundary condition w=0 by construction,
    for ANY choice of network weights -- the optimizer can no longer trade off
    boundary satisfaction against the PDE/moment residuals, which was causing
    a visible, non-shrinking violation of w=0 at the edges (and a systematic
    bias in the interior solution) under the previous soft bc_loss penalty.
    Normalizing by (a/2)^2*(b/2)^2 keeps D(x,y)=1 at the plate center, so it
    doesn't change the overall output scale.
    """
    def __init__(self, hidden=64, n_layers=3):
        super().__init__()
        layers = []
        in_dim = 2
        for _ in range(n_layers):
            linear = nn.Linear(in_dim, hidden, dtype=DTYPE)
            nn.init.xavier_normal_(linear.weight)   # matches original kernel_initializer='glorot_normal'
            nn.init.zeros_(linear.bias)              # matches keras Dense default bias_initializer='zeros'
            layers.append(linear)
            layers.append(nn.SiLU())
            in_dim = hidden
        self.hidden = nn.Sequential(*layers)
        self.out = nn.Linear(in_dim, 1, dtype=DTYPE)
        nn.init.zeros_(self.out.weight)   # matches original output_layer kernel_initializer='zeros'
        nn.init.zeros_(self.out.bias)     # network output is exactly 0 at init, controlled by `scale` below
        self.scale = nn.Parameter(torch.tensor(0.01, dtype=DTYPE))

    def forward(self, x, y):
        h = torch.cat([x / a_nd, y / b_nd], dim=1)
        h = self.hidden(h)
        n_out = self.out(h)
        dist = (x * (a_nd - x) * y * (b_nd - y)) / ((a_nd / 2) ** 2 * (b_nd / 2) ** 2)
        return self.scale * dist * n_out


model = PlatePINN().to(DEVICE)

optimizer = torch.optim.Adam(model.parameters(), lr=LR_INIT)
scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer, lr_lambda=lambda step: LR_DECAY_RATE ** (step / LR_DECAY_STEPS)
)


# ============================================================
# FINITE-DIFFERENCE STENCIL HELPERS
# ============================================================
def batched_stencil_eval(x, y, offsets):
    """
    Evaluate the model at (x+dx, y+dy) for every (dx,dy) in `offsets`,
    all in a single stacked forward pass.

    x, y : (N,1) tensors
    offsets : list of (dx, dy) tuples
    returns : dict mapping (dx,dy) -> (N,1) tensor of model outputs
    """
    n = x.shape[0]
    xs = torch.cat([x + dx for dx, dy in offsets], dim=0)
    ys = torch.cat([y + dy for dx, dy in offsets], dim=0)
    w_all = model(xs, ys)
    return {off: w_all[i * n:(i + 1) * n] for i, off in enumerate(offsets)}


def fd_interior_derivatives(x, y, h):
    """Return w_c, w_xxxx, w_yyyy, w_xxyy at interior points via FD stencils."""
    offsets = [
        (0, 0),
        (-2 * h, 0), (-h, 0), (h, 0), (2 * h, 0),      # x-line (for xxxx)
        (0, -2 * h), (0, -h), (0, h), (0, 2 * h),      # y-line (for yyyy)
        (h, h), (h, -h), (-h, h), (-h, -h),            # diagonals (for xxyy)
    ]
    v = batched_stencil_eval(x, y, offsets)

    w_c = v[(0, 0)]
    w_xxxx = (v[(-2 * h, 0)] - 4 * v[(-h, 0)] + 6 * w_c - 4 * v[(h, 0)] + v[(2 * h, 0)]) / h ** 4
    w_yyyy = (v[(0, -2 * h)] - 4 * v[(0, -h)] + 6 * w_c - 4 * v[(0, h)] + v[(0, 2 * h)]) / h ** 4
    w_xxyy = (4 * w_c
              - 2 * (v[(h, 0)] + v[(-h, 0)] + v[(0, h)] + v[(0, -h)])
              + v[(h, h)] + v[(h, -h)] + v[(-h, h)] + v[(-h, -h)]) / h ** 4
    return w_c, w_xxxx, w_yyyy, w_xxyy


def fd_boundary_moments(x, y, h):
    """Return w_c, w_xx, w_yy at boundary points via 3-point FD stencils.

    Note: the central stencil evaluates the network at x-h/x+h (or y-h/y+h),
    which for points exactly on the domain edge means one of the two
    evaluation points lies slightly outside [0,a]x[0,b] (e.g. at x=0 this
    needs w(-h,y)). This is not automatically wrong -- the network is a
    smooth function defined everywhere, not only on the physical domain --
    but it is a methodological choice a reviewer could question. It is
    documented here explicitly as the chosen FDM-PINN stencil treatment for
    this study: central differences, evaluated with a small extension of the
    network beyond the physical edge. A one-sided (forward/backward)
    boundary stencil is a valid alternative if strict domain-only evaluation
    is required; it was not used here so that all boundary derivatives share
    the same (central, 2nd-order accurate) truncation-error behavior as the
    interior stencils."""
    offsets = [(0, 0), (-h, 0), (h, 0), (0, -h), (0, h)]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    return w_c, w_xx, w_yy


def fd_full_field_derivatives(x, y, h):
    """Same as fd_interior_derivatives but also returns w_xy (for Mxy) —
    used only in post-processing over the full grid."""
    offsets = [
        (0, 0),
        (-h, 0), (h, 0), (0, -h), (0, h),
        (h, h), (h, -h), (-h, h), (-h, -h),
    ]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    w_xy = (v[(h, h)] - v[(h, -h)] - v[(-h, h)] + v[(-h, -h)]) / (4 * h ** 2)
    return w_c, w_xx, w_yy, w_xy


# ============================================================
# LOSS (finite-difference residual — replaces autograd chain from Vanilla)
# ============================================================
def compute_loss():
    x_int, y_int = X_int[:, 0:1], X_int[:, 1:2]
    x_bc, y_bc = X_bc[:, 0:1], X_bc[:, 1:2]
    x_ex, y_ex = X_edge_x[:, 0:1], X_edge_x[:, 1:2]
    x_ey, y_ey = X_edge_y[:, 0:1], X_edge_y[:, 1:2]

    _, w_xxxx, w_yyyy, w_xxyy = fd_interior_derivatives(x_int, y_int, FD_STEP)
    q = sinusoidal_load_nd(x_int, y_int)
    pde = w_xxxx + 2 * w_xxyy + w_yyyy - q
    pde_loss = torch.mean(pde ** 2)

    w_bc = model(x_bc, y_bc)
    bc_loss = torch.mean(w_bc ** 2)

    _, w_xx_ex, w_yy_ex = fd_boundary_moments(x_ex, y_ex, FD_STEP)
    Mx_edge = -(w_xx_ex + nu_nd * w_yy_ex)

    _, w_xx_ey, w_yy_ey = fd_boundary_moments(x_ey, y_ey, FD_STEP)
    My_edge = -(w_yy_ey + nu_nd * w_xx_ey)

    moment_loss = torch.mean(Mx_edge ** 2) + torch.mean(My_edge ** 2)

    total = pde_loss + moment_loss

    with torch.no_grad():
        w_full = model(X_grid[:, 0:1], X_grid[:, 1:2])

    return total, pde_loss, bc_loss, moment_loss, w_full


# ============================================================
# ============================================================
epoch_history, loss_history, pde_history, bc_history, moment_history = [], [], [], [], []
w_center_history, epoch_time_history = [], []

best_loss = float("inf")
wait = 0
weights_path = os.path.join(OUT_DIR, "best_weights.pt")

if DEVICE.type == "cuda":
    torch.cuda.reset_peak_memory_stats()

train_start = time.perf_counter()
early_stop_epoch = MAX_EPOCHS  # epoch index where early stopping triggered; MAX_EPOCHS if it never did. This is an early-stopping bookkeeping value, NOT a proven mathematical convergence criterion.

for epoch in range(MAX_EPOCHS):
    epoch_t0 = time.perf_counter()

    optimizer.zero_grad()
    total, pde, bc, moment, w_full = compute_loss()
    total.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)  # matches original Adam(clipnorm=1.0)
    optimizer.step()
    scheduler.step()

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    epoch_dt = time.perf_counter() - epoch_t0

    loss_val = total.item()
    w_center_val = w_full[center_idx, 0].item()

    epoch_history.append(epoch)
    loss_history.append(loss_val)
    pde_history.append(pde.item())
    bc_history.append(bc.item())
    moment_history.append(moment.item())
    w_center_history.append(w_center_val)
    epoch_time_history.append(epoch_dt)

    if epoch % PRINT_EVERY == 0:
        print(f"Epoch {epoch} Loss {loss_val:.6e} Center {w_center_val:.6e} "
              f"({epoch_dt*1000:.2f} ms/epoch)")

    if loss_val < best_loss:
        best_loss = loss_val
        wait = 0
        torch.save(model.state_dict(), weights_path)
    else:
        wait += 1
        if wait >= PATIENCE:
            print(f"Early stopping triggered at epoch {epoch}")
            early_stop_epoch = epoch
            break

total_train_time = time.perf_counter() - train_start
peak_mem_mb = (torch.cuda.max_memory_allocated() / 1e6) if DEVICE.type == "cuda" else float("nan")

model.load_state_dict(torch.load(weights_path, map_location=DEVICE))
model.eval()

# ============================================================
# SAVE TRAINING CONVERGENCE CSV
# ============================================================
w_center_dim_history = np.array(w_center_history) * W
df_hist = pd.DataFrame({
    "epoch": epoch_history,
    "total_loss": loss_history,
    "pde_loss": pde_history,
    "bc_loss": bc_history,
    "moment_loss": moment_history,
    "w_center_nd": w_center_history,
    "w_center_dim": w_center_dim_history,
    "epoch_time_s": epoch_time_history,
})
df_hist.to_csv(os.path.join(OUT_DIR, "training_convergence.csv"), index=False)
print("Saved training_convergence.csv")

# ============================================================
# ============================================================
cost_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "device": str(DEVICE),
    "fd_step_h": FD_STEP,
    "epochs_run": len(epoch_history),
    "epochs_to_early_stop": early_stop_epoch,  # renamed from epochs_to_convergence: this marks when the
    "mean_time_per_epoch_s": float(np.mean(epoch_time_history)),
    "median_time_per_epoch_s": float(np.median(epoch_time_history)),
    "total_training_time_s": total_train_time,
    "peak_gpu_memory_mb": peak_mem_mb,
    "n_trainable_params": sum(p.numel() for p in model.parameters() if p.requires_grad),
}
with open(os.path.join(OUT_DIR, "cost_metrics.json"), "w") as f:
    json.dump(cost_metrics, f, indent=2)
print("Saved cost_metrics.json:", cost_metrics)

# ============================================================
# POST-PROCESSING: full-field predictions via FD stencils
# ============================================================
def compute_full_fields():
    x = X_grid[:, 0:1]
    y = X_grid[:, 1:2]
    with torch.no_grad():
        w_c, w_xx, w_yy, w_xy = fd_full_field_derivatives(x, y, FD_STEP)
        Mx = -(w_xx + nu_nd * w_yy)
        My = -(w_yy + nu_nd * w_xx)
        Mxy = -(1 - nu_nd) * w_xy

    w_grid = w_c.cpu().numpy().reshape(ny, nx)
    Mx_grid = Mx.cpu().numpy().reshape(ny, nx)
    My_grid = My.cpu().numpy().reshape(ny, nx)
    Mxy_grid = Mxy.cpu().numpy().reshape(ny, nx)
    return w_grid, Mx_grid, My_grid, Mxy_grid


w_pred_nd, Mx_pred_nd, My_pred_nd, Mxy_pred_nd = compute_full_fields()

w_anal_nd = analytical_w_nd(X_nd, Y_nd)
Mx_anal_nd = analytical_Mx_nd(X_nd, Y_nd)
My_anal_nd = analytical_My_nd(X_nd, Y_nd)
Mxy_anal_nd = analytical_Mxy_nd(X_nd, Y_nd)

w_dim_pred, w_dim_anal = w_pred_nd * W, w_anal_nd * W
moment_scale = D_dim * W / L ** 2
Mx_dim_pred, Mx_dim_anal = Mx_pred_nd * moment_scale, Mx_anal_nd * moment_scale
My_dim_pred, My_dim_anal = My_pred_nd * moment_scale, My_anal_nd * moment_scale
Mxy_dim_pred, Mxy_dim_anal = Mxy_pred_nd * moment_scale, Mxy_anal_nd * moment_scale

w_signed_error = w_dim_pred - w_dim_anal
w_error = np.abs(w_signed_error)
Mx_error = np.abs(Mx_dim_pred - Mx_dim_anal)
My_error = np.abs(My_dim_pred - My_dim_anal)
Mxy_error = np.abs(Mxy_dim_pred - Mxy_dim_anal)

# ============================================================
# ============================================================
iy, ix = ny // 2, nx // 2
w_pred_center, w_anal_center = w_dim_pred[iy, ix], w_dim_anal[iy, ix]
center_abs_err = abs(w_pred_center - w_anal_center)
center_pct_err = center_abs_err / abs(w_anal_center) * 100

L2_error_w = np.sqrt(np.mean((w_dim_pred - w_dim_anal) ** 2))
rel_L2_error_w = np.linalg.norm(w_dim_pred - w_dim_anal) / np.linalg.norm(w_dim_anal)

error_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "center_deflection_pred_m": float(w_pred_center),
    "center_deflection_anal_m": float(w_anal_center),
    "center_abs_error_m": float(center_abs_err),
    "center_pct_error": float(center_pct_err),
    "w_max_abs_error_m": float(np.max(w_error)),
    "w_rms_error_m": float(np.sqrt(np.mean(w_error ** 2))),
    "w_L2_error_m": float(L2_error_w),
    "w_relative_L2_error": float(rel_L2_error_w),
    "Mx_max_abs_error_Nm": float(np.max(Mx_error)),
    "Mx_rms_error_Nm": float(np.sqrt(np.mean(Mx_error ** 2))),
    "My_max_abs_error_Nm": float(np.max(My_error)),
    "My_rms_error_Nm": float(np.sqrt(np.mean(My_error ** 2))),
    "Mxy_max_abs_error_Nm": float(np.max(Mxy_error)),
    "Mxy_rms_error_Nm": float(np.sqrt(np.mean(Mxy_error ** 2))),
}
with open(os.path.join(OUT_DIR, "error_metrics.json"), "w") as f:
    json.dump(error_metrics, f, indent=2)
print("Saved error_metrics.json:", error_metrics)

df_field = pd.DataFrame({
    "x_nd": X_nd.ravel(), "y_nd": Y_nd.ravel(),
    "w_dim_anal": w_dim_anal.ravel(), "w_dim_pred": w_dim_pred.ravel(),
    "Mx_dim_anal": Mx_dim_anal.ravel(), "Mx_dim_pred": Mx_dim_pred.ravel(),
    "My_dim_anal": My_dim_anal.ravel(), "My_dim_pred": My_dim_pred.ravel(),
    "Mxy_dim_anal": Mxy_dim_anal.ravel(), "Mxy_dim_pred": Mxy_dim_pred.ravel(),
})
df_field["w_abs_error"] = np.abs(df_field["w_dim_anal"] - df_field["w_dim_pred"])
df_field["w_rel_error"] = df_field["w_abs_error"] / (np.abs(df_field["w_dim_anal"]) + 1e-12)
df_field.to_csv(os.path.join(OUT_DIR, "field_data.csv"), index=False)
print("Saved field_data.csv")

# ============================================================
# ============================================================
plt.rcParams.update({"figure.dpi": 150})


def save_contour(field, title, cbar_label, fname, cmap=cm.jet):
    plt.figure(figsize=(7, 6))
    cf = plt.contourf(X_nd, Y_nd, field, levels=50, cmap=cmap)
    plt.colorbar(cf, label=cbar_label)
    plt.title(title, fontstyle="italic", fontsize=13)
    plt.xlabel("x/a", fontstyle="italic")
    plt.ylabel("y/b", fontstyle="italic")
    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, fname))
    plt.close()


plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], df_hist["w_center_dim"], linewidth=2, label=f"{MODEL_NAME} Prediction")
plt.axhline(y=w_anal_center, linestyle="--", linewidth=2, color="k", label="Analytical")
plt.xlabel("Epoch"); plt.ylabel("Center Deflection w (m)")
plt.title("Convergence of Center Deflection")
plt.legend(); plt.grid(True, linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Center_Deflection.png")); plt.close()

plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], np.abs(df_hist["w_center_dim"] - w_anal_center), linewidth=2)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Absolute Error at Center (m)")
plt.title("Convergence Error (Log Scale)")
plt.grid(True, which="both", linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Error_Log_Scale.png")); plt.close()

plt.figure(figsize=(8, 6))
plt.plot(df_hist["epoch"], df_hist["total_loss"], label="Total Loss")
plt.plot(df_hist["epoch"], df_hist["pde_loss"], label="PDE Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["bc_loss"], label="BC Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["moment_loss"], label="Moment Loss", alpha=0.7)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Training Loss Curve")
plt.legend(); plt.grid(True, which="both", ls="--", alpha=0.7); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Training_Loss_Curve.png")); plt.close()

save_contour(w_dim_anal, "Analytical Plate Deflection", "Deflection (m)",
             "Analytical_Plate_Deflection.png")
save_contour(w_dim_pred, f"{MODEL_NAME} Predicted Plate Deflection", "Deflection (m)",
             "FDM-PINN_Predicted_Plate_Deflection.png")
save_contour(w_error, "Absolute Error in Plate Deflection", "|w_pred - w_anal| (m)",
             "Absolute_Error_in_Plate_Deflection.png", cmap=cm.inferno)
save_contour(w_signed_error, "Signed Error w_pred - w_anal", "w_pred - w_anal (m)",
             "Absolute_Error_w_pred_-_w_anal.png", cmap="coolwarm")

save_contour(Mx_dim_anal, "Analytical Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x_Analytical.png")
save_contour(Mx_dim_pred, "Predicted Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x.png")
save_contour(Mx_error, "Absolute Error in $M_x$", "|Mx_pred - Mx_anal| (N·m/m)",
             "Absolute_Error_in_M_x.png", cmap="hot")

save_contour(My_dim_anal, "Analytical Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y_Analytical.png")
save_contour(My_dim_pred, "Predicted Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y.png")
save_contour(My_error, "Absolute Error in $M_y$", "|My_pred - My_anal| (N·m/m)",
             "Absolute_Error_in_M_y.png", cmap="hot")

save_contour(Mxy_dim_anal, "Analytical Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy_Analytical.png")
save_contour(Mxy_dim_pred, "Predicted Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy.png")
save_contour(Mxy_error, "Absolute Error in $M_{xy}$", "|Mxy_pred - Mxy_anal| (N·m/m)",
             "Absolute_Error_in_M_xy.png", cmap="hot")

plt.figure(figsize=(6, 6))
plt.scatter(w_dim_anal.ravel(), w_dim_pred.ravel(), s=8, alpha=0.5)
lims = [min(w_dim_anal.min(), w_dim_pred.min()), max(w_dim_anal.max(), w_dim_pred.max())]
plt.plot(lims, lims, "k--", linewidth=1, label="Perfect agreement")
plt.xlabel("Analytical displacement w (m)", fontstyle="italic")
plt.ylabel("Predicted displacement w (m)", fontstyle="italic")
plt.title("Analytical vs Predicted Displacement", fontstyle="italic")
plt.legend(); plt.grid(True); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Analytical_vs_Predicted_Displacement.png")); plt.close()

err_fft = np.fft.fftshift(np.fft.fft2(w_error))
err_fft_mag = np.log1p(np.abs(err_fft))
plt.figure(figsize=(7, 6))
plt.imshow(err_fft_mag, cmap="magma", extent=[-nx // 2, nx // 2, -ny // 2, ny // 2])
plt.colorbar(label="log(1 + |FFT error|)")
plt.title(f"Error Spectrum (FFT) — {LOADING_CASE}")
plt.xlabel("kx"); plt.ylabel("ky")
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Error_Spectrum_FFT.png")); plt.close()

fig, axes = plt.subplots(4, 3, figsize=(16, 18))
rows = [
    ("Displacement w (m)", w_dim_anal, w_dim_pred, w_error),
    ("Moment Mx (N·m)", Mx_dim_anal, Mx_dim_pred, Mx_error),
    ("Moment My (N·m)", My_dim_anal, My_dim_pred, My_error),
    ("Moment Mxy (N·m)", Mxy_dim_anal, Mxy_dim_pred, Mxy_error),
]
for r, (label, anal, pred, err) in enumerate(rows):
    for c, (title, field, cmap) in enumerate([
        (f"Analytical {label}", anal, cm.jet),
        (f"Predicted {label}", pred, cm.jet),
        (f"Absolute Error {label}", err, cm.inferno),
    ]):
        ax = axes[r, c]
        cf = ax.contourf(X_nd, Y_nd, field, levels=20, cmap=cmap)
        fig.colorbar(cf, ax=ax)
        ax.set_title(title, fontsize=10)
        ax.set_xlabel("x/a"); ax.set_ylabel("y/b")
plt.suptitle(f"{MODEL_NAME} — {LOADING_CASE} loading, SS-SS-SS-SS (summary grid)", fontsize=14)
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "_summary_grid_all_fields.png")); plt.close()

print(f"\nAll outputs written to: {os.path.abspath(OUT_DIR)}")
print(f"Total training time: {total_train_time:.2f}s | "
      f"Epochs to early stop: {early_stop_epoch} | "
      f"FD step h: {FD_STEP}")
```

```python
# ---------------------------------------------------------------------------
# Physics-Informed Neural Network for Kirchhoff Plate Bending
# Architecture : FDM-PINN
# BC Type      : SSSS (Simply Supported: w=0 and Mn=0 on all edges)
# Loading      : N/A - Epoch selection study
# Authors      : Vivek Kumar Yadav, Kshitij Kumar Yadav, Rajesh Kumar
# Institution  : Dept. of Civil Engineering, IIT (BHU) Varanasi
# ---------------------------------------------------------------------------


import os
import json
import time
import random

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from matplotlib import cm

# ============================================================
# CONFIG
# ============================================================
MODEL_NAME = "FDM_PINN"
LOADING_CASE = "sinusoidal"

OUT_DIR = os.path.join("outputs", MODEL_NAME)
FIG_DIR = os.path.join(OUT_DIR, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

SEED = 42
MAX_EPOCHS = 50000     # matches original notebook's training budget (range(15000)); same PATIENCE=3000
PATIENCE = 3000
PRINT_EVERY = 100
LR_INIT = 1e-3
LR_DECAY_STEPS = 2000
LR_DECAY_RATE = 0.95

FD_STEP = 0.01     # finite-difference step size h (non-dimensional units).

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
DTYPE = torch.float64

# ============================================================
# SEEDING
# ============================================================
np.random.seed(SEED)
random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)

# ============================================================
# ============================================================
real_a, real_b, real_h = 1.0, 1.0, 0.01
real_E, real_nu, real_q0 = 200e9, 0.3, 1000.0

D_dim = real_E * (real_h ** 3) / (12 * (1 - real_nu ** 2))
L = real_a
W = real_q0 * (L ** 4) / D_dim
D_nd = 1.0
a_nd, b_nd, q0_nd, nu_nd = 1.0, 1.0, 1.0, real_nu

# ============================================================
# DOMAIN
# ============================================================
nx, ny = 51, 51
x_nd = np.linspace(0, a_nd, nx)
y_nd = np.linspace(0, b_nd, ny)
X_nd, Y_nd = np.meshgrid(x_nd, y_nd)
grid_points_nd = np.vstack([X_nd.ravel(), Y_nd.ravel()]).T

X_grid = torch.tensor(grid_points_nd, dtype=DTYPE, device=DEVICE)

center_x, center_y = a_nd / 2, b_nd / 2
center_idx = int(np.argmin((grid_points_nd[:, 0] - center_x) ** 2 +
                            (grid_points_nd[:, 1] - center_y) ** 2))

boundary_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
interior_indices = np.setdiff1d(np.arange(nx * ny), boundary_indices)

edge_x_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
edge_y_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
]))

bc_idx = torch.tensor(boundary_indices, dtype=torch.long, device=DEVICE)
int_idx = torch.tensor(interior_indices, dtype=torch.long, device=DEVICE)
edge_x_idx = torch.tensor(edge_x_indices, dtype=torch.long, device=DEVICE)
edge_y_idx = torch.tensor(edge_y_indices, dtype=torch.long, device=DEVICE)

X_int = X_grid[int_idx]        # (N_int, 2) interior collocation points
X_bc = X_grid[bc_idx]          # (N_bc, 2) boundary points (for w=0)
X_edge_x = X_grid[edge_x_idx]  # (N_ex, 2) x=0,a edge points (for Mx=0)
X_edge_y = X_grid[edge_y_idx]  # (N_ey, 2) y=0,b edge points (for My=0)

# ============================================================
# ============================================================
def analytical_w_nd(x, y):
    denom = np.pi ** 4 * D_nd * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mx_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (1 / a_nd ** 2 + nu_nd / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_My_nd(x, y):
    denom = np.pi ** 2 * (1 / a_nd ** 2 + 1 / b_nd ** 2) ** 2
    return (q0_nd / denom) * (nu_nd / a_nd ** 2 + 1 / b_nd ** 2) * np.sin(np.pi * x / a_nd) * np.sin(np.pi * y / b_nd)


def analytical_Mxy_nd(x, y):
    denom = ((1 / a_nd ** 2) + (1 / b_nd ** 2)) ** 2
    return (-q0_nd * (1 - nu_nd) / (np.pi ** 2 * denom * a_nd * b_nd)) * np.cos(np.pi * x / a_nd) * np.cos(np.pi * y / b_nd)


def sinusoidal_load_nd(x, y):
    return q0_nd * torch.sin(np.pi * x / a_nd) * torch.sin(np.pi * y / b_nd)


# ============================================================
# ============================================================
class PlatePINN(nn.Module):
    """
    w(x,y) = D(x,y) * scale * NN(x,y), where D(x,y) = x(a-x)y(b-y) / [(a/2)^2(b/2)^2]
    is a smooth distance function that is EXACTLY zero on all four edges
    (x=0, x=a, y=0, y=b) and positive inside the domain.

    This hard-enforces the essential SS boundary condition w=0 by construction,
    for ANY choice of network weights -- the optimizer can no longer trade off
    boundary satisfaction against the PDE/moment residuals, which was causing
    a visible, non-shrinking violation of w=0 at the edges (and a systematic
    bias in the interior solution) under the previous soft bc_loss penalty.
    Normalizing by (a/2)^2*(b/2)^2 keeps D(x,y)=1 at the plate center, so it
    doesn't change the overall output scale.
    """
    def __init__(self, hidden=64, n_layers=3):
        super().__init__()
        layers = []
        in_dim = 2
        for _ in range(n_layers):
            linear = nn.Linear(in_dim, hidden, dtype=DTYPE)
            nn.init.xavier_normal_(linear.weight)   # matches original kernel_initializer='glorot_normal'
            nn.init.zeros_(linear.bias)              # matches keras Dense default bias_initializer='zeros'
            layers.append(linear)
            layers.append(nn.SiLU())
            in_dim = hidden
        self.hidden = nn.Sequential(*layers)
        self.out = nn.Linear(in_dim, 1, dtype=DTYPE)
        nn.init.zeros_(self.out.weight)   # matches original output_layer kernel_initializer='zeros'
        nn.init.zeros_(self.out.bias)     # network output is exactly 0 at init, controlled by `scale` below
        self.scale = nn.Parameter(torch.tensor(0.01, dtype=DTYPE))

    def forward(self, x, y):
        h = torch.cat([x / a_nd, y / b_nd], dim=1)
        h = self.hidden(h)
        n_out = self.out(h)
        dist = (x * (a_nd - x) * y * (b_nd - y)) / ((a_nd / 2) ** 2 * (b_nd / 2) ** 2)
        return self.scale * dist * n_out


model = PlatePINN().to(DEVICE)

optimizer = torch.optim.Adam(model.parameters(), lr=LR_INIT)
scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer, lr_lambda=lambda step: LR_DECAY_RATE ** (step / LR_DECAY_STEPS)
)


# ============================================================
# FINITE-DIFFERENCE STENCIL HELPERS
# ============================================================
def batched_stencil_eval(x, y, offsets):
    """
    Evaluate the model at (x+dx, y+dy) for every (dx,dy) in `offsets`,
    all in a single stacked forward pass.

    x, y : (N,1) tensors
    offsets : list of (dx, dy) tuples
    returns : dict mapping (dx,dy) -> (N,1) tensor of model outputs
    """
    n = x.shape[0]
    xs = torch.cat([x + dx for dx, dy in offsets], dim=0)
    ys = torch.cat([y + dy for dx, dy in offsets], dim=0)
    w_all = model(xs, ys)
    return {off: w_all[i * n:(i + 1) * n] for i, off in enumerate(offsets)}


def fd_interior_derivatives(x, y, h):
    """Return w_c, w_xxxx, w_yyyy, w_xxyy at interior points via FD stencils."""
    offsets = [
        (0, 0),
        (-2 * h, 0), (-h, 0), (h, 0), (2 * h, 0),      # x-line (for xxxx)
        (0, -2 * h), (0, -h), (0, h), (0, 2 * h),      # y-line (for yyyy)
        (h, h), (h, -h), (-h, h), (-h, -h),            # diagonals (for xxyy)
    ]
    v = batched_stencil_eval(x, y, offsets)

    w_c = v[(0, 0)]
    w_xxxx = (v[(-2 * h, 0)] - 4 * v[(-h, 0)] + 6 * w_c - 4 * v[(h, 0)] + v[(2 * h, 0)]) / h ** 4
    w_yyyy = (v[(0, -2 * h)] - 4 * v[(0, -h)] + 6 * w_c - 4 * v[(0, h)] + v[(0, 2 * h)]) / h ** 4
    w_xxyy = (4 * w_c
              - 2 * (v[(h, 0)] + v[(-h, 0)] + v[(0, h)] + v[(0, -h)])
              + v[(h, h)] + v[(h, -h)] + v[(-h, h)] + v[(-h, -h)]) / h ** 4
    return w_c, w_xxxx, w_yyyy, w_xxyy


def fd_boundary_moments(x, y, h):
    """Return w_c, w_xx, w_yy at boundary points via 3-point FD stencils.

    Note: the central stencil evaluates the network at x-h/x+h (or y-h/y+h),
    which for points exactly on the domain edge means one of the two
    evaluation points lies slightly outside [0,a]x[0,b] (e.g. at x=0 this
    needs w(-h,y)). This is not automatically wrong -- the network is a
    smooth function defined everywhere, not only on the physical domain --
    but it is a methodological choice a reviewer could question. It is
    documented here explicitly as the chosen FDM-PINN stencil treatment for
    this study: central differences, evaluated with a small extension of the
    network beyond the physical edge. A one-sided (forward/backward)
    boundary stencil is a valid alternative if strict domain-only evaluation
    is required; it was not used here so that all boundary derivatives share
    the same (central, 2nd-order accurate) truncation-error behavior as the
    interior stencils."""
    offsets = [(0, 0), (-h, 0), (h, 0), (0, -h), (0, h)]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    return w_c, w_xx, w_yy


def fd_full_field_derivatives(x, y, h):
    """Same as fd_interior_derivatives but also returns w_xy (for Mxy) —
    used only in post-processing over the full grid."""
    offsets = [
        (0, 0),
        (-h, 0), (h, 0), (0, -h), (0, h),
        (h, h), (h, -h), (-h, h), (-h, -h),
    ]
    v = batched_stencil_eval(x, y, offsets)
    w_c = v[(0, 0)]
    w_xx = (v[(-h, 0)] - 2 * w_c + v[(h, 0)]) / h ** 2
    w_yy = (v[(0, -h)] - 2 * w_c + v[(0, h)]) / h ** 2
    w_xy = (v[(h, h)] - v[(h, -h)] - v[(-h, h)] + v[(-h, -h)]) / (4 * h ** 2)
    return w_c, w_xx, w_yy, w_xy


# ============================================================
# LOSS (finite-difference residual — replaces autograd chain from Vanilla)
# ============================================================
def compute_loss():
    x_int, y_int = X_int[:, 0:1], X_int[:, 1:2]
    x_bc, y_bc = X_bc[:, 0:1], X_bc[:, 1:2]
    x_ex, y_ex = X_edge_x[:, 0:1], X_edge_x[:, 1:2]
    x_ey, y_ey = X_edge_y[:, 0:1], X_edge_y[:, 1:2]

    _, w_xxxx, w_yyyy, w_xxyy = fd_interior_derivatives(x_int, y_int, FD_STEP)
    q = sinusoidal_load_nd(x_int, y_int)
    pde = w_xxxx + 2 * w_xxyy + w_yyyy - q
    pde_loss = torch.mean(pde ** 2)

    w_bc = model(x_bc, y_bc)
    bc_loss = torch.mean(w_bc ** 2)

    _, w_xx_ex, w_yy_ex = fd_boundary_moments(x_ex, y_ex, FD_STEP)
    Mx_edge = -(w_xx_ex + nu_nd * w_yy_ex)

    _, w_xx_ey, w_yy_ey = fd_boundary_moments(x_ey, y_ey, FD_STEP)
    My_edge = -(w_yy_ey + nu_nd * w_xx_ey)

    moment_loss = torch.mean(Mx_edge ** 2) + torch.mean(My_edge ** 2)

    total = pde_loss + moment_loss

    with torch.no_grad():
        w_full = model(X_grid[:, 0:1], X_grid[:, 1:2])

    return total, pde_loss, bc_loss, moment_loss, w_full


# ============================================================
# ============================================================
epoch_history, loss_history, pde_history, bc_history, moment_history = [], [], [], [], []
w_center_history, epoch_time_history = [], []

best_loss = float("inf")
wait = 0
weights_path = os.path.join(OUT_DIR, "best_weights.pt")

if DEVICE.type == "cuda":
    torch.cuda.reset_peak_memory_stats()

train_start = time.perf_counter()
early_stop_epoch = MAX_EPOCHS  # epoch index where early stopping triggered; MAX_EPOCHS if it never did. This is an early-stopping bookkeeping value, NOT a proven mathematical convergence criterion.

for epoch in range(MAX_EPOCHS):
    epoch_t0 = time.perf_counter()

    optimizer.zero_grad()
    total, pde, bc, moment, w_full = compute_loss()
    total.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)  # matches original Adam(clipnorm=1.0)
    optimizer.step()
    scheduler.step()

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    epoch_dt = time.perf_counter() - epoch_t0

    loss_val = total.item()
    w_center_val = w_full[center_idx, 0].item()

    epoch_history.append(epoch)
    loss_history.append(loss_val)
    pde_history.append(pde.item())
    bc_history.append(bc.item())
    moment_history.append(moment.item())
    w_center_history.append(w_center_val)
    epoch_time_history.append(epoch_dt)

    if epoch % PRINT_EVERY == 0:
        print(f"Epoch {epoch} Loss {loss_val:.6e} Center {w_center_val:.6e} "
              f"({epoch_dt*1000:.2f} ms/epoch)")

    if loss_val < best_loss:
        best_loss = loss_val
        wait = 0
        torch.save(model.state_dict(), weights_path)
    else:
        wait += 1
        if wait >= PATIENCE:
            print(f"Early stopping triggered at epoch {epoch}")
            early_stop_epoch = epoch
            break

total_train_time = time.perf_counter() - train_start
peak_mem_mb = (torch.cuda.max_memory_allocated() / 1e6) if DEVICE.type == "cuda" else float("nan")

model.load_state_dict(torch.load(weights_path, map_location=DEVICE))
model.eval()

# ============================================================
# SAVE TRAINING CONVERGENCE CSV
# ============================================================
w_center_dim_history = np.array(w_center_history) * W
df_hist = pd.DataFrame({
    "epoch": epoch_history,
    "total_loss": loss_history,
    "pde_loss": pde_history,
    "bc_loss": bc_history,
    "moment_loss": moment_history,
    "w_center_nd": w_center_history,
    "w_center_dim": w_center_dim_history,
    "epoch_time_s": epoch_time_history,
})
df_hist.to_csv(os.path.join(OUT_DIR, "training_convergence.csv"), index=False)
print("Saved training_convergence.csv")

# ============================================================
# ============================================================
cost_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "device": str(DEVICE),
    "fd_step_h": FD_STEP,
    "epochs_run": len(epoch_history),
    "epochs_to_early_stop": early_stop_epoch,  # renamed from epochs_to_convergence: this marks when the
    "mean_time_per_epoch_s": float(np.mean(epoch_time_history)),
    "median_time_per_epoch_s": float(np.median(epoch_time_history)),
    "total_training_time_s": total_train_time,
    "peak_gpu_memory_mb": peak_mem_mb,
    "n_trainable_params": sum(p.numel() for p in model.parameters() if p.requires_grad),
}
with open(os.path.join(OUT_DIR, "cost_metrics.json"), "w") as f:
    json.dump(cost_metrics, f, indent=2)
print("Saved cost_metrics.json:", cost_metrics)

# ============================================================
# POST-PROCESSING: full-field predictions via FD stencils
# ============================================================
def compute_full_fields():
    x = X_grid[:, 0:1]
    y = X_grid[:, 1:2]
    with torch.no_grad():
        w_c, w_xx, w_yy, w_xy = fd_full_field_derivatives(x, y, FD_STEP)
        Mx = -(w_xx + nu_nd * w_yy)
        My = -(w_yy + nu_nd * w_xx)
        Mxy = -(1 - nu_nd) * w_xy

    w_grid = w_c.cpu().numpy().reshape(ny, nx)
    Mx_grid = Mx.cpu().numpy().reshape(ny, nx)
    My_grid = My.cpu().numpy().reshape(ny, nx)
    Mxy_grid = Mxy.cpu().numpy().reshape(ny, nx)
    return w_grid, Mx_grid, My_grid, Mxy_grid


w_pred_nd, Mx_pred_nd, My_pred_nd, Mxy_pred_nd = compute_full_fields()

w_anal_nd = analytical_w_nd(X_nd, Y_nd)
Mx_anal_nd = analytical_Mx_nd(X_nd, Y_nd)
My_anal_nd = analytical_My_nd(X_nd, Y_nd)
Mxy_anal_nd = analytical_Mxy_nd(X_nd, Y_nd)

w_dim_pred, w_dim_anal = w_pred_nd * W, w_anal_nd * W
moment_scale = D_dim * W / L ** 2
Mx_dim_pred, Mx_dim_anal = Mx_pred_nd * moment_scale, Mx_anal_nd * moment_scale
My_dim_pred, My_dim_anal = My_pred_nd * moment_scale, My_anal_nd * moment_scale
Mxy_dim_pred, Mxy_dim_anal = Mxy_pred_nd * moment_scale, Mxy_anal_nd * moment_scale

w_signed_error = w_dim_pred - w_dim_anal
w_error = np.abs(w_signed_error)
Mx_error = np.abs(Mx_dim_pred - Mx_dim_anal)
My_error = np.abs(My_dim_pred - My_dim_anal)
Mxy_error = np.abs(Mxy_dim_pred - Mxy_dim_anal)

# ============================================================
# ============================================================
iy, ix = ny // 2, nx // 2
w_pred_center, w_anal_center = w_dim_pred[iy, ix], w_dim_anal[iy, ix]
center_abs_err = abs(w_pred_center - w_anal_center)
center_pct_err = center_abs_err / abs(w_anal_center) * 100

L2_error_w = np.sqrt(np.mean((w_dim_pred - w_dim_anal) ** 2))
rel_L2_error_w = np.linalg.norm(w_dim_pred - w_dim_anal) / np.linalg.norm(w_dim_anal)

error_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "center_deflection_pred_m": float(w_pred_center),
    "center_deflection_anal_m": float(w_anal_center),
    "center_abs_error_m": float(center_abs_err),
    "center_pct_error": float(center_pct_err),
    "w_max_abs_error_m": float(np.max(w_error)),
    "w_rms_error_m": float(np.sqrt(np.mean(w_error ** 2))),
    "w_L2_error_m": float(L2_error_w),
    "w_relative_L2_error": float(rel_L2_error_w),
    "Mx_max_abs_error_Nm": float(np.max(Mx_error)),
    "Mx_rms_error_Nm": float(np.sqrt(np.mean(Mx_error ** 2))),
    "My_max_abs_error_Nm": float(np.max(My_error)),
    "My_rms_error_Nm": float(np.sqrt(np.mean(My_error ** 2))),
    "Mxy_max_abs_error_Nm": float(np.max(Mxy_error)),
    "Mxy_rms_error_Nm": float(np.sqrt(np.mean(Mxy_error ** 2))),
}
with open(os.path.join(OUT_DIR, "error_metrics.json"), "w") as f:
    json.dump(error_metrics, f, indent=2)
print("Saved error_metrics.json:", error_metrics)

df_field = pd.DataFrame({
    "x_nd": X_nd.ravel(), "y_nd": Y_nd.ravel(),
    "w_dim_anal": w_dim_anal.ravel(), "w_dim_pred": w_dim_pred.ravel(),
    "Mx_dim_anal": Mx_dim_anal.ravel(), "Mx_dim_pred": Mx_dim_pred.ravel(),
    "My_dim_anal": My_dim_anal.ravel(), "My_dim_pred": My_dim_pred.ravel(),
    "Mxy_dim_anal": Mxy_dim_anal.ravel(), "Mxy_dim_pred": Mxy_dim_pred.ravel(),
})
df_field["w_abs_error"] = np.abs(df_field["w_dim_anal"] - df_field["w_dim_pred"])
df_field["w_rel_error"] = df_field["w_abs_error"] / (np.abs(df_field["w_dim_anal"]) + 1e-12)
df_field.to_csv(os.path.join(OUT_DIR, "field_data.csv"), index=False)
print("Saved field_data.csv")

# ============================================================
# ============================================================
plt.rcParams.update({"figure.dpi": 150})


def save_contour(field, title, cbar_label, fname, cmap=cm.jet):
    plt.figure(figsize=(7, 6))
    cf = plt.contourf(X_nd, Y_nd, field, levels=50, cmap=cmap)
    plt.colorbar(cf, label=cbar_label)
    plt.title(title, fontstyle="italic", fontsize=13)
    plt.xlabel("x/a", fontstyle="italic")
    plt.ylabel("y/b", fontstyle="italic")
    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, fname))
    plt.close()


plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], df_hist["w_center_dim"], linewidth=2, label=f"{MODEL_NAME} Prediction")
plt.axhline(y=w_anal_center, linestyle="--", linewidth=2, color="k", label="Analytical")
plt.xlabel("Epoch"); plt.ylabel("Center Deflection w (m)")
plt.title("Convergence of Center Deflection")
plt.legend(); plt.grid(True, linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Center_Deflection.png")); plt.close()

plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], np.abs(df_hist["w_center_dim"] - w_anal_center), linewidth=2)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Absolute Error at Center (m)")
plt.title("Convergence Error (Log Scale)")
plt.grid(True, which="both", linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Error_Log_Scale.png")); plt.close()

plt.figure(figsize=(8, 6))
plt.plot(df_hist["epoch"], df_hist["total_loss"], label="Total Loss")
plt.plot(df_hist["epoch"], df_hist["pde_loss"], label="PDE Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["bc_loss"], label="BC Loss", alpha=0.7)
plt.plot(df_hist["epoch"], df_hist["moment_loss"], label="Moment Loss", alpha=0.7)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Training Loss Curve")
plt.legend(); plt.grid(True, which="both", ls="--", alpha=0.7); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Training_Loss_Curve.png")); plt.close()

save_contour(w_dim_anal, "Analytical Plate Deflection", "Deflection (m)",
             "Analytical_Plate_Deflection.png")
save_contour(w_dim_pred, f"{MODEL_NAME} Predicted Plate Deflection", "Deflection (m)",
             "FDM-PINN_Predicted_Plate_Deflection.png")
save_contour(w_error, "Absolute Error in Plate Deflection", "|w_pred - w_anal| (m)",
             "Absolute_Error_in_Plate_Deflection.png", cmap=cm.inferno)
save_contour(w_signed_error, "Signed Error w_pred - w_anal", "w_pred - w_anal (m)",
             "Absolute_Error_w_pred_-_w_anal.png", cmap="coolwarm")

save_contour(Mx_dim_anal, "Analytical Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x_Analytical.png")
save_contour(Mx_dim_pred, "Predicted Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x.png")
save_contour(Mx_error, "Absolute Error in $M_x$", "|Mx_pred - Mx_anal| (N·m/m)",
             "Absolute_Error_in_M_x.png", cmap="hot")

save_contour(My_dim_anal, "Analytical Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y_Analytical.png")
save_contour(My_dim_pred, "Predicted Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y.png")
save_contour(My_error, "Absolute Error in $M_y$", "|My_pred - My_anal| (N·m/m)",
             "Absolute_Error_in_M_y.png", cmap="hot")

save_contour(Mxy_dim_anal, "Analytical Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy_Analytical.png")
save_contour(Mxy_dim_pred, "Predicted Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy.png")
save_contour(Mxy_error, "Absolute Error in $M_{xy}$", "|Mxy_pred - Mxy_anal| (N·m/m)",
             "Absolute_Error_in_M_xy.png", cmap="hot")

plt.figure(figsize=(6, 6))
plt.scatter(w_dim_anal.ravel(), w_dim_pred.ravel(), s=8, alpha=0.5)
lims = [min(w_dim_anal.min(), w_dim_pred.min()), max(w_dim_anal.max(), w_dim_pred.max())]
plt.plot(lims, lims, "k--", linewidth=1, label="Perfect agreement")
plt.xlabel("Analytical displacement w (m)", fontstyle="italic")
plt.ylabel("Predicted displacement w (m)", fontstyle="italic")
plt.title("Analytical vs Predicted Displacement", fontstyle="italic")
plt.legend(); plt.grid(True); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Analytical_vs_Predicted_Displacement.png")); plt.close()

err_fft = np.fft.fftshift(np.fft.fft2(w_error))
err_fft_mag = np.log1p(np.abs(err_fft))
plt.figure(figsize=(7, 6))
plt.imshow(err_fft_mag, cmap="magma", extent=[-nx // 2, nx // 2, -ny // 2, ny // 2])
plt.colorbar(label="log(1 + |FFT error|)")
plt.title(f"Error Spectrum (FFT) — {LOADING_CASE}")
plt.xlabel("kx"); plt.ylabel("ky")
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Error_Spectrum_FFT.png")); plt.close()

fig, axes = plt.subplots(4, 3, figsize=(16, 18))
rows = [
    ("Displacement w (m)", w_dim_anal, w_dim_pred, w_error),
    ("Moment Mx (N·m)", Mx_dim_anal, Mx_dim_pred, Mx_error),
    ("Moment My (N·m)", My_dim_anal, My_dim_pred, My_error),
    ("Moment Mxy (N·m)", Mxy_dim_anal, Mxy_dim_pred, Mxy_error),
]
for r, (label, anal, pred, err) in enumerate(rows):
    for c, (title, field, cmap) in enumerate([
        (f"Analytical {label}", anal, cm.jet),
        (f"Predicted {label}", pred, cm.jet),
        (f"Absolute Error {label}", err, cm.inferno),
    ]):
        ax = axes[r, c]
        cf = ax.contourf(X_nd, Y_nd, field, levels=20, cmap=cmap)
        fig.colorbar(cf, ax=ax)
        ax.set_title(title, fontsize=10)
        ax.set_xlabel("x/a"); ax.set_ylabel("y/b")
plt.suptitle(f"{MODEL_NAME} — {LOADING_CASE} loading, SS-SS-SS-SS (summary grid)", fontsize=14)
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "_summary_grid_all_fields.png")); plt.close()

print(f"\nAll outputs written to: {os.path.abspath(OUT_DIR)}")
print(f"Total training time: {total_train_time:.2f}s | "
      f"Epochs to early stop: {early_stop_epoch} | "
      f"FD step h: {FD_STEP}")
```
