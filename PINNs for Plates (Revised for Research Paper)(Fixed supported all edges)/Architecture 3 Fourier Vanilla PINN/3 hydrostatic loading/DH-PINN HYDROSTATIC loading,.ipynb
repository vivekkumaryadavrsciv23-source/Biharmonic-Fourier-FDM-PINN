# DH-PINN plate bending under HYDROSTATIC (linearly-varying / "triangular") loading, ALL-CLAMPED / FIXED edges (FFFF case).

```python
# ---------------------------------------------------------------------------
# Physics-Informed Neural Network for Kirchhoff Plate Bending
# Architecture : Fourier (DH) Vanilla PINN
# BC Type      : FFFF (Fully Clamped: w=0 and dw/dn=0 on all edges)
# Loading      : Hydrostatic  q(x,y) = q0 * x/a
# Authors      : Vivek Kumar Yadav, Kshitij Kumar Yadav, Rajesh Kumar
# Institution  : Dept. of Civil Engineering, IIT (BHU) Varanasi
# ---------------------------------------------------------------------------


import os
import json
import time
import random

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from matplotlib import cm
from scipy.optimize import brentq
from scipy.integrate import simpson

# ============================================================
# CONFIG
# ============================================================
LOADING_CASE = "hydrostatic"
BC_CASE = "FFFF"   # Fixed (clamped) on all four edges: x=0, x=a, y=0, y=b

# --- DH-PINN specific config -------------------------------------------
N_HARMONICS = 6
INCLUDE_RAW_COORDS = True
USE_PRODUCT_TERMS = True
GRAD_CLIP_NORM = 1.0
EMBEDDING_TYPE = "deterministic_harmonic_modal" if USE_PRODUCT_TERMS else "deterministic_harmonic"

# --- Rayleigh-Ritz (clamped-clamped beam function) reference solution --
# Convergence knob for the ANALYTICAL REFERENCE ONLY (unrelated to
# N_HARMONICS, which sizes the PINN's own input embedding). Raise these
# and re-run to check the center deflection has stopped changing before
# trusting the error metrics.
RITZ_M_TERMS = 10     # number of clamped-clamped beam modes in x
RITZ_N_TERMS = 10     # number of clamped-clamped beam modes in y
RITZ_QUAD_POINTS = 4001   # quadrature resolution for the 1-D beam-function
                           # overlap integrals (odd, for Simpson's rule)

OUT_DIR = os.path.join("results", MODEL_NAME)
FIG_DIR = os.path.join(OUT_DIR, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

SEED = 42
MAX_EPOCHS = 15000
PATIENCE = 3000
PRINT_EVERY = 100
LR_INIT = 1e-3
LR_DECAY_STEPS = 2000
LR_DECAY_RATE = 0.95

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
DTYPE = torch.float64

# ============================================================
# SEEDING
# ============================================================
np.random.seed(SEED)
random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)

# ============================================================
# PHYSICAL / NON-DIMENSIONAL PARAMETERS
# ============================================================
real_a, real_b, real_h = 1.0, 1.0, 0.01
real_E, real_nu, real_q0 = 200e9, 0.3, 1000.0

D_dim = real_E * (real_h ** 3) / (12 * (1 - real_nu ** 2))
L = real_a
W = real_q0 * (L ** 4) / D_dim
D_nd = 1.0
a_nd, b_nd, q0_nd, nu_nd = 1.0, 1.0, 1.0, real_nu

# ============================================================
# DOMAIN
# ============================================================
nx, ny = 51, 51
x_nd = np.linspace(0, a_nd, nx)
y_nd = np.linspace(0, b_nd, ny)
X_nd, Y_nd = np.meshgrid(x_nd, y_nd)   # 'xy' indexing -> shape (ny, nx)
grid_points_nd = np.vstack([X_nd.ravel(), Y_nd.ravel()]).T

X_grid = torch.tensor(grid_points_nd, dtype=DTYPE, device=DEVICE)

center_x, center_y = a_nd / 2, b_nd / 2
center_idx = int(np.argmin((grid_points_nd[:, 0] - center_x) ** 2 +
                            (grid_points_nd[:, 1] - center_y) ** 2))

# --- Full boundary -- kept for the (now two-part) diagnostic bc_loss ------
boundary_indices = np.unique(np.concatenate([
    np.where(grid_points_nd[:, 1] == 0)[0],
    np.where(grid_points_nd[:, 1] == b_nd)[0],
    np.where(grid_points_nd[:, 0] == 0)[0],
    np.where(grid_points_nd[:, 0] == a_nd)[0],
]))
interior_indices = np.setdiff1d(np.arange(nx * ny), boundary_indices)

bc_idx = torch.tensor(boundary_indices, dtype=torch.long, device=DEVICE)
int_idx = torch.tensor(interior_indices, dtype=torch.long, device=DEVICE)

# per-edge indices, now used only for the SLOPE diagnostic (dw/dn ~ 0),
# since there is no soft moment BC in the clamped case.
edge_x0 = np.where(grid_points_nd[:, 0] == 0)[0]
edge_xa = np.where(grid_points_nd[:, 0] == a_nd)[0]
edge_y0 = np.where(grid_points_nd[:, 1] == 0)[0]
edge_yb = np.where(grid_points_nd[:, 1] == b_nd)[0]

edge_x0_idx = torch.tensor(edge_x0, dtype=torch.long, device=DEVICE)
edge_xa_idx = torch.tensor(edge_xa, dtype=torch.long, device=DEVICE)
edge_y0_idx = torch.tensor(edge_y0, dtype=torch.long, device=DEVICE)
edge_yb_idx = torch.tensor(edge_yb, dtype=torch.long, device=DEVICE)

# ============================================================
# HYDROSTATIC LOAD (torch, used inside the training loop) -- unchanged
# ============================================================
def hydrostatic_load_nd(x, y):
    """q(x, y) = q0 * (x / a) -- zero at x=0, ramping linearly to q0 at x=a."""
    return q0_nd * (x / a_nd)


# ============================================================
# CLAMPED-CLAMPED BEAM EIGENFUNCTIONS (building blocks for the
# Rayleigh-Ritz analytical reference)
# ============================================================
def _clamped_beam_roots(n_terms):
    """
    Positive roots beta*L of the clamped-clamped beam characteristic
    equation  cos(beta*L)*cosh(beta*L) = 1  (nondimensional, L=1 here;
    physical derivatives are rescaled by a/b afterwards).

    Roots are found numerically (scipy.optimize.brentq) rather than
    quoted from memorized literature constants, so they can be verified
    directly from the equation above. The bracket for each root is seeded
    from the standard asymptotic estimate beta_m*L ~ (2m+1)*pi/2 for
    m = 1, 2, 3, ... (this asymptote is accurate for m >= 3; the first two
    roots deviate from it more and are given a wider bracket).
    """
    def f(beta):
        return np.cos(beta) * np.cosh(beta) - 1.0

    roots = []
    for m in range(1, n_terms + 1):
        guess = (2 * m + 1) * np.pi / 2
        if m == 1:
            lo, hi = 3.5, 5.5
        elif m == 2:
            lo, hi = 6.5, 9.0
        else:
            lo, hi = guess - 1.4, guess + 1.4
        try:
            r = brentq(f, lo, hi)
        except ValueError:
            r = brentq(f, guess - 3.0, guess + 3.0)
        roots.append(r)
    return np.array(roots)


def _beam_mode_and_derivs(beta, xi):
    """
    Clamped-clamped beam eigenfunction phi(xi) on nondimensional xi in
    [0, 1] (L=1), plus its first and second derivatives w.r.t. xi. sigma
    is fixed by phi'(0)=0 automatically (the cosh-cos, sinh-sin
    combination already gives phi(0)=phi'(0)=0); sigma is chosen so that
    phi(1)=0 too, and phi'(1)=0 then follows automatically from beta
    being a root of the characteristic equation.
    """
    sigma = (np.cosh(beta) - np.cos(beta)) / (np.sinh(beta) - np.sin(beta))
    phi = (np.cosh(beta * xi) - np.cos(beta * xi)
           - sigma * (np.sinh(beta * xi) - np.sin(beta * xi)))
    dphi = beta * (np.sinh(beta * xi) + np.sin(beta * xi)
                    - sigma * (np.cosh(beta * xi) - np.cos(beta * xi)))
    ddphi = beta ** 2 * (np.cosh(beta * xi) + np.cos(beta * xi)
                          - sigma * (np.sinh(beta * xi) + np.sin(beta * xi)))
    return phi, dphi, ddphi


def _beam_overlap_matrices(betas, n_quad=RITZ_QUAD_POINTS):
    """
    1-D nondimensional (L=1) overlap integrals for a set of beam modes:
        A0[m,k] = int_0^1 phi_m * phi_k          dxi
        A1[m,k] = int_0^1 phi_m' * phi_k'        dxi
        A2[m,k] = int_0^1 phi_m'' * phi_k''      dxi
        X1[m]   = int_0^1 xi * phi_m             dxi   (for the hydrostatic
                                                          load projection)
        Y0[m]   = int_0^1 phi_m                  dxi   (uniform-in-y load
                                                          projection)
    Evaluated by Simpson's rule on a fine grid -- numerical quadrature,
    not a hand-derived closed form, to avoid transcription errors.
    """
    xi = np.linspace(0.0, 1.0, n_quad)
    M = len(betas)
    phis = np.zeros((M, n_quad))
    dphis = np.zeros((M, n_quad))
    ddphis = np.zeros((M, n_quad))
    for i, beta in enumerate(betas):
        phis[i], dphis[i], ddphis[i] = _beam_mode_and_derivs(beta, xi)

    A0 = np.array([[simpson(phis[m] * phis[k], xi) for k in range(M)] for m in range(M)])
    A1 = np.array([[simpson(dphis[m] * dphis[k], xi) for k in range(M)] for m in range(M)])
    A2 = np.array([[simpson(ddphis[m] * ddphis[k], xi) for k in range(M)] for m in range(M)])
    X1 = np.array([simpson(xi * phis[m], xi) for m in range(M)])
    Y0 = np.array([simpson(phis[m], xi) for m in range(M)])
    return A0, A1, A2, X1, Y0, xi, phis, dphis, ddphis


def _solve_rayleigh_ritz_clamped_hydrostatic():
    """
    Assemble and solve K @ C = F for the Rayleigh-Ritz coefficients C_mn
    of w(x,y) = sum_mn C_mn * phi_m(x/a) * phi_n(y/b), for the clamped
    rectangular plate under q(x,y) = q0*x/a. Returns everything needed to
    reconstruct w, Mx, My, Mxy pointwise (betas_x, betas_y, C matrix).
    """
    betas_x = _clamped_beam_roots(RITZ_M_TERMS)
    betas_y = _clamped_beam_roots(RITZ_N_TERMS)

    A0, A1, A2, X1, _, _, _, _, _ = _beam_overlap_matrices(betas_x)
    B0, B1, B2, _, Y0, _, _, _, _ = _beam_overlap_matrices(betas_y)

    M, N = RITZ_M_TERMS, RITZ_N_TERMS
    P = M * N

    # rescale nondimensional (L=1) integrals to the physical [0,a]x[0,b]
    # domain: d/dx = (1/a) d/dxi, dx = a*dxi, etc.
    A0_p = a_nd * A0
    A1_p = A1 / a_nd
    A2_p = A2 / a_nd ** 3
    B0_p = b_nd * B0
    B1_p = B1 / b_nd
    B2_p = B2 / b_nd ** 3

    K = np.zeros((P, P))
    F = np.zeros(P)

    for m in range(M):
        for n in range(N):
            p = m * N + n
            # load: F_{mn} = int int q0*(x/a) * phi_m(x) * phi_n(y) dA
            #             = (q0/a) * [a^2 * X1_m] * [b * Y0_n]
            F[p] = (q0_nd / a_nd) * (a_nd ** 2 * X1[m]) * (b_nd * Y0[n])
            for k in range(M):
                for l in range(N):
                    q = k * N + l
                    K[p, q] = D_nd * (A2_p[m, k] * B0_p[n, l]
                                       + A0_p[m, k] * B2_p[n, l]
                                       + 2.0 * A1_p[m, k] * B1_p[n, l])

    C = np.linalg.solve(K, F)
    C = C.reshape(M, N)
    return betas_x, betas_y, C


print("Solving Rayleigh-Ritz (clamped-clamped beam function) reference "
      f"solution with {RITZ_M_TERMS}x{RITZ_N_TERMS} modes...")
_RITZ_BETAS_X, _RITZ_BETAS_Y, _RITZ_C = _solve_rayleigh_ritz_clamped_hydrostatic()
print("Rayleigh-Ritz reference solution assembled.")


def _clamped_plate_fields(x, y):
    """
    Vectorized evaluation of w, Mx, My, Mxy from the Rayleigh-Ritz
    coefficients _RITZ_C, at arbitrary (x, y) arrays of identical shape
    (values in [0,a] x [0,b]). Mirrors the interface of
    `_navier_series_fields` from the SS-SS-SS-SS script.
    """
    orig_shape = x.shape
    x_flat = x.ravel() / a_nd     # nondimensional xi in [0,1]
    y_flat = y.ravel() / b_nd     # nondimensional eta in [0,1]

    M, N = RITZ_M_TERMS, RITZ_N_TERMS
    phi_x = np.zeros((M, x_flat.size))
    dphi_x = np.zeros((M, x_flat.size))
    ddphi_x = np.zeros((M, x_flat.size))
    for m in range(M):
        phi_x[m], dphi_x[m], ddphi_x[m] = _beam_mode_and_derivs(_RITZ_BETAS_X[m], x_flat)

    phi_y = np.zeros((N, y_flat.size))
    dphi_y = np.zeros((N, y_flat.size))
    ddphi_y = np.zeros((N, y_flat.size))
    for n in range(N):
        phi_y[n], dphi_y[n], ddphi_y[n] = _beam_mode_and_derivs(_RITZ_BETAS_Y[n], y_flat)

    w = np.zeros(x_flat.size)
    w_xx = np.zeros(x_flat.size)
    w_yy = np.zeros(x_flat.size)
    w_xy = np.zeros(x_flat.size)
    for m in range(M):
        for n in range(N):
            c = _RITZ_C[m, n]
            if c == 0.0:
                continue
            w += c * phi_x[m] * phi_y[n]
            w_xx += c * (ddphi_x[m] / a_nd ** 2) * phi_y[n]
            w_yy += c * phi_x[m] * (ddphi_y[n] / b_nd ** 2)
            w_xy += c * (dphi_x[m] / a_nd) * (dphi_y[n] / b_nd)

    Mx = -D_nd * (w_xx + nu_nd * w_yy)
    My = -D_nd * (w_yy + nu_nd * w_xx)
    Mxy = -D_nd * (1 - nu_nd) * w_xy

    return (w.reshape(orig_shape), Mx.reshape(orig_shape),
            My.reshape(orig_shape), Mxy.reshape(orig_shape))


def analytical_w_nd(x, y):
    w, _, _, _ = _clamped_plate_fields(x, y)
    return w


def analytical_Mx_nd(x, y):
    _, Mx, _, _ = _clamped_plate_fields(x, y)
    return Mx


def analytical_My_nd(x, y):
    _, _, My, _ = _clamped_plate_fields(x, y)
    return My


def analytical_Mxy_nd(x, y):
    _, _, _, Mxy = _clamped_plate_fields(x, y)
    return Mxy


# ============================================================
# DETERMINISTIC HARMONIC EMBEDDING -- unchanged from the SS-SS-SS-SS script
# ============================================================
class HarmonicEmbedding(nn.Module):
    """
    Deterministic sinusoidal (harmonic) input feature transformation.
    Unchanged from the SS-SS-SS-SS hydrostatic script: this is purely an
    input-feature block for the network and is independent of which
    boundary condition the hard-BC envelope (below) enforces.
    """

    def __init__(self, n_harmonics: int, include_raw_coords: bool = True,
                 use_product_terms: bool = True):
        super().__init__()
        self.n_harmonics = n_harmonics
        self.include_raw_coords = include_raw_coords
        self.use_product_terms = use_product_terms
        k = torch.arange(1, n_harmonics + 1, dtype=DTYPE, device=DEVICE).view(1, -1)
        self.register_buffer("k", k)

        out_dim = (2 if include_raw_coords else 0) + 4 * n_harmonics
        if use_product_terms:
            out_dim += n_harmonics * n_harmonics
        self.out_dim = out_dim

    def forward(self, xi, eta):
        kx = self.k * np.pi * xi
        ky = self.k * np.pi * eta
        sin_x, cos_x = torch.sin(kx), torch.cos(kx)
        sin_y, cos_y = torch.sin(ky), torch.cos(ky)

        feats = [sin_x, cos_x, sin_y, cos_y]
        if self.include_raw_coords:
            feats = [xi, eta] + feats
        if self.use_product_terms:
            modal = sin_x.unsqueeze(2) * sin_y.unsqueeze(1)
            feats.append(modal.reshape(modal.shape[0], -1))
        return torch.cat(feats, dim=1)


# ============================================================
# MODEL -- backbone unchanged; envelope updated for the CLAMPED BC
# ============================================================
class PlatePINN(nn.Module):
    """
    Same SiLU-MLP backbone and HarmonicEmbedding input block as the
    SS-SS-SS-SS script. The only structural change is the hard-BC
    envelope: squaring each factor makes both the envelope AND its first
    derivative vanish at every edge, which hard-enforces the CLAMPED
    condition w=0, dw/dn=0 on all four sides (vs. just w=0 for simple
    support). No soft moment-BC term is needed anymore -- see module
    docstring.
    """

    def __init__(self, hidden=64, n_layers=3,
                 n_harmonics=N_HARMONICS, include_raw_coords=INCLUDE_RAW_COORDS,
                 use_product_terms=USE_PRODUCT_TERMS):
        super().__init__()
        self.embed = HarmonicEmbedding(n_harmonics, include_raw_coords, use_product_terms)

        layers = []
        in_dim = self.embed.out_dim
        for _ in range(n_layers):
            layers.append(nn.Linear(in_dim, hidden, dtype=DTYPE))
            layers.append(nn.SiLU())
            in_dim = hidden
        self.hidden = nn.Sequential(*layers)
        self.out = nn.Linear(in_dim, 1, dtype=DTYPE)
        self.scale = nn.Parameter(torch.tensor(1.0, dtype=DTYPE))

    def forward(self, x, y):
        xi = x / a_nd
        eta = y / b_nd

        h = self.embed(xi, eta)
        h = self.hidden(h)
        raw = self.out(h)

        # CLAMPED hard-BC envelope: squared factors force both the
        # function value AND its first derivative to vanish at each edge
        # (xi=0, xi=1, eta=0, eta=1), for any `raw`.
        envelope = (xi * (1 - xi)) ** 2 * (eta * (1 - eta)) ** 2
        return self.scale * envelope * raw


model = PlatePINN().to(DEVICE)

# ============================================================
# OPTIMIZER + LR SCHEDULE -- unchanged
# ============================================================
optimizer = torch.optim.Adam(model.parameters(), lr=LR_INIT)
scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer, lr_lambda=lambda step: LR_DECAY_RATE ** (step / LR_DECAY_STEPS)
)


# ============================================================
# LOSS (4th-order derivatives via autograd)
# ============================================================
def compute_loss():
    x = X_grid[:, 0:1].clone().requires_grad_(True)
    y = X_grid[:, 1:2].clone().requires_grad_(True)

    w = model(x, y)
    ones = torch.ones_like(w)

    w_x = torch.autograd.grad(w, x, grad_outputs=ones, create_graph=True)[0]
    w_y = torch.autograd.grad(w, y, grad_outputs=ones, create_graph=True)[0]

    w_xx = torch.autograd.grad(w_x, x, grad_outputs=ones, create_graph=True)[0]
    w_yy = torch.autograd.grad(w_y, y, grad_outputs=ones, create_graph=True)[0]

    w_xxx = torch.autograd.grad(w_xx, x, grad_outputs=ones, create_graph=True)[0]
    w_yyy = torch.autograd.grad(w_yy, y, grad_outputs=ones, create_graph=True)[0]
    w_xxy = torch.autograd.grad(w_xx, y, grad_outputs=ones, create_graph=True)[0]

    w_xxxx = torch.autograd.grad(w_xxx, x, grad_outputs=ones, create_graph=True)[0]
    w_yyyy = torch.autograd.grad(w_yyy, y, grad_outputs=ones, create_graph=True)[0]
    w_xxyy = torch.autograd.grad(w_xxy, y, grad_outputs=ones, create_graph=True)[0]

    q = hydrostatic_load_nd(x, y)

    pde = (w_xxxx + 2 * w_xxyy + w_yyyy - q)
    pde_loss = torch.mean(pde[int_idx] ** 2)

    # DIAGNOSTICS ONLY -- both should be ~0 exactly, by construction of
    # the squared envelope. Neither enters `total`; gradients through them
    # are not used for optimization.
    bc_w_loss = torch.mean(w[bc_idx] ** 2)
    bc_slope_loss = (torch.mean(w_x[edge_x0_idx] ** 2) + torch.mean(w_x[edge_xa_idx] ** 2)
                      + torch.mean(w_y[edge_y0_idx] ** 2) + torch.mean(w_y[edge_yb_idx] ** 2))

    # No soft moment-BC term for the clamped case: both essential
    # conditions (w=0, dw/dn=0) are already hard-enforced above.
    total = pde_loss
    return total, pde_loss, bc_w_loss, bc_slope_loss, w


# ============================================================
# TRAINING LOOP with computational-cost logging + early stopping
# ============================================================
epoch_history, loss_history, pde_history = [], [], []
bc_w_history, bc_slope_history = [], []
w_center_history, epoch_time_history = [], []

best_loss = float("inf")
wait = 0
weights_path = os.path.join(OUT_DIR, "best_weights.pt")

if DEVICE.type == "cuda":
    torch.cuda.reset_peak_memory_stats()

train_start = time.perf_counter()
converged_epoch = MAX_EPOCHS

for epoch in range(MAX_EPOCHS):
    epoch_t0 = time.perf_counter()

    optimizer.zero_grad()
    total, pde, bc_w, bc_slope, w_pred = compute_loss()
    total.backward()
    if GRAD_CLIP_NORM is not None:
        torch.nn.utils.clip_grad_norm_(model.parameters(), GRAD_CLIP_NORM)
    optimizer.step()
    scheduler.step()

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    epoch_dt = time.perf_counter() - epoch_t0

    loss_val = total.item()
    w_center_val = w_pred[center_idx, 0].item()

    epoch_history.append(epoch)
    loss_history.append(loss_val)
    pde_history.append(pde.item())
    bc_w_history.append(bc_w.item())
    bc_slope_history.append(bc_slope.item())
    w_center_history.append(w_center_val)
    epoch_time_history.append(epoch_dt)

    if epoch % PRINT_EVERY == 0:
        print(f"Epoch {epoch} Loss {loss_val:.6e} (pde {pde.item():.3e}) "
              f"hardBC_w_residual {bc_w.item():.3e} hardBC_slope_residual {bc_slope.item():.3e} "
              f"Center {w_center_val:.6e} ({epoch_dt*1000:.2f} ms/epoch)")

    if loss_val < best_loss:
        best_loss = loss_val
        wait = 0
        torch.save(model.state_dict(), weights_path)
    else:
        wait += 1
        if wait >= PATIENCE:
            print(f"Early stopping triggered at epoch {epoch}")
            converged_epoch = epoch
            break

total_train_time = time.perf_counter() - train_start
peak_mem_mb = (torch.cuda.max_memory_allocated() / 1e6) if DEVICE.type == "cuda" else float("nan")

model.load_state_dict(torch.load(weights_path, map_location=DEVICE))
model.eval()

# ============================================================
# SAVE TRAINING CONVERGENCE CSV
# ============================================================
w_center_dim_history = np.array(w_center_history) * W
df_hist = pd.DataFrame({
    "epoch": epoch_history,
    "total_loss": loss_history,
    "pde_loss": pde_history,
    "hard_bc_w_residual_diagnostic": bc_w_history,
    "hard_bc_slope_residual_diagnostic": bc_slope_history,
    "w_center_nd": w_center_history,
    "w_center_dim": w_center_dim_history,
    "epoch_time_s": epoch_time_history,
})
df_hist.to_csv(os.path.join(OUT_DIR, "training_convergence.csv"), index=False)
print("Saved training_convergence.csv")

# ============================================================
# LOSS BALANCE DIAGNOSTIC
# ============================================================
final_pde = pde_history[-1]
final_bc_w, final_bc_slope = bc_w_history[-1], bc_slope_history[-1]
loss_balance = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "bc_case": BC_CASE,
    "embedding_type": EMBEDDING_TYPE,
    "n_harmonics": N_HARMONICS,
    "include_raw_coords": INCLUDE_RAW_COORDS,
    "use_product_terms": USE_PRODUCT_TERMS,
    "grad_clip_norm": GRAD_CLIP_NORM,
    "final_pde_loss": final_pde,
    "final_hard_bc_w_residual_diagnostic": final_bc_w,
    "final_hard_bc_slope_residual_diagnostic": final_bc_slope,
    "note": "total loss is pde_loss only; both BC components are hard-enforced "
            "by the squared envelope and are diagnostic-only here.",
}
if final_bc_w > 1e-20 or final_bc_slope > 1e-20:
    print(f"WARNING: hard BC residual diagnostics are not ~0 "
          f"(w: {final_bc_w:.3e}, slope: {final_bc_slope:.3e}) -- "
          f"double-check the envelope function / grid point exactness (float64 expected).")
with open(os.path.join(OUT_DIR, "loss_balance.json"), "w") as f:
    json.dump(loss_balance, f, indent=2)
print("Saved loss_balance.json:", loss_balance)

# ============================================================
# COMPUTATIONAL COST METRICS
# ============================================================
cost_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "bc_case": BC_CASE,
    "embedding_type": EMBEDDING_TYPE,
    "n_harmonics": N_HARMONICS,
    "include_raw_coords": INCLUDE_RAW_COORDS,
    "use_product_terms": USE_PRODUCT_TERMS,
    "grad_clip_norm": GRAD_CLIP_NORM,
    "input_feature_dim": model.embed.out_dim,
    "device": str(DEVICE),
    "epochs_run": len(epoch_history),
    "epochs_to_early_stop": converged_epoch,
    "epochs_to_convergence": converged_epoch,
    "mean_time_per_epoch_s": float(np.mean(epoch_time_history)),
    "median_time_per_epoch_s": float(np.median(epoch_time_history)),
    "total_training_time_s": total_train_time,
    "peak_gpu_memory_mb": peak_mem_mb,
    "n_trainable_params": sum(p.numel() for p in model.parameters() if p.requires_grad),
}
with open(os.path.join(OUT_DIR, "cost_metrics.json"), "w") as f:
    json.dump(cost_metrics, f, indent=2)
print("Saved cost_metrics.json:", cost_metrics)

# ============================================================
# POST-PROCESSING: full-field predictions (w, Mx, My, Mxy) via autograd
# ============================================================
def compute_full_fields():
    x = X_grid[:, 0:1].clone().requires_grad_(True)
    y = X_grid[:, 1:2].clone().requires_grad_(True)
    w = model(x, y)
    ones = torch.ones_like(w)

    w_x = torch.autograd.grad(w, x, grad_outputs=ones, create_graph=True)[0]
    w_y = torch.autograd.grad(w, y, grad_outputs=ones, create_graph=True)[0]
    w_xx = torch.autograd.grad(w_x, x, grad_outputs=ones, create_graph=True)[0]
    w_yy = torch.autograd.grad(w_y, y, grad_outputs=ones, create_graph=True)[0]
    w_xy = torch.autograd.grad(w_x, y, grad_outputs=ones, create_graph=True)[0]

    Mx = -(w_xx + nu_nd * w_yy)
    My = -(w_yy + nu_nd * w_xx)
    Mxy = -(1 - nu_nd) * w_xy

    w_grid = w.detach().cpu().numpy().reshape(ny, nx)
    Mx_grid = Mx.detach().cpu().numpy().reshape(ny, nx)
    My_grid = My.detach().cpu().numpy().reshape(ny, nx)
    Mxy_grid = Mxy.detach().cpu().numpy().reshape(ny, nx)
    return w_grid, Mx_grid, My_grid, Mxy_grid


w_pred_nd, Mx_pred_nd, My_pred_nd, Mxy_pred_nd = compute_full_fields()

w_anal_nd, Mx_anal_nd, My_anal_nd, Mxy_anal_nd = _clamped_plate_fields(X_nd, Y_nd)

w_dim_pred, w_dim_anal = w_pred_nd * W, w_anal_nd * W
moment_scale = D_dim * W / L ** 2
Mx_dim_pred, Mx_dim_anal = Mx_pred_nd * moment_scale, Mx_anal_nd * moment_scale
My_dim_pred, My_dim_anal = My_pred_nd * moment_scale, My_anal_nd * moment_scale
Mxy_dim_pred, Mxy_dim_anal = Mxy_pred_nd * moment_scale, Mxy_anal_nd * moment_scale

w_signed_error = w_dim_pred - w_dim_anal
w_error = np.abs(w_signed_error)
Mx_error = np.abs(Mx_dim_pred - Mx_dim_anal)
My_error = np.abs(My_dim_pred - My_dim_anal)
Mxy_error = np.abs(Mxy_dim_pred - Mxy_dim_anal)

# ============================================================
# ERROR METRICS
# ============================================================
ix, iy = nx // 2, ny // 2
w_pred_center, w_anal_center = w_dim_pred[iy, ix], w_dim_anal[iy, ix]
center_abs_err = abs(w_pred_center - w_anal_center)
center_pct_err = center_abs_err / abs(w_anal_center) * 100

L2_error_w = np.sqrt(np.mean((w_dim_pred - w_dim_anal) ** 2))
rel_L2_error_w = np.linalg.norm(w_dim_pred - w_dim_anal) / np.linalg.norm(w_dim_anal)

error_metrics = {
    "model": MODEL_NAME,
    "loading_case": LOADING_CASE,
    "bc_case": BC_CASE,
    "embedding_type": EMBEDDING_TYPE,
    "n_harmonics": N_HARMONICS,
    "use_product_terms": USE_PRODUCT_TERMS,
    "reference_solution_method": "rayleigh_ritz_clamped_beam_functions",
    "ritz_m_terms": RITZ_M_TERMS,
    "ritz_n_terms": RITZ_N_TERMS,
    "center_deflection_pred_m": float(w_pred_center),
    "center_deflection_anal_m": float(w_anal_center),
    "center_abs_error_m": float(center_abs_err),
    "center_pct_error": float(center_pct_err),
    "w_max_abs_error_m": float(np.max(w_error)),
    "w_rms_error_m": float(np.sqrt(np.mean(w_error ** 2))),
    "w_L2_error_m": float(L2_error_w),
    "w_relative_L2_error": float(rel_L2_error_w),
    "Mx_max_abs_error_Nm": float(np.max(Mx_error)),
    "Mx_rms_error_Nm": float(np.sqrt(np.mean(Mx_error ** 2))),
    "My_max_abs_error_Nm": float(np.max(My_error)),
    "My_rms_error_Nm": float(np.sqrt(np.mean(My_error ** 2))),
    "Mxy_max_abs_error_Nm": float(np.max(Mxy_error)),
    "Mxy_rms_error_Nm": float(np.sqrt(np.mean(Mxy_error ** 2))),
    "boundary_w_max_abs_value_m": float(np.max(np.abs(w_dim_pred[0, :]))
                                         + np.max(np.abs(w_dim_pred[-1, :]))
                                         + np.max(np.abs(w_dim_pred[:, 0]))
                                         + np.max(np.abs(w_dim_pred[:, -1]))),
    "_caveat": "reference solution is a convergent Rayleigh-Ritz approximation "
               "(clamped-clamped beam functions), not an exact closed form -- "
               "raise ritz_m_terms/ritz_n_terms and re-run to check convergence "
               "before trusting these numbers as ground truth.",
}
with open(os.path.join(OUT_DIR, "error_metrics.json"), "w") as f:
    json.dump(error_metrics, f, indent=2)
print("Saved error_metrics.json:", error_metrics)

df_field = pd.DataFrame({
    "x_nd": X_nd.ravel(), "y_nd": Y_nd.ravel(),
    "w_dim_anal": w_dim_anal.ravel(), "w_dim_pred": w_dim_pred.ravel(),
    "Mx_dim_anal": Mx_dim_anal.ravel(), "Mx_dim_pred": Mx_dim_pred.ravel(),
    "My_dim_anal": My_dim_anal.ravel(), "My_dim_pred": My_dim_pred.ravel(),
    "Mxy_dim_anal": Mxy_dim_anal.ravel(), "Mxy_dim_pred": Mxy_dim_pred.ravel(),
})
df_field["w_abs_error"] = np.abs(df_field["w_dim_anal"] - df_field["w_dim_pred"])
df_field["w_rel_error"] = df_field["w_abs_error"] / (np.abs(df_field["w_dim_anal"]) + 1e-12)
df_field.to_csv(os.path.join(OUT_DIR, "field_data.csv"), index=False)
print("Saved field_data.csv")

# ============================================================
# PLOTTING
# ============================================================
plt.rcParams.update({"figure.dpi": 150})


def save_contour(field, title, cbar_label, fname, cmap=cm.jet):
    plt.figure(figsize=(7, 6))
    cf = plt.contourf(X_nd, Y_nd, field, levels=50, cmap=cmap)
    plt.colorbar(cf, label=cbar_label)
    plt.title(title, fontstyle="italic", fontsize=13)
    plt.xlabel("x/a", fontstyle="italic")
    plt.ylabel("y/b", fontstyle="italic")
    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, fname))
    plt.close()


plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], df_hist["w_center_dim"], linewidth=2, label=f"{MODEL_NAME} Prediction")
plt.axhline(y=w_anal_center, linestyle="--", linewidth=2, color="k", label="Rayleigh-Ritz reference")
plt.xlabel("Epoch"); plt.ylabel("Center Deflection w (m)")
plt.title("Convergence of Center Deflection")
plt.legend(); plt.grid(True, linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Center_Deflection.png")); plt.close()

plt.figure(figsize=(8, 5))
plt.plot(df_hist["epoch"], np.abs(df_hist["w_center_dim"] - w_anal_center), linewidth=2)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Absolute Error at Center (m)")
plt.title("Convergence Error (Log Scale)")
plt.grid(True, which="both", linestyle="--", alpha=0.6); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Convergence_Error_Log_Scale.png")); plt.close()

plt.figure(figsize=(8, 6))
plt.plot(df_hist["epoch"], df_hist["total_loss"], label="Total Loss (= PDE Loss)")
plt.plot(df_hist["epoch"], df_hist["hard_bc_w_residual_diagnostic"],
         label="Hard-BC w residual (diagnostic, ~0)", alpha=0.6, linestyle=":")
plt.plot(df_hist["epoch"], df_hist["hard_bc_slope_residual_diagnostic"],
         label="Hard-BC slope residual (diagnostic, ~0)", alpha=0.6, linestyle=":")
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Training Loss Curve")
plt.legend(); plt.grid(True, which="both", ls="--", alpha=0.7); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Training_Loss_Curve.png")); plt.close()

save_contour(w_dim_anal, "Rayleigh-Ritz Reference Plate Deflection (clamped, FFFF)",
             "Deflection (m)", "Analytical_Plate_Deflection.png")
save_contour(w_dim_pred, f"{MODEL_NAME} Predicted Plate Deflection", "Deflection (m)",
             "PINN_Predicted_Plate_Deflection.png")
save_contour(w_error, "Absolute Error in Plate Deflection", "|w_pred - w_ref| (m)",
             "Absolute_Error_in_Plate_Deflection.png", cmap=cm.inferno)
save_contour(w_signed_error, "Signed Error w_pred - w_ref", "w_pred - w_ref (m)",
             "Absolute_Error_w_pred_-_w_anal.png", cmap="coolwarm")

save_contour(Mx_dim_anal, "Rayleigh-Ritz Reference Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x_Analytical.png")
save_contour(Mx_dim_pred, "Predicted Bending Moment $M_x$", "Mx (N·m/m)",
             "Bending_Moment_M_x.png")
save_contour(Mx_error, "Absolute Error in $M_x$", "|Mx_pred - Mx_ref| (N·m/m)",
             "Absolute_Error_in_M_x.png", cmap="hot")

save_contour(My_dim_anal, "Rayleigh-Ritz Reference Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y_Analytical.png")
save_contour(My_dim_pred, "Predicted Bending Moment $M_y$", "My (N·m/m)",
             "Bending_Moment_M_y.png")
save_contour(My_error, "Absolute Error in $M_y$", "|My_pred - My_ref| (N·m/m)",
             "Absolute_Error_in_M_y.png", cmap="hot")

save_contour(Mxy_dim_anal, "Rayleigh-Ritz Reference Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy_Analytical.png")
save_contour(Mxy_dim_pred, "Predicted Twisting Moment $M_{xy}$", "Mxy (N·m/m)",
             "Twisting_Moment_M_xy.png")
save_contour(Mxy_error, "Absolute Error in $M_{xy}$", "|Mxy_pred - Mxy_ref| (N·m/m)",
             "Absolute_Error_in_M_xy.png", cmap="hot")

plt.figure(figsize=(6, 6))
plt.scatter(w_dim_anal.ravel(), w_dim_pred.ravel(), s=8, alpha=0.5)
lims = [min(w_dim_anal.min(), w_dim_pred.min()), max(w_dim_anal.max(), w_dim_pred.max())]
plt.plot(lims, lims, "k--", linewidth=1, label="Perfect agreement")
plt.xlabel("Rayleigh-Ritz reference displacement w (m)", fontstyle="italic")
plt.ylabel("Predicted displacement w (m)", fontstyle="italic")
plt.title("Reference vs Predicted Displacement", fontstyle="italic")
plt.legend(); plt.grid(True); plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Analytical_vs_Predicted_Displacement.png")); plt.close()

err_fft = np.fft.fftshift(np.fft.fft2(w_error))
err_fft_mag = np.log1p(np.abs(err_fft))
plt.figure(figsize=(7, 6))
plt.imshow(err_fft_mag, cmap="magma", extent=[-nx // 2, nx // 2, -ny // 2, ny // 2])
plt.colorbar(label="log(1 + |FFT error|)")
plt.title(f"Error Spectrum (FFT) — {LOADING_CASE}, {BC_CASE}")
plt.xlabel("kx"); plt.ylabel("ky")
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "Error_Spectrum_FFT.png")); plt.close()

fig, axes = plt.subplots(4, 3, figsize=(16, 18))
rows = [
    ("Displacement w (m)", w_dim_anal, w_dim_pred, w_error),
    ("Moment Mx (N·m)", Mx_dim_anal, Mx_dim_pred, Mx_error),
    ("Moment My (N·m)", My_dim_anal, My_dim_pred, My_error),
    ("Moment Mxy (N·m)", Mxy_dim_anal, Mxy_dim_pred, Mxy_error),
]
for r, (label, anal, pred, err) in enumerate(rows):
    for c, (title, field, cmap) in enumerate([
        (f"Reference {label}", anal, cm.jet),
        (f"Predicted {label}", pred, cm.jet),
        (f"Absolute Error {label}", err, cm.inferno),
    ]):
        ax = axes[r, c]
        cf = ax.contourf(X_nd, Y_nd, field, levels=20, cmap=cmap)
        fig.colorbar(cf, ax=ax)
        ax.set_title(title, fontsize=10)
        ax.set_xlabel("x/a"); ax.set_ylabel("y/b")
plt.suptitle(f"{MODEL_NAME} — {LOADING_CASE} loading, {BC_CASE} (summary grid)", fontsize=14)
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "_summary_grid_all_fields.png")); plt.close()

print(f"\nAll outputs written to: {os.path.abspath(OUT_DIR)}")
print(f"Total training time: {total_train_time:.2f}s | "
      f"Epochs to early-stop: {converged_epoch} | "
      f"Peak GPU mem: {peak_mem_mb:.1f} MB" if DEVICE.type == "cuda" else
      f"\nAll outputs written to: {os.path.abspath(OUT_DIR)} | "
      f"Total training time: {total_train_time:.2f}s | Epochs: {converged_epoch}")
```
